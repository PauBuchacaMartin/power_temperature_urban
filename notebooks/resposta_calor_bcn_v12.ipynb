**El model, ara amb les dues branques**
$$\ln E(d) = \sum_{k=0}^{L} \chi_c(k)\,\mathrm{CDD}(d-k) \;+\; \sum_{k=0}^{L} \chi_h(k)\,\mathrm{HDD}(d-k) \;+\; \text{efectes fixos} + \varepsilon$$
$$\mathrm{CDD}(d)=\max\big(T(d)-T_c,\,0\big), \qquad \mathrm{HDD}(d)=\max\big(T_h-T(d),\,0\big)$$
Cada branca s'ajusta per separat amb els dies de la seva estació (juny-setembre i novembre-març). $\chi^0_c=\sum_k\chi_c(k)$ i $\chi^0_h=\sum_k\chi_h(k)$ són el % de consum de més per cada grau-dia.

**Què NO fa (encara)**
- No hi ha termes creuats $\chi_{ij}$ ($i\neq j$): amb tres estacions no es poden identificar.
- El bucle $G$ només és un ordre de magnitud (22j): $h_\text{eff}$ no està verificat.
- La memòria té coeficients lliures (retards 0-3); el nucli exponencial (temps de relaxació $\tau$) queda per a una versió posterior.
- **No veu el gas:** la branca de fred només mesura la part elèctrica de la calefacció.
- El test de les onades (18) suposa una $\chi$ igual tots els estius; amb la tendència, caldria fer-lo amb la $\chi$ de cada estiu (pendent).

```python
ANYS         = [2021, 2022, 2023, 2024, 2025]  # 2020 exclòs per defecte: confinament COVID
MESOS_ESTIU  = [6, 7, 8, 9]                   # juny-setembre
SECTOR       = "Residencial"                  # sector principal que analitzem (el text exacte s'ajusta a la secció 4)
LAGS         = 3                              # memòria: dies de retard (0..LAGS)
TC_CANDIDATS = np.arange(18.0, 25.5, 0.5)     # temperatures de confort a provar (°C)
PERCENTIL_ONADA = 0.95                        # llindar de Tmax per definir "dia molt calent"
DIES_MIN_ONADA  = 3                           # dies seguits per considerar-ho onada de calor
MIN_DIES_CP     = 200                         # descartem codis postals amb poques dades
CONSUM_MIN_FRAC = 0.05                        # descartem CP que algun any consumeixen menys del 5 % de la mediana (CP gairebé buits)
EFECTE_ESTACIONAL = "setmana"                # "setmana" (v0.2.1) o "mes" (com la v0.2)

# Barcelona (centre aproximat) per a la sèrie de temperatura
LAT, LON = 41.39, 2.17

# D'on llegim les dades: "api" (descàrrega automàtica) o "local" (fitxers pujats a mà)
FONT_ELEC = "local"
FONT_TEMP = "api"
FONT_EST  = "api"     # estacions Meteocat: "api" o "local" (xema.csv a CARPETA_LOCAL)
FONT_TEMP_CP = "estacions"   # temperatura per CP: "estacions" (Meteocat, v0.3) o "era5" (com la v0.2.3)
CARPETA_LOCAL = "/content/drive/MyDrive/dades_tesi"   # carpeta amb els CSV (Drive)

# v0.11 · MEMÒRIA CAU: tot el que es baixa d'internet es desa a CARPETA_LOCAL i la propera vegada es llegeix d'allà
# (així una caiguda o un límit d'un servidor no atura el notebook). Per forçar una baixada nova, esborra el fitxer.
import os
def de_la_cau(fitxer, baixa, col_data="data", fins=None, **kw_read):
    """Llegeix CARPETA_LOCAL/fitxer si existeix i cobreix fins a 'fins'; si no, crida baixa(), desa i retorna."""
    ruta = f"{CARPETA_LOCAL}/{fitxer}"
    if os.path.exists(ruta):
        d = pd.read_csv(ruta, parse_dates=[col_data], **kw_read)
        if fins is None or d[col_data].max() >= pd.Timestamp(fins) - pd.Timedelta(days=7):
            print(f"[cau] {fitxer} llegit del Drive ({len(d):,} files)")
            return d
        print(f"[cau] {fitxer} no arriba fins a {fins}: el baixo de nou")
    d = baixa()
    try:
        d.to_csv(ruta, index=False); print(f"[cau] {fitxer} baixat i desat al Drive")
    except Exception as e:
        print(f"[cau] {fitxer} baixat però no desat ({e})")
    return d

# v0.12 · Branca de fred (secció 24) i histèresi (secció 25)
HIVERNS      = [2021, 2022, 2023, 2024]   # hivern 2021 = novembre 2021 a març 2022. El 2020-21 queda fora:
                                          # només hi ha gener-març del 2021, amb toc de queda (COVID)
MESOS_HIVERN = (11, 12, 1, 2, 3)          # novembre-març
TH_CANDIDATS = np.arange(10.0, 19.5, 0.5) # llindars de calefacció a provar (°C)
```

## 3. Descàrrega del consum elèctric (Open Data BCN · Datadis)

El portal Open Data BCN funciona amb CKAN. Demanem la llista de fitxers del conjunt de dades `consum-electricitat-bcn` i baixem els CSV dels anys que volem.

**Si falla** (el portal té protecció anti-robots i de vegades bloqueja descàrregues automàtiques):
1. Obre https://opendata-ajuntament.barcelona.cat/data/ca/dataset/consum-electricitat-bcn al navegador.
2. Baixa els CSV `2021_consum_electricitat_bcn.csv` … `2025_...`.
3. A Colab, carpeta de l'esquerra → crea la carpeta `dades` i puja-hi els fitxers.
4. A la secció 2, posa `FONT_ELEC = "local"` i torna a executar.

```python
from pathlib import Path

URL_API = "https://opendata-ajuntament.barcelona.cat/data/api/3/action/package_show?id=consum-electricitat-bcn"
CAPCALERA = {"User-Agent": "Mozilla/5.0 (recerca doctoral UPC)"}

def llegeix_csv_flexible(contingut_o_ruta):
    # Els CSV de l'Ajuntament solen ser UTF-8 i separats per comes; per si de cas provem alternatives
    for enc in ["utf-8", "utf-8-sig", "latin-1"]:
        for sep in [",", ";"]:
            try:
                if isinstance(contingut_o_ruta, (str, Path)):
                    df = pd.read_csv(contingut_o_ruta, encoding=enc, sep=sep)
                else:
                    df = pd.read_csv(io.BytesIO(contingut_o_ruta), encoding=enc, sep=sep)
                if df.shape[1] >= 4:
                    return df
            except Exception:
                pass
    raise ValueError("No he pogut llegir el CSV")

def descarrega_elec_api(anys):
    r = requests.get(URL_API, headers=CAPCALERA, timeout=60)
    r.raise_for_status()
    recursos = r.json()["result"]["resources"]
    trossos = []
    for any_ in anys:
        cand = [x for x in recursos if str(any_) in (x.get("name","") + x.get("url",""))
                and x.get("url","").lower().endswith(".csv")]
        if not cand:
            print(f"  [!] No trobo el CSV de {any_}")
            continue
        url = cand[0]["url"]
        print(f"  Baixant {any_}: {url}")
        c = requests.get(url, headers=CAPCALERA, timeout=300)
        c.raise_for_status()
        trossos.append(llegeix_csv_flexible(c.content))
        time.sleep(1)
    return pd.concat(trossos, ignore_index=True)

def llegeix_elec_local(carpeta, anys):
    # Si la carpeta és al Google Drive, el muntem automàticament (demana permís la primera vegada)
    if str(carpeta).startswith("/content/drive") and not Path("/content/drive/MyDrive").exists():
        from google.colab import drive
        drive.mount("/content/drive")
    if not Path(carpeta).exists():
        raise FileNotFoundError(f"La carpeta '{carpeta}' no existeix. Comprova el camí a la secció 2.")
    print(f"Fitxers a la carpeta: {[f.name for f in Path(carpeta).iterdir()]}")
    trossos = []
    for f in sorted(p for p in Path(carpeta).iterdir() if p.suffix.lower() == ".csv"):
        if any(str(a) in f.name for a in anys) and "temp" not in f.name.lower():
            print(f"  Llegint {f.name}")
            trossos.append(llegeix_csv_flexible(f))
    if not trossos:
        raise FileNotFoundError(f"No hi ha CSV d'electricitat a '{carpeta}/'")
    return pd.concat(trossos, ignore_index=True)

if FONT_ELEC == "api":
    try:
        elec_raw = descarrega_elec_api(ANYS)
    except Exception as e:
        raise SystemExit(f"[ERROR] Descàrrega automàtica fallida ({e}).\n"
                         "Segueix les instruccions de dalt i posa FONT_ELEC = 'local'.")
else:
    elec_raw = llegeix_elec_local(CARPETA_LOCAL, ANYS)

print(f"\n[OK] {len(elec_raw):,} files · columnes: {list(elec_raw.columns)}")
elec_raw.head()
```

## 4. Neteja: identificar columnes i sumar a consum diari

No confiem en els noms exactes de les columnes: les busquem per paraules clau. **Mira la sortida d'aquesta cel·la** i comprova que ha triat bé cada columna i el sector.

```python
def troba_columna(df, paraules):
    for c in df.columns:
        nom = c.lower().replace("_", " ")
        if any(p in nom for p in paraules):
            return c
    raise KeyError(f"Cap columna conté {paraules}. Columnes: {list(df.columns)}")

COL_DATA   = troba_columna(elec_raw, ["data", "fecha", "date"])
COL_CP     = troba_columna(elec_raw, ["postal", "cp"])
COL_SECTOR = troba_columna(elec_raw, ["sector"])
COL_TRAM   = troba_columna(elec_raw, ["tram", "franja", "interval"])
COL_VALOR  = troba_columna(elec_raw, ["valor", "consum", "kwh", "value"])
print("Columnes triades:", dict(data=COL_DATA, cp=COL_CP, sector=COL_SECTOR, tram=COL_TRAM, valor=COL_VALOR))

print("\nSectors que apareixen:")
print(elec_raw[COL_SECTOR].value_counts())
print("\nTrams horaris que apareixen:")
print(elec_raw[COL_TRAM].value_counts())
```

```python
df = elec_raw.rename(columns={COL_DATA:"data", COL_CP:"cp", COL_SECTOR:"sector",
                              COL_TRAM:"tram", COL_VALOR:"kwh"})[["data","cp","sector","tram","kwh"]].copy()
df["data"] = pd.to_datetime(df["data"], errors="coerce")
# El CP pot venir com a text "08001" o com a número 8001: el normalitzem a 5 dígits
df["cp"]   = df["cp"].astype(str).str.extract(r"(\d{4,5})")[0].str.zfill(5)
df["kwh"]  = pd.to_numeric(df["kwh"].astype(str).str.replace(",", "."), errors="coerce")
df = df.dropna(subset=["data","cp","kwh"])

# Files duplicades (mateix dia, CP, sector i tram): si n'hi ha, la suma diària surt doblada
n_dup = df.duplicated(["data","cp","sector","tram"]).sum()
df = df.drop_duplicates(["data","cp","sector","tram"])
print(f"Files duplicades eliminades: {n_dup}")

# Triem el sector: busquem el text que conté SECTOR (ignorant majúscules i accents bàsics)
sectors = df["sector"].unique()
sector_triat = [s for s in sectors if SECTOR.lower()[:6] in str(s).lower()]
if not sector_triat:
    raise ValueError(f"No trobo el sector '{SECTOR}'. Opcions: {sectors}")
sector_triat = sector_triat[0]
print(f"Sector triat: '{sector_triat}'")

# Consum DIARI per codi postal: sumem els trams horaris
diari = (df[df["sector"] == sector_triat]
         .groupby(["cp","data"], as_index=False)["kwh"].sum())

# Traiem codis postals amb poques dades i valors no positius (no es pot fer el logaritme)
diari = diari[diari["kwh"] > 0]
n_dies = diari.groupby("cp")["data"].count()
cps_ok = n_dies[n_dies >= MIN_DIES_CP].index
diari = diari[diari["cp"].isin(cps_ok)]

# Codis postals gairebé buits: si en ALGUN trimestre la mediana diària del CP és menor que CONSUM_MIN_FRAC
# vegades la mediana dels CP, el marquem. Fins a la v0.10 es treien; des de la v0.11 es mantenen (vegeu més avall).
nivell_trim = diari.groupby(["cp", diari["data"].dt.to_period("Q")])["kwh"].median().unstack()
llindar_cp = CONSUM_MIN_FRAC * nivell_trim.median().median()
petits = sorted(nivell_trim.index[(nivell_trim < llindar_cp).any(axis=1)])
print(f"CP amb algun trimestre gairebé buit (< {llindar_cp:,.0f} kWh/dia): {petits if petits else 'cap'}")
# v0.11 (corregit): NO traiem cap CP. Els CP que es buiden (08007, 08009, 08011: Dreta de l'Eixample i Sant Antoni,
# ~3 % del consum residencial) són CP normals fins al juliol del 2025, quan els seus comptadors passen a altres CP
# (alerta de dades 1). Continuen tenint dades cada dia, però molt petites. El salt el detecta la secció 12 i es tracta
# amb trams; la suma de la ciutat es conserva. Només els llistem perquè quedi constància.
for cp in petits:
    q_ = nivell_trim.loc[cp]
    print(f"   {cp}: trimestres gairebé buits {list(q_.index[q_ < llindar_cp].astype(str))}")

print(f"[OK] {diari['cp'].nunique()} codis postals · {diari['data'].min().date()} → {diari['data'].max().date()}")

# Guardem també la sèrie de ciutat (suma de tots els CP) per a les figures
ciutat = diari.groupby("data", as_index=False)["kwh"].sum()

# Completesa: quantes files (CP × tram) hi ha cada dia. Si un dia en té moltes menys que la resta,
# la suma diària surt artificialment baixa.
files_dia = df[df["sector"] == sector_triat].groupby("data").size().rename("files")
print(f"Files per dia: mediana {files_dia.median():.0f}, mínim {files_dia.min()} ({files_dia.idxmin().date()})")
```

## 5. Validació de les dades (abans de modelar res)

**Regla:** cap ajust sense mirar les dades primer. Aquesta secció fa quatre controls i **treu automàticament** el que no passa, imprimint què ha tret i per què:

1. **Completesa:** dies amb menys codis postals dels habituals.
2. **Salt de nivell de la ciutat:** dies en què el total de la ciutat surt de l'interval $[0{,}5,\ 2]\times$ la seva mediana mòbil de 61 dies.
3. **Salt de nivell d'un CP:** dies en què un CP surt de $[1/3,\ 3]\times$ la seva pròpia mediana mòbil.
4. **Ordre de magnitud:** consum anual total en GWh, per comparar amb xifres conegudes.

La mediana mòbil és robusta: un bloc dolent de menys d'un mes no l'arrossega. **Un bloc dolent de més d'un mes sí que pot passar desapercebut**, i per això també mirem el gràfic i els totals anuals.

```python
UMBRAL_CIUTAT = (0.5, 2.0)    # rati total / mediana mòbil acceptat
UMBRAL_CP     = (1/3, 3.0)    # rati CP / mediana mòbil del CP acceptat
FINESTRA      = "61D"         # amplada de la mediana mòbil

# --- 1 i 2. Controls a escala de ciutat ---
tot = diari.groupby("data").agg(total=("kwh", "sum"), cps=("cp", "nunique"))
N_CP_ESPERAT = int(tot["cps"].mode()[0])          # el nombre de CP més habitual
tot["mediana"] = tot["total"].rolling(FINESTRA, center=True, min_periods=15).median()
tot["rati"] = tot["total"] / tot["mediana"]
tot["incomplet"] = tot["cps"] < N_CP_ESPERAT
tot["salt"] = (tot["rati"] < UMBRAL_CIUTAT[0]) | (tot["rati"] > UMBRAL_CIUTAT[1])
dies_dolents = tot.index[tot["incomplet"] | tot["salt"]]

# --- 3. Control per CP ---
diari = diari.sort_values(["cp", "data"]).reset_index(drop=True)
# transform: retorna sempre una columna alineada amb 'diari' (amb apply, si tots els CP tenen
# el mateix nombre de dies, pandas retorna una taula ampla i falla)
diari["mediana_cp"] = (diari.set_index("data").groupby("cp")["kwh"]
                       .transform(lambda x: x.rolling(FINESTRA, center=True, min_periods=15).median())
                       .values)
r_cp = diari["kwh"] / diari["mediana_cp"]
cp_dolent = (r_cp < UMBRAL_CP[0]) | (r_cp > UMBRAL_CP[1])

# --- Registre del que traiem ---
print(f"CP esperats per dia: {N_CP_ESPERAT}")
print(f"Dies exclosos (tota la ciutat): {len(dies_dolents)}  "
      f"[incomplets: {tot['incomplet'].sum()}, salt de nivell: {tot['salt'].sum()}]")
if len(dies_dolents):
    print(pd.Series(dies_dolents).dt.to_period("M").value_counts().sort_index().to_string())
fora_dies = cp_dolent & ~diari["data"].isin(dies_dolents)
print(f"Parells CP-dia exclosos a més (salt d'un sol CP): {fora_dies.sum()}")

registre = pd.concat([
    tot.loc[dies_dolents, ["total", "cps", "rati"]].assign(cp="TOTS",
        motiu=np.where(tot.loc[dies_dolents, "incomplet"], "incomplet", "salt_ciutat")).reset_index(),
    diari.loc[fora_dies, ["data", "cp", "kwh"]].assign(motiu="salt_cp"),
], ignore_index=True)
registre.to_csv("dies_exclosos_v12.csv", index=False)

# --- 4. Ordre de magnitud ---
anual = tot.groupby(tot.index.year)["total"].sum() / 1e6
print("\nConsum anual del sector (GWh, suma de CP, abans de filtrar):")
print(anual.round(0).to_string())

# --- Gràfic: tota la sèrie, amb els dies exclosos en vermell ---
fig, ax = plt.subplots(2, 1, figsize=(13, 5.5), sharex=True)
ax[0].plot(tot.index, tot["total"] / 1e3, lw=0.6, label="total diari")
ax[0].plot(tot.index, tot["mediana"] / 1e3, lw=1, c="k", label="mediana mòbil")
ax[0].scatter(dies_dolents, tot.loc[dies_dolents, "total"] / 1e3, c="red", s=10, zorder=3, label="exclosos")
ax[0].set_ylabel("MWh / dia"); ax[0].legend(fontsize=8)
ax[1].plot(tot.index, tot["cps"], lw=0.8); ax[1].set_ylabel("Nre. de CP")
plt.suptitle(f"Validació: consum diari {sector_triat}"); plt.tight_layout(); plt.show()

# --- Apliquem el filtre ---
diari = diari[~diari["data"].isin(dies_dolents) & ~cp_dolent].drop(columns="mediana_cp")
# La sèrie de ciutat només amb dies complets (si no, la suma surt baixa)
compl = diari.groupby("data")["cp"].nunique()
ciutat = (diari[diari["data"].isin(compl[compl == N_CP_ESPERAT].index)]
          .groupby("data", as_index=False)["kwh"].sum())
print(f"\n[OK] Després del filtre: {len(diari):,} parells CP-dia · ciutat: {len(ciutat)} dies")
```

## 6. Temperatura diària

Fem servir **ERA5** (reanàlisi europea, Copernicus) a través de l'API gratuïta d'Open-Meteo: és una graella d'uns 25 km, no una estació. És suficient per a la v0.2. **Millora pendent:** substituir-la per les estacions del Servei Meteorològic de Catalunya (més precises dins la ciutat).

Si la descàrrega falla: posa un CSV `dades/temperatura.csv` amb columnes `data, tmax, tmean` i `FONT_TEMP = "local"`.

```python
def descarrega_temperatura(inici, final):
    url = ("https://archive-api.open-meteo.com/v1/archive"
           f"?latitude={LAT}&longitude={LON}&start_date={inici}&end_date={final}"
           "&daily=temperature_2m_max,temperature_2m_mean&timezone=Europe%2FMadrid")
    r = requests.get(url, timeout=120); r.raise_for_status()
    d = r.json()["daily"]
    return pd.DataFrame({"data": pd.to_datetime(d["time"]),
                         "tmax": d["temperature_2m_max"], "tmean": d["temperature_2m_mean"]})

inici = f"{min(ANYS)}-01-01"
final = min(pd.Timestamp(f"{max(ANYS)}-12-31"), diari["data"].max()).strftime("%Y-%m-%d")

if FONT_TEMP == "api":
    temp = de_la_cau("temperatura.csv", lambda: descarrega_temperatura(inici, final), fins=final)[["data", "tmax", "tmean"]]
else:
    temp = pd.read_csv(f"{CARPETA_LOCAL}/temperatura.csv", parse_dates=["data"])

temp = temp.dropna().sort_values("data").reset_index(drop=True)
print(f"[OK] Temperatura: {len(temp)} dies · Tmean de {temp.tmean.min():.1f} a {temp.tmean.max():.1f} °C")
```

## 7. Temperatura de les estacions del Meteocat

ERA5 dona **una sola temperatura per a tota la ciutat** (una cel·la de ~9-25 km que barreja mar, Collserola i ciutat). Les estacions automàtiques del Meteocat (xarxa XEMA) mesuren dins la ciutat: **el Raval (X4)**, **Zona Universitària (X8)**, i a tocar, **Badalona (WU)**. L'Observatori Fabra (D5) és a 410 m a Collserola: el baixem només com a referència. Dades obertes diàries de la Generalitat (dataset `7bvh-jvq2`).

**Compte:** l'estació del Raval és en un terrat; mesura el centre dens, però no exactament el carrer.

```python
# Estacions automàtiques (XEMA) del Meteocat a Barcelona i rodalia: codi -> (nom, lat, lon, altitud m)
# Font de les coordenades: metadades XEMA (dataset yqwd-vj5e de la Generalitat)
ESTACIONS_INFO = {
    "X4": ("Barcelona - el Raval",           41.38386, 2.16774,  33),
    "X8": ("Barcelona - Zona Universitària", 41.37922, 2.10537,  82),
    "WU": ("Badalona - Museu",               41.45215, 2.24757,  42),
    "D5": ("Barcelona - Observatori Fabra",  41.41864, 2.12379, 410),   # Collserola: només com a referència
}
URL_XEMA = "https://analisi.transparenciacatalunya.cat/resource/7bvh-jvq2.csv"   # dades diàries XEMA
VARS_XEMA = {1000: "tmean", 1001: "tmax"}   # codis de variable: temperatura mitjana i màxima diàries

def descarrega_xema(codis, inici, final):
    llista = ",".join(f"'{c}'" for c in codis)
    params = {
        "$select": "codi_estacio,data_lectura,codi_variable,valor",
        "$where": (f"codi_estacio in({llista}) AND codi_variable in('1000','1001') "
                   f"AND data_lectura >= '{inici}T00:00:00' AND data_lectura <= '{final}T00:00:00'"),
        "$limit": 500000,
    }
    r = requests.get(URL_XEMA, params=params, timeout=300)
    r.raise_for_status()
    return pd.read_csv(io.StringIO(r.text))

if FONT_EST == "api":
    xema = de_la_cau("xema.csv", lambda: descarrega_xema(list(ESTACIONS_INFO), inici, final), col_data="data_lectura", fins=final)
else:
    xema = pd.read_csv(f"{CARPETA_LOCAL}/xema.csv")   # mateix format que la descàrrega

# De format llarg (una fila per estació, dia i variable) a una fila per estació i dia
xema["data"] = pd.to_datetime(xema["data_lectura"]).dt.normalize()
xema["var"] = xema["codi_variable"].astype(int).map(VARS_XEMA)
xema["valor"] = pd.to_numeric(xema["valor"], errors="coerce")
est = (xema.pivot_table(index=["data", "codi_estacio"], columns="var", values="valor")
           .reset_index().rename(columns={"codi_estacio": "estacio"}))
print(f"[OK] Meteocat: {len(est):,} dies-estació · {est['estacio'].nunique()} estacions")

# ---------- Comparació amb ERA5 ----------
p95 = temp.loc[temp["data"].dt.month.isin(MESOS_ESTIU), "tmax"].quantile(PERCENTIL_ONADA)
cmp_ = est.merge(temp[["data", "tmean", "tmax"]], on="data", suffixes=("", "_era5"))
cmp_ = cmp_[cmp_["data"].dt.month.isin(MESOS_ESTIU)]
cmp_["calent"] = cmp_["tmax_era5"] >= p95
files_cmp = []
for e, g in cmp_.groupby("estacio"):
    files_cmp.append({
        "estacio": e, "nom": ESTACIONS_INFO[e][0], "dies_estiu": len(g),
        "ΔTmitjana_estiu": (g["tmean"] - g["tmean_era5"]).mean(),
        "ΔTmitjana_dies_calents": (g.loc[g["calent"], "tmean"] - g.loc[g["calent"], "tmean_era5"]).mean(),
        "ΔTmax_dies_calents": (g.loc[g["calent"], "tmax"] - g.loc[g["calent"], "tmax_era5"]).mean(),
        "correlacio": g[["tmean", "tmean_era5"]].corr().iloc[0, 1],
    })
print("\nEstiu: estació menys ERA5 (°C). Positiu = l'estació és més calenta que ERA5")
print(pd.DataFrame(files_cmp).round(2).to_string(index=False))

# Gràfic: un estiu, totes les estacions i ERA5
ANY_GRAFIC = 2023
fig, ax = plt.subplots(figsize=(13, 4))
m = lambda d: (d["data"].dt.year == ANY_GRAFIC) & d["data"].dt.month.isin(MESOS_ESTIU)
ax.plot(temp.loc[m(temp), "data"], temp.loc[m(temp), "tmean"], c="k", lw=2, label="ERA5 (una cel·la)")
for e, g in est[m(est)].groupby("estacio"):
    ax.plot(g["data"], g["tmean"], lw=1, label=f"{e} · {ESTACIONS_INFO[e][0]}")
ax.set_ylabel("Temperatura mitjana diària (°C)"); ax.legend(fontsize=8)
ax.set_title(f"Estiu {ANY_GRAFIC}: estacions del Meteocat vs ERA5")
plt.tight_layout(); plt.show()
```

## 8. Geografia dels codis postals

OSM no té els límits dels CP: els **reconstruïm a partir de les adreces** (Voronoi). En traiem el polígon, el centre i l'àrea de cada CP. Pot trigar uns minuts.

```python
import geopandas as gpd
import osmnx as ox
import contextily as cx
from shapely.geometry import MultiPoint
from shapely.ops import voronoi_diagram  # cal shapely >= 2.0 (Colab ja la té)

ox.settings.requests_timeout = 600   # la consulta és gran: deixem-li temps
# (pot trigar uns minuts la primera vegada)

import os, time
cps = set(diari["cp"])
LLOC = "Barcelona, Catalonia, Spain"

# v0.11: MEMÒRIA CAU al Drive. El servidor d'OSM (Overpass) sovint està saturat. Si ja tenim els polígons
# desats a CARPETA_LOCAL, els fem servir i no cal internet. Si no, els baixem d'OSM (provant dos servidors)
# i els desem per a la propera vegada.
F_POL = f"{CARPETA_LOCAL}/codis_postals.geojson"     # polígons dels CP
F_CEN = f"{CARPETA_LOCAL}/centres_cp.csv"            # centre (lat, lon) i àrea de cada CP
F_ADR = f"{CARPETA_LOCAL}/adreces_cp.csv"            # adreces amb CP (per a les seccions A2 i A3)
SERVIDORS_OSM = ["https://overpass-api.de/api", "https://overpass.kumi.systems/api"]

def unio(gdf):
    return gdf.union_all() if hasattr(gdf, "union_all") else gdf.unary_union

adr = None
if os.path.exists(F_POL) and os.path.exists(F_CEN):
    pol = gpd.read_file(F_POL).to_crs(epsg=25831)
    pol["cp"] = pol["cp"].astype(str).str.zfill(5)
    cen = pd.read_csv(F_CEN, dtype={"cp": str}).set_index("cp")
    centroides = cen[["lat", "lon"]]
    area_cp = pol.set_index("cp").geometry.area
    bcn = gpd.GeoDataFrame(geometry=[unio(pol)], crs=pol.crs)
    limit = bcn.geometry.iloc[0]
    if os.path.exists(F_ADR):
        a_ = pd.read_csv(F_ADR, dtype={"cp": str})
        adr = gpd.GeoDataFrame(a_[["cp"]], geometry=gpd.points_from_xy(a_["lon"], a_["lat"]), crs="EPSG:4326").to_crs(epsg=25831)
    print(f"[OK] Polígons llegits de la memòria cau ({len(pol)} CP): {F_POL}")
    falten = sorted(cps - set(pol["cp"]))
    if falten:
        print(f"  [!] CP sense polígon a la memòria cau (no sortiran als mapes; temperatura ERA5): {falten}")
        print("      Per tenir-los, esborra els fitxers de la memòria cau i torna a executar quan OSM funcioni.")
else:
    for servidor in SERVIDORS_OSM:
        try:
            try:
                ox.settings.overpass_url = servidor
            except Exception:
                ox.settings.overpass_endpoint = servidor
            print(f"Provo el servidor d'OSM {servidor} ...")
            # ---------- 1) Límit del municipi i adreces amb codi postal ----------
            LLOC = "Barcelona, Catalonia, Spain"
            bcn = ox.geocode_to_gdf(LLOC).to_crs(epsg=25831)      # 25831: UTM 31N, en metres
            limit = bcn.geometry.iloc[0]

            adr = ox.features_from_place(LLOC, tags={"addr:postcode": True})
            adr = adr[["addr:postcode", "geometry"]].copy()
            adr["cp"] = adr["addr:postcode"].astype(str).str.extract(r"(\d{5})")[0]
            adr = adr[adr["cp"].isin(cps)].to_crs(epsg=25831)
            adr["geometry"] = adr.geometry.representative_point()  # edificis -> un punt
            print(f"Adreces amb codi postal: {len(adr):,}")

            # Traiem adreces mal etiquetades: les que són a més de 2,5 km del centre del seu CP
            cx_ = adr.geometry.x.groupby(adr["cp"]).transform("median")
            cy_ = adr.geometry.y.groupby(adr["cp"]).transform("median")
            adr = adr[np.hypot(adr.geometry.x - cx_, adr.geometry.y - cy_) < 2500]

            # Com a molt 400 punts per CP (més no millora la forma i alenteix el càlcul)
            adr = adr.groupby("cp", group_keys=False).apply(lambda g: g.sample(min(len(g), 400), random_state=0))
            recompte = adr["cp"].value_counts()
            print(f"CP amb adreces: {len(recompte)} de {len(cps)} · mínim {recompte.min()} adreces ({recompte.idxmin()})")
            falten = sorted(cps - set(recompte.index))
            if falten:
                print("  [!] CP sense adreces a OSM (no sortiran al mapa):", falten)

            # ---------- 2) Voronoi i 3) unió per CP ----------
            adr = adr.drop_duplicates(subset="geometry")
            vor = voronoi_diagram(MultiPoint(list(adr.geometry)), envelope=limit.envelope.buffer(2000))
            cel = gpd.GeoDataFrame(geometry=list(vor.geoms), crs=adr.crs)
            cel = gpd.sjoin(cel, adr[["cp", "geometry"]], predicate="contains", how="inner")
            pol = cel[["cp", "geometry"]].dissolve(by="cp").reset_index()
            pol["geometry"] = pol.geometry.intersection(limit)   # retallem pel límit municipal
            print(f"[OK] Polígons aproximats: {len(pol)} CP")

            # Centre de cada CP (mediana de les seves adreces) i àrea (m²)
            _c = adr.to_crs(epsg=4326)
            centroides = (pd.DataFrame({"cp": _c["cp"].values, "lat": _c.geometry.y.values, "lon": _c.geometry.x.values})
                            .groupby("cp").median())
            area_cp = pol.set_index("cp").geometry.area     # EPSG:25831 -> m²
            break
        except Exception as e:
            print(f"  [!] Ha fallat ({str(e)[:120]}). Espero 60 s i provo el següent.")
            time.sleep(60)
    else:
        raise RuntimeError("Cap servidor d'OSM respon. Torna-ho a provar més tard o puja a CARPETA_LOCAL "
                           "codis_postals.geojson i centres_cp.csv d'una execució anterior.")
    # Desem la memòria cau per a la propera vegada
    pol.to_crs(epsg=4326).to_file(F_POL, driver="GeoJSON")
    centroides.join(area_cp.rename("area_m2")).to_csv(F_CEN)
    _a = adr.to_crs(epsg=4326)
    pd.DataFrame({"cp": _a["cp"].values, "lat": _a.geometry.y.values, "lon": _a.geometry.x.values}).to_csv(F_ADR, index=False)
    print(f"[OK] Memòria cau desada a {CARPETA_LOCAL}")
```

## 9. Temperatura local de cada codi postal

Per a cada CP, fem la mitjana de les estacions ponderada per $1/d^2$ (distància des del centre del CP). Si un dia no hi ha cap estació, fem servir ERA5 més la diferència mitjana. Amb tres estacions, el camp de temperatura és suau: capta el gradient centre-perifèria, no el carrer.

```python
# Temperatura de cada CP: mitjana de les estacions ponderada per la inversa de la distància al quadrat (IDW)
# des del centre del CP (mediana de les seves adreces). Fabra (D5) queda fora: és a 410 m, a Collserola.
EST_CIUTAT = ["X4", "X8", "WU"]
POTENCIA_IDW = 2

def dist_km(lat1, lon1, lat2, lon2):
    # distància plana (prou bona a escala de ciutat)
    kx = 111.32 * np.cos(np.radians((lat1 + lat2) / 2))
    return np.hypot((lon1 - lon2) * kx, (lat1 - lat2) * 110.57)

dies = temp[["data"]].copy()
taules = {}
for var in ["tmean", "tmax"]:
    taules[var] = (est.pivot(index="data", columns="estacio", values=var)
                      .reindex(columns=EST_CIUTAT).reindex(dies["data"]))

def serie_idw(lat, lon):
    d = np.array([dist_km(lat, lon, ESTACIONS_INFO[e][1], ESTACIONS_INFO[e][2]) for e in EST_CIUTAT])
    w = 1.0 / np.maximum(d, 0.3) ** POTENCIA_IDW
    out = pd.DataFrame({"data": dies["data"].values})
    for var, tw in taules.items():
        W = pd.DataFrame(np.tile(w, (len(tw), 1)), index=tw.index, columns=EST_CIUTAT).where(tw.notna())
        v = ((tw * W).sum(axis=1) / W.sum(axis=1)).values          # NaN si cap estació té dada
        era5 = temp[var].values
        offset = np.nanmean(v - era5)                               # diferència mitjana amb ERA5
        out[var] = np.where(np.isnan(v), era5 + offset, v)          # forats: ERA5 + diferència mitjana
    return out, w / w.sum()

tloc, pesos = {}, {}
for cp, c in centroides.iterrows():
    tloc[cp], pesos[cp] = serie_idw(c["lat"], c["lon"])
pesos = pd.DataFrame(pesos, index=EST_CIUTAT).T.round(2)

# Temperatura "de ciutat" segons les estacions: mitjana simple de les tres
temp_est_ciutat, _ = serie_idw(*centroides[["lat", "lon"]].mean().values)

estiu = temp["data"].dt.month.isin(MESOS_ESTIU).values
resum_t = pd.DataFrame({cp: {"Tmitjana_estiu": t.loc[estiu, "tmean"].mean(),
                             "estacio_principal": pesos.loc[cp].idxmax(),
                             "pes_principal": pesos.loc[cp].max()} for cp, t in tloc.items()}).T
print(f"[OK] Temperatura local per a {len(tloc)} CP")
print("Estació que domina cada CP:", resum_t["estacio_principal"].value_counts().to_dict())
print(f"Tmitjana d'estiu per CP: de {resum_t['Tmitjana_estiu'].min():.2f} a {resum_t['Tmitjana_estiu'].max():.2f} °C "
      f"(ERA5: {temp.loc[estiu, 'tmean'].mean():.2f} °C)")
```

## 10. Illa de calor per satèl·lit (Landsat 8 i 9)

**Què fa:** baixa les imatges tèrmiques dels estius 2021-2025 sense gaires núvols (catàleg obert Microsoft Planetary Computer, sense compte), treu núvols, ombres i mar, i calcula la temperatura de superfície mitjana de cada CP a cada imatge.

**Per què:** les tres estacions donen un camp de temperatura molt suau. El satèl·lit veu la ciutat a ~100 m i ens diu **quins barris són més calents** i **quant s'amplifica la calor** a cada barri.

**Dos números per CP:**
- $\Delta LST_i$: quants graus més calenta és la superfície del CP respecte de la mitjana de la ciutat (illa de calor estàtica).
- $s_i$ (amplificació): pendent de $LST_i$ respecte de $LST_{\text{ciutat}}$ entre imatges. $s_i = 1{,}2$ vol dir que, quan la ciutat s'escalfa 1 K, aquell CP s'escalfa 1,2 K. **És el que pot inflar $\chi^0$**, perquè $\chi^0_{\text{mesurat}} \approx \chi^0_{\text{real}}\cdot s_i$.

**Què mirar a la sortida:** el nombre d'imatges útils (com més, millor; amb menys de ~15 cal desconfiar), la taula ordenada per ΔLST i el mapa. Esperem el centre dens i els polígons industrials calents, i Collserola, Montjuïc i la façana marítima frescos.

**Limitacions:** és temperatura de **superfície** (terrats, asfalt), no de l'aire, i a l'hora de pas del satèl·lit (~10:30 h). La superfície oscil·la més que l'aire: $s_i$ exagera les diferències. Pot trigar uns minuts.

```python
# v0.11: MEMÒRIA CAU al Drive. El catàleg de Planetary Computer limita el nombre de consultes ("rate limit").
# Si ja tenim la illa de calor desada a CARPETA_LOCAL, la llegim; si no, la calculem i la desem.
import os
F_ILLA = f"{CARPETA_LOCAL}/illa_calor.csv"
if os.path.exists(F_ILLA):
    illa = pd.read_csv(F_ILLA, dtype={"cp": str}).set_index("cp")
    print(f"[OK] Illa de calor llegida de la memòria cau ({len(illa)} CP): {F_ILLA}")
    print(illa.sort_values("dLST", ascending=False).round(2).head(10).to_string())
else:
    # Imatges Landsat 8 i 9 (Collection 2, nivell 2) del catàleg obert de Microsoft Planetary Computer.
    # La banda tèrmica 'lwir11' és la temperatura de superfície (en Kelvin, després d'aplicar escala i offset).
    import pystac_client, planetary_computer, rioxarray
    from rasterio.features import rasterize

    MAX_NUVOLS   = 20      # % màxim de núvols de l'escena sencera
    MIN_VALID    = 0.7     # fracció mínima de píxels vàlids dins la ciutat per acceptar una imatge
    ESCALA, OFFSET = 0.00341802, 149.0   # conversió de la banda tèrmica a Kelvin (Landsat C2 L2)

    cataleg = pystac_client.Client.open("https://planetarycomputer.microsoft.com/api/stac/v1",
                                        modifier=planetary_computer.sign_inplace)
    bbox = list(bcn.to_crs(epsg=4326).total_bounds)
    cerca = cataleg.search(collections=["landsat-c2-l2"], bbox=bbox,
                           datetime=f"{min(ANYS)}-06-01/{max(ANYS)}-09-30",
                           query={"eo:cloud_cover": {"lt": MAX_NUVOLS},
                                  "platform": {"in": ["landsat-8", "landsat-9"]}})
    items = [it for it in cerca.items() if it.datetime.month in MESOS_ESTIU]
    print(f"Imatges d'estiu candidates: {len(items)}")

    def llegeix(item, ref=None):
        # Retalla la banda tèrmica i la de qualitat al rectangle de Barcelona; si hi ha 'ref', s'alinea a la seva graella
        st = rioxarray.open_rasterio(item.assets["lwir11"].href).squeeze(drop=True)
        qa = rioxarray.open_rasterio(item.assets["qa_pixel"].href).squeeze(drop=True)
        caixa = bcn.to_crs(st.rio.crs).total_bounds
        st, qa = st.rio.clip_box(*caixa), qa.rio.clip_box(*caixa)
        if ref is not None:
            st, qa = st.rio.reproject_match(ref), qa.rio.reproject_match(ref)
        q = qa.values.astype(np.int64)
        # Bits de qualitat: 1 núvol dilatat, 3 núvol, 4 ombra, 7 aigua -> fora
        dolent = ((q >> 1) & 1) | ((q >> 3) & 1) | ((q >> 4) & 1) | ((q >> 7) & 1)
        lst = st.values * ESCALA + OFFSET - 273.15                       # °C
        lst = np.where((dolent == 0) & (st.values > 0), lst, np.nan)
        return st, lst

    def mitjana_per_zona(valors, etiquetes, n):
        # Mitjana dels píxels vàlids de cada zona (etiqueta 1..n); 0 = fora de la ciutat
        ok = np.isfinite(valors) & (etiquetes > 0)
        suma = np.bincount(etiquetes[ok], weights=valors[ok], minlength=n + 1)
        cnt  = np.bincount(etiquetes[ok], minlength=n + 1)
        with np.errstate(invalid="ignore", divide="ignore"):
            return suma / cnt, cnt

    ref, etiquetes, files_lst = None, None, []
    codis = list(pol["cp"])
    for it in sorted(items, key=lambda x: x.datetime):
        try:
            st, lst = llegeix(it, ref)
        except Exception as e:
            print(f"  [!] {it.id}: {e}"); continue
        if ref is None:   # la primera imatge vàlida defineix la graella i els polígons rasteritzats
            ref = st
            etiquetes = rasterize([(g, k + 1) for k, g in enumerate(pol.to_crs(st.rio.crs).geometry)],
                                  out_shape=st.shape, transform=st.rio.transform(), fill=0, dtype="int32")
            n_ciutat = (etiquetes > 0).sum()
        mitj, cnt = mitjana_per_zona(lst, etiquetes, len(codis))
        valid = cnt[1:].sum() / n_ciutat
        if valid < MIN_VALID:
            continue
        fila = {"data": pd.Timestamp(it.datetime.date()), "valid": valid, "lst_ciutat": np.nanmean(lst[etiquetes > 0])}
        fila.update({cp: mitj[k + 1] for k, cp in enumerate(codis)})
        files_lst.append(fila)

    lst_cp = pd.DataFrame(files_lst).set_index("data")
    print(f"[OK] Imatges útils: {len(lst_cp)} (de {lst_cp.index.min().date()} a {lst_cp.index.max().date()})")

    # ---------- Dos números per CP ----------
    # ΔLST: quants graus més calenta (o freda) és la superfície del CP respecte de la mitjana de la ciutat
    # s (amplificació): si la ciutat s'escalfa 1 K més un dia que un altre, quant s'escalfa el CP (pendent)
    anom = lst_cp[codis].sub(lst_cp["lst_ciutat"], axis=0)
    illa = pd.DataFrame({"dLST": anom.mean(), "dLST_se": anom.std() / np.sqrt(anom.count())})
    x = lst_cp["lst_ciutat"]
    illa["s_amplificacio"] = [np.polyfit(x[lst_cp[cp].notna()], lst_cp[cp].dropna(), 1)[0]
                              if lst_cp[cp].notna().sum() > 5 else np.nan for cp in codis]
    illa.index.name = "cp"
    print("\nIlla de calor de superfície per CP (mitjana d'estiu, ~10:30 h):")
    print(illa.sort_values("dLST", ascending=False).round(2).to_string())

    # Mapa de ΔLST
    g_illa = pol.merge(illa.reset_index(), on="cp").to_crs(epsg=3857)
    fig, ax = plt.subplots(figsize=(10, 10))
    lim = np.nanmax(np.abs(g_illa["dLST"]))
    g_illa.plot(ax=ax, column="dLST", cmap="RdBu_r", vmin=-lim, vmax=lim, edgecolor="white", linewidth=0.6,
                legend=True, legend_kwds={"label": "ΔLST respecte de la mitjana de la ciutat (°C)", "shrink": 0.55})
    for _, r in g_illa.iterrows():
        p = r.geometry.representative_point(); ax.annotate(r["cp"][-2:], (p.x, p.y), ha="center", fontsize=8)
    ax.set_axis_off(); ax.set_title(f"Illa de calor de superfície (Landsat, {len(lst_cp)} imatges d'estiu)")
    plt.tight_layout(); plt.savefig("illa_calor_landsat_v12.png", dpi=200); plt.show()
    illa.to_csv(F_ILLA)
    print(f"[OK] Illa de calor desada a la memòria cau: {F_ILLA}")
```

## 11. Valors atípics condicionats a la temperatura

La secció 5 només treu errors grossos (factor 2). Per detectar errors més petits no n'hi ha prou de comparar amb una mediana mòbil: **un dia d'onada de calor consumeix un 30-50 % més que la mitjana del mes, i un filtre ingenu l'esborraria**. Seria eliminar justament el senyal que estudiem.

La solució és definir "atípic" **respecte d'un model simple que ja coneix la temperatura**:
$$\ln E_{\text{ciutat}}(d) = f\big(\lfloor T(d)\rceil\big) + \text{dia setmana} + \text{festiu} + \text{setmana de l\'any} + \text{any} + r(d)$$
amb $f$ un valor lliure per a cada grau de temperatura (una "V" sense forma imposada). Un dia és atípic si el seu residu supera $K$ vegades la dispersió robusta $\hat\sigma = 1{,}4826\cdot\mathrm{MAD}(r)$. Fem servir la MAD i no la desviació típica perquè els mateixos atípics no inflin el llindar.

**Els dies calents (Tmax per sobre del P95 d'estiu) no s'exclouen mai automàticament**: si surten marcats, s'imprimeixen amb l'avís "revisa" i decidim a mà. És la regla per no esborrar el senyal.

```python
K_ATIPIC   = 6.0    # quantes sigmes robustes considerem atípic
SIGMA_MIN  = 0.02   # dispersió mínima (2 %): evita que un model massa ajustat marqui dies normals

v = ciutat.merge(temp, on="data")
fe = pd.get_dummies(pd.DataFrame({
    "tbin":   v["tmean"].round().astype(int).astype(str),
    "dia":    v["data"].dt.dayofweek.astype(str),
    "festiu": v["data"].isin(pd.to_datetime(list(holidays.country_holidays("ES", subdiv="CT", years=ANYS).keys()))).astype(str),
    "setmana": v["data"].dt.isocalendar().week.astype(int).astype(str).values,  # capta el buidatge de mitjan agost
    "any":    v["data"].dt.year.astype(str),
}), drop_first=True).astype(float)
X = np.column_stack([np.ones(len(v)), fe.values])
y = np.log(v["kwh"].values)
beta, *_ = np.linalg.lstsq(X, y, rcond=None)
v["residu"] = y - X @ beta
sigma = max(1.4826 * np.median(np.abs(v["residu"] - np.median(v["residu"]))), SIGMA_MIN)
v["z"] = v["residu"] / sigma
atipics = v[v["z"].abs() > K_ATIPIC].copy()

print(f"Dispersió robusta del model simple: σ = {100*sigma:.1f} %")
print(f"Dies atípics (|z| > {K_ATIPIC}): {len(atipics)}")
if len(atipics):
    # "calent": Tmax per sobre del percentil 95 dels estius (mateix criteri que les onades de la secció 14)
    p95 = temp.loc[temp["data"].dt.month.isin(MESOS_ESTIU), "tmax"].quantile(PERCENTIL_ONADA)
    # v0.12: tampoc s'exclouen els dies MOLT FREDS (Tmean per sota del percentil 5 dels hiverns):
    # són el senyal de la branca de fred (secció 24). Només es marquen per revisar-los.
    p05 = temp.loc[temp["data"].dt.month.isin([11, 12, 1, 2, 3]), "tmean"].quantile(0.05)
    atipics["calent"] = np.where(atipics["tmax"] >= p95, "SÍ: revisa!",
                                 np.where(atipics["tmean"] <= p05, "FRED: revisa!", ""))
    print(atipics.assign(data=atipics["data"].dt.date, kwh=atipics["kwh"].round(0),
                         residu_pct=(100*atipics["residu"]).round(1), z=atipics["z"].round(1))
          [["data","tmean","kwh","residu_pct","z","calent"]].to_string(index=False))

fig, ax = plt.subplots(figsize=(13,3.5))
ax.plot(v["data"], 100*v["residu"], lw=0.5)
ax.axhline( K_ATIPIC*100*sigma, c="red", ls="--", lw=0.8); ax.axhline(-K_ATIPIC*100*sigma, c="red", ls="--", lw=0.8)
ax.scatter(atipics["data"], 100*atipics["residu"], c="red", s=12, zorder=3)
ax.set_ylabel("Residu (%)"); ax.set_title("Residus del model simple (vermell: atípics exclosos)")
plt.tight_layout(); plt.show()

# Excloem els dies atípics de tota l'anàlisi, EXCEPTE els calents (només es marquen per revisar)
if len(atipics):
    atipics = atipics[atipics["calent"] == ""]
dies_atipics = set(atipics["data"])
print(f"Dies exclosos per atípics (no calents): {len(dies_atipics)}")
diari  = diari[~diari["data"].isin(dies_atipics)]
ciutat = ciutat[~ciutat["data"].isin(dies_atipics)]
registre = pd.concat([registre, atipics[["data","kwh"]].assign(cp="TOTS", motiu="atipic_temperatura")], ignore_index=True)
registre.to_csv("dies_exclosos_v12.csv", index=False)
print(f"\n[OK] Queden {len(ciutat)} dies de ciutat. Registre complet: dies_exclosos_v12.csv")
```

## 12. Salts de nivell persistents: comptadors que canvien de codi postal

**Per què:** la 5 treu anomalies de pocs dies, però no veu els **esglaons que es mantenen**. Al centre de l'Eixample n'hi ha de molt clars: 08037 es multiplica per ~6 a l'estiu del 2025, 08008 cau a un terç a l'estiu del 2023 i torna a pujar el 2025, i 08010 baixa a meitat el 2025. Com que el consum total no canvia, el més probable és que Datadis **reassigni comptadors entre CP veïns**. Els efectes fixos per any i setmana no absorbeixen un salt a mig estiu, i això infla els errors i fa que la $\chi^0$ d'aquests CP no sigui fiable.

**Què fa:**
1. Per a cada CP calcula la seva **quota** sobre la resta de la ciutat (així el cicle anual gairebé desapareix).
2. Compara la mediana dels 60 dies abans i dels 60 dies després de cada data. Si el factor és de ×1,5 o més (o ÷1,5), és un salt.
3. Si `CORREGIR_SALTS = True`, **reescala tot el tram anterior** al salt perquè empalmi amb el nivell nou.

**Què mirar a la sortida:** la taula de salts (CP, data, factor). Si dos CP veïns tenen salts el mateix dia amb factors oposats, és una reassignació. Compara després la $\chi^0$ de l'Eixample central amb `CORREGIR_SALTS = True` i `False`.

**Compte:** el reescalat corregeix el **nivell**, no la **composició**. Si els comptadors que entren són d'un altre tipus (per exemple, pisos turístics), la $\chi^0$ del CP pot canviar de veritat abans i després del salt. Provat amb dades sintètiques: troba les dates amb un error d'una setmana i els factors amb un error d'un ~5 %.


```python
# ===== 12. Salts de nivell persistents (comptadors que canvien de codi postal) =====
# La 5 només detecta anomalies de pocs dies. Aquí busquem ESGLAONS que es mantenen:
# el nivell de després és molt diferent del d'abans i no torna. Hi corregim escalant
# tot el tram anterior al salt perquè empalmi amb el nivell nou.
CORREGIR_SALTS = True   # posa False per veure els resultats sense corregir
FINESTRA = 60           # dies que comparem abans i després de cada possible salt
LLINDAR  = 1.5          # factor mínim per considerar-ho un salt (×1,5 o ÷1,5)

TOTAL = diari.groupby("data")["kwh"].sum()   # consum de tota la ciutat, per treure l'estacionalitat

def troba_salts(s):
    # s: consum diari d'un CP (índex = data). Treballem amb la QUOTA del CP sobre la ciutat:
    # el cicle de l'any (estiu, agost, Nadal) gairebé desapareix i només queden els esglaons.
    # Retorna llista de (data del salt, factor despres/abans)
    l = np.log(s / (TOTAL.reindex(s.index) - s)).asfreq("D")   # resta de la ciutat, sense el propi CP
    abans   = l.rolling(FINESTRA, min_periods=40).median()                       # mediana dels 60 dies fins avui
    despres = l[::-1].rolling(FINESTRA, min_periods=40).median()[::-1].shift(-1)  # mediana dels 60 dies següents
    salt = (despres - abans).dropna()
    salts = []
    while len(salt) and salt.abs().max() > np.log(LLINDAR):
        d = salt.abs().idxmax()
        salts.append((d, float(np.exp(salt[d]))))
        salt = salt[(salt.index < d - pd.Timedelta(days=FINESTRA)) | (salt.index > d + pd.Timedelta(days=FINESTRA))]
    return sorted(salts)

llista_salts = []
for cp, g in diari.groupby("cp"):
    s = g.groupby("data")["kwh"].sum().sort_index()
    for d, f in troba_salts(s):
        llista_salts.append({"cp": cp, "data_salt": d.date(), "factor": round(f, 2)})
salts_df = pd.DataFrame(llista_salts)
print(f"Salts de nivell persistents (factor ≥ {LLINDAR}): {len(salts_df)} en {salts_df['cp'].nunique() if len(salts_df) else 0} CP")
print(salts_df.to_string(index=False) if len(salts_df) else "  cap")

diari_brut = diari.copy()   # còpia sense corregir: la secció 19 agrupa CP a partir d'aquí
if CORREGIR_SALTS and len(salts_df):
    diari = diari.copy()
    for _, r in salts_df.sort_values("data_salt", ascending=False).iterrows():
        # del salt més recent al més antic: el tram anterior s'escala al nivell posterior
        m = (diari["cp"] == r["cp"]) & (diari["data"] <= pd.Timestamp(r["data_salt"]))
        diari.loc[m, "kwh"] *= r["factor"]
    print("[OK] Trams anteriors als salts reescalats (CORREGIR_SALTS = True)")
salts_df.to_csv("salts_nivell_v12.csv", index=False)
```

## 14. Construcció de les variables del model

Per a cada dia $d$:
- $\mathrm{CDD}_k(d) = \max(T_{mean}(d-k) - T_c, 0)$ per a $k=0,\dots,L$ (el forçament i la seva memòria).
- **Efectes fixos** (el que NO és temperatura): any (tendència, preus), mes (vacances d'agost!), dia de la setmana, festius.

Definim els **dies d'onada de calor** a partir de les dades, sense posar dates a mà: dies amb $T_{max}$ per sobre del percentil 95 dels estius, en ratxes de com a mínim 3 dies seguits.

```python
festius = holidays.country_holidays("ES", subdiv="CT", years=ANYS)

def afegeix_cdd(t, Tc, lags):
    t = t.copy()
    for k in range(lags + 1):
        t[f"cdd{k}"] = np.maximum(t["tmean"].shift(k) - Tc, 0)
    return t

# --- Onades de calor, definides a partir de les dades ---
estiu_t = temp[temp["data"].dt.month.isin(MESOS_ESTIU)]
llindar = estiu_t["tmax"].quantile(PERCENTIL_ONADA)
calent = temp["tmax"] >= llindar
ratxa_id = (calent != calent.shift()).cumsum()           # numera les ratxes consecutives
mida_ratxa = calent.groupby(ratxa_id).transform("sum")
temp["onada"] = calent & (mida_ratxa >= DIES_MIN_ONADA)

onades = temp[temp["onada"]]
print(f"Llindar Tmax (P{int(PERCENTIL_ONADA*100)} estiu): {llindar:.1f} °C")
print(f"Dies d'onada detectats: {len(onades)}")
for (a, m), g in onades.groupby([onades.data.dt.year, onades.data.dt.month]):
    print(f"  {a}-{m:02d}: {len(g)} dies ({g.data.min().date()} → {g.data.max().date()})")
```

## 15. Triar $T_c$ amb la sèrie de ciutat

Provem diferents temperatures de confort i ens quedem amb la que millor explica el consum d'estiu de la ciutat (menor error quadràtic, ajust només amb dies normals). És un **perfil de versemblança** d'un sol paràmetre.

```python
def matriu_disseny(taula, lags):
    # Regressors: CDD amb retards + efectes fixos (variables dummy)
    X_cdd = taula[[f"cdd{k}" for k in range(lags + 1)]]
    # Categories fixes: així els dies normals i els d'onada tenen exactament les mateixes columnes
    cat = lambda valors, cats: pd.Categorical(valors, categories=cats)
    fe = pd.get_dummies(pd.DataFrame({
        "any":    cat(taula["data"].dt.year, ANYS),
        "estacio": (cat(taula["data"].dt.isocalendar().week.astype(int).values, list(range(1, 54)))
                    if EFECTE_ESTACIONAL == "setmana"
                    else cat(taula["data"].dt.month, MESOS_ESTIU)),
        "dia":    cat(taula["data"].dt.dayofweek, list(range(7))),
        "festiu": cat(taula["data"].isin(pd.to_datetime(list(festius.keys()))), [False, True]),
    }), drop_first=True).astype(float)
    X = pd.concat([X_cdd.reset_index(drop=True), fe.reset_index(drop=True)], axis=1)
    X.insert(0, "const", 1.0)
    return X

def ajusta(taula, lags):
    # Mínims quadrats ordinaris amb numpy (sense caixes negres).
    # v0.9: l'error es calcula per CLÚSTERS DE SETMANA (fórmula "sandwich"): els residus de dies seguits
    # s'assemblen i l'error clàssic de MQO era massa petit. Afecta les seccions 15-18 i A6.
    X = matriu_disseny(taula, lags)
    y = np.log(taula["kwh"].values)
    Xv = X.values.astype(float)
    beta, *_ = np.linalg.lstsq(Xv, y, rcond=None)
    res = y - Xv @ beta
    n, p = X.shape
    pinv = np.linalg.pinv(Xv.T @ Xv)
    iso = taula["data"].dt.isocalendar()
    g = (iso["year"].astype(int) * 100 + iso["week"].astype(int)).values
    G = len(np.unique(g))
    if G > 1:
        meat = np.zeros((p, p))
        for gi in np.unique(g):
            s_ = Xv[g == gi].T @ res[g == gi]
            meat += np.outer(s_, s_)
        cov = pinv @ meat @ pinv * (G / (G - 1)) * ((n - 1) / max(n - p, 1))
    else:
        cov = (res @ res / max(n - p, 1)) * pinv
    # clip: evita arrels de nombres negatius minúsculs per errors d'arrodoniment (efectes fixos sense dades)
    return pd.Series(beta, index=X.columns), pd.Series(np.sqrt(np.clip(np.diag(cov), 0, None)), index=X.columns), res

def prepara(serie, Tc, tsrc=None):
    # tsrc: sèrie de temperatura a fer servir (per defecte, ERA5 de tota la ciutat)
    t = afegeix_cdd(temp if tsrc is None else tsrc, Tc, LAGS)
    taula = serie.merge(t, on="data")
    taula = taula[taula["data"].dt.month.isin(MESOS_ESTIU) & taula["data"].dt.year.isin(ANYS)]
    return taula.dropna().reset_index(drop=True)

perfil = []
for Tc in TC_CANDIDATS:
    taula = prepara(ciutat, Tc)
    normals = taula[~taula["onada"]]
    _, _, res = ajusta(normals, LAGS)
    perfil.append((Tc, np.mean(res**2)))
perfil = pd.DataFrame(perfil, columns=["Tc", "mse"])
TC = perfil.loc[perfil["mse"].idxmin(), "Tc"]

plt.figure(figsize=(6,3.5))
plt.plot(perfil["Tc"], perfil["mse"], "o-")
plt.axvline(TC, ls="--", c="grey")
plt.xlabel("$T_c$ (°C)"); plt.ylabel("Error quadràtic mitjà"); plt.title(f"Tria de $T_c$ → {TC} °C")
plt.show()
print(f"[OK] Temperatura de confort triada: Tc = {TC} °C")

# --- La mateixa tria amb la temperatura de les estacions (és més alta que ERA5: Tc també ho serà) ---
temp_est_ciutat = temp_est_ciutat.merge(temp[["data", "onada"]], on="data")
perfil_est = []
for Tc in TC_CANDIDATS:
    taula = prepara(ciutat, Tc, temp_est_ciutat)
    _, _, res = ajusta(taula[~taula["onada"]], LAGS)
    perfil_est.append((Tc, np.mean(res**2)))
perfil_est = pd.DataFrame(perfil_est, columns=["Tc", "mse"])
TC_EST = perfil_est.loc[perfil_est["mse"].idxmin(), "Tc"]
print(f"[OK] Tc amb les estacions: {TC_EST} °C")
```

## 16. Resposta de la ciutat sencera

Primer, el sistema com un tot. Com que el consum va en logaritme, cada coeficient $\chi(k)$ és el **tant per u de consum extra per cada grau per sobre de $T_c$**, $k$ dies després.

- $\chi(k)$ per a cada $k$ = **funció de resposta** (amb memòria).
- $\chi^0=\sum_k \chi(k)$ = **susceptibilitat estàtica**: resposta a un escalfament sostingut.

```python
taula_c = prepara(ciutat, TC)
beta_c, se_c, _ = ajusta(taula_c[~taula_c["onada"]], LAGS)
chi_c    = np.array([beta_c[f"cdd{k}"] for k in range(LAGS + 1)])
chi_c_se = np.array([se_c[f"cdd{k}"]   for k in range(LAGS + 1)])

plt.figure(figsize=(6,3.5))
plt.errorbar(range(LAGS + 1), 100*chi_c, yerr=200*chi_c_se, fmt="o-", capsize=4)
plt.axhline(0, c="grey", lw=0.8)
plt.xlabel("Retard k (dies)"); plt.ylabel("χ(k)  (% consum / K)")
plt.title("Funció de resposta de la ciutat (barres: ±2 errors estàndard aprox.)")
plt.show()
print(f"Susceptibilitat estàtica de la ciutat: χ⁰ = {100*chi_c.sum():.2f} % per K")
```

## 17. Resposta per codi postal

Repetim el mateix ajust, per separat, per a cada codi postal. Obtenim un $\chi^0_i$ per CP: quant creix el consum d'aquell tros de ciutat per cada grau de calor.

```python
def font_cp(cp, font=None):
    # Retorna (sèrie de temperatura, Tc) per a un CP segons la font triada
    font = font or FONT_TEMP_CP
    if font == "estacions" and cp in tloc:
        return tloc[cp].merge(temp[["data", "onada"]], on="data"), TC_EST
    return None, TC

def ajusta_cps(font):
    resultats = []
    for cp, serie_cp in diari.groupby("cp"):
        tsrc, Tc = font_cp(cp, font)
        taula = prepara(serie_cp[["data","kwh"]], Tc, tsrc)
        normals = taula[~taula["onada"]]
        if len(normals) < 100:
            continue
        beta, se, res = ajusta(normals, LAGS)
        chi = np.array([beta[f"cdd{k}"] for k in range(LAGS + 1)])
        chi0_se = np.sqrt(sum(se[f"cdd{k}"]**2 for k in range(LAGS + 1)))   # aprox.: ignora covariàncies
        resultats.append({"cp": cp, "chi_estatica": chi.sum(), "chi_estatica_se": chi0_se,
                          **{f"chi_lag{k}": chi[k] for k in range(LAGS + 1)},
                          "consum_mitja_kwh": serie_cp["kwh"].mean(), "n_dies": len(normals)})
    return pd.DataFrame(resultats).sort_values("chi_estatica", ascending=False).reset_index(drop=True)

res_era5 = ajusta_cps("era5")
res_est  = ajusta_cps("estacions")
res_cp = res_est if FONT_TEMP_CP == "estacions" else res_era5
TC_CP  = TC_EST if FONT_TEMP_CP == "estacions" else TC

# --- Comparació: canvia el mapa de χ⁰ quan cada CP té la seva temperatura? ---
cmp_chi = res_era5[["cp","chi_estatica"]].merge(res_est[["cp","chi_estatica"]], on="cp", suffixes=("_era5","_est"))
cmp_chi["canvi_pct"] = 100 * (cmp_chi["chi_estatica_est"] - cmp_chi["chi_estatica_era5"])
r_s, _ = stats.spearmanr(cmp_chi["chi_estatica_era5"], cmp_chi["chi_estatica_est"])
print(f"χ⁰ mediana: ERA5 {100*res_era5.chi_estatica.median():.2f} %/K · estacions {100*res_est.chi_estatica.median():.2f} %/K")
print(f"Ordre dels CP (Spearman ERA5 vs estacions): ρ = {r_s:.2f}")
print("\nCP on més baixa χ⁰ amb la temperatura local (part de la seva 'susceptibilitat' era escalfament local):")
print(cmp_chi.sort_values("canvi_pct").head(6).round(3).to_string(index=False))
print("\nCP on més puja:")
print(cmp_chi.sort_values("canvi_pct", ascending=False).head(6).round(3).to_string(index=False))

fig, ax = plt.subplots(1, 2, figsize=(14, 4.5), gridspec_kw={"width_ratios": [1, 2.3]})
ax[0].scatter(100*cmp_chi["chi_estatica_era5"], 100*cmp_chi["chi_estatica_est"], s=18)
lim = [0, 100*cmp_chi[["chi_estatica_era5","chi_estatica_est"]].max().max()*1.05]
ax[0].plot(lim, lim, "k--", lw=0.8)
for _, r in cmp_chi.iterrows():
    ax[0].annotate(r["cp"][-2:], (100*r["chi_estatica_era5"], 100*r["chi_estatica_est"]), fontsize=7)
ax[0].set_xlabel("χ⁰ amb ERA5 (%/K)"); ax[0].set_ylabel("χ⁰ amb estacions (%/K)"); ax[0].set_title("ERA5 vs temperatura local")

ax[1].bar(res_cp["cp"], 100*res_cp["chi_estatica"], yerr=200*res_cp["chi_estatica_se"], capsize=2)
ax[1].set_ylabel("χ⁰  (% consum / K)"); ax[1].set_title(f"Susceptibilitat per CP (temperatura: {FONT_TEMP_CP}, Tc = {TC_CP} °C)")
ax[1].tick_params(axis="x", rotation=90, labelsize=7)
plt.tight_layout(); plt.show()
res_cp.head(10)
```

## 18. El test: prediu la resposta lineal les onades de calor?

Aquest és el resultat que importa. El model s'ha ajustat **sense** els dies d'onada. Ara prediem aquests dies i mirem el residu:
$$r_i = \ln E_i^{\text{observat}} - \ln E_i^{\text{predit}}$$

- $r_i \approx 0$ → la resposta lineal funciona fins i tot en condicions extremes.
- $r_i > 0$ → el consum **s'amplifica** més del que diu la llei lineal (més aparells encesos, efecte acumulat, bucle amb la calor antropogènica?).
- $r_i < 0$ → el consum **satura** (límit de capacitat, gent que marxa de la ciutat, parc d'aparells ja al màxim).

Compararem amb el residu dels dies normals per saber si la diferència és real.

```python
def residus_onada(serie, Tc, tsrc=None):
    taula = prepara(serie, Tc, tsrc)
    normals, calents = taula[~taula["onada"]], taula[taula["onada"]]
    if len(calents) == 0:
        return None
    beta, _, res_n = ajusta(normals, LAGS)
    Xc = matriu_disseny(calents, LAGS).reindex(columns=beta.index, fill_value=0.0)
    res_c = np.log(calents["kwh"].values) - Xc.values @ beta.values
    return res_n, res_c, calents["data"].values

# --- Ciutat ---
rn, rc, dates_c = residus_onada(ciutat, TC)
print(f"Ciutat · residu mitjà en dies d'onada: {100*rc.mean():+.2f} %  "
      f"(dispersió dies normals: ±{100*rn.std():.2f} %)")

# --- Per codi postal ---
nl = []
for cp, serie_cp in diari.groupby("cp"):
    tsrc, Tc_i = font_cp(cp)
    out = residus_onada(serie_cp[["data","kwh"]], Tc_i, tsrc)
    if out is None or len(out[1]) < 3:
        continue
    rn_i, rc_i, _ = out
    # z: quantes desviacions típiques (de l'error de la mitjana) s'aparta el residu d'onada
    z = rc_i.mean() / (rn_i.std() / np.sqrt(len(rc_i)))
    nl.append({"cp": cp, "residu_onada": rc_i.mean(), "z": z})
nl = pd.DataFrame(nl)
res_cp = res_cp.merge(nl, on="cp", how="left")

fig, ax = plt.subplots(1, 2, figsize=(13,4.5))
ax[0].hist(100*rn, bins=40, alpha=0.6, density=True, label="dies normals")
ax[0].hist(100*rc, bins=15, alpha=0.6, density=True, label="dies d'onada")
ax[0].set_xlabel("Residu (%)"); ax[0].set_title("Ciutat: residus del model lineal"); ax[0].legend()

ax[1].scatter(100*res_cp["chi_estatica"], 100*res_cp["residu_onada"], s=18)
for _, r in res_cp.iterrows():
    if abs(r["z"]) > 3:
        ax[1].annotate(r["cp"], (100*r["chi_estatica"], 100*r["residu_onada"]), fontsize=7)
ax[1].axhline(0, c="grey", lw=0.8)
ax[1].set_xlabel("χ⁰ (% / K)"); ax[1].set_ylabel("Residu mitjà en onada (%)")
ax[1].set_title("On es trenca la linealitat? (etiquetats: |z|>3)")
plt.tight_layout(); plt.show()

print(f"Codis postals amb |z| > 3: {(res_cp['z'].abs() > 3).sum()} de {len(res_cp)}")
```

## 18b. Diagnòstic: què passa exactament els dies d'onada?

Si el residu d'onada és molt gran i del mateix signe a tota la ciutat, abans de parlar de física cal descartar tres coses:
1. **Calendari:** les onades cauen a mitjan agost, quan la ciutat es buida (els efectes fixos no ho capten del tot).
2. **Dades:** dies amb menys files (trams o CP que falten).
3. **Extrapolació:** el model lineal prediu massa quan el forçament és molt més gran que en els dies d'ajust.

La taula mostra, dia per dia (ciutat sencera): la data, el dia de la setmana, la temperatura, la contribució de la calor que prediu el model i el residu. **Copia-la i enganxa-la a la conversa** (és text, no cal captura).

```python
taula_c = prepara(ciutat, TC)
normals_c, calents_c = taula_c[~taula_c["onada"]], taula_c[taula_c["onada"]]
beta_c, _, _ = ajusta(normals_c, LAGS)
Xc = matriu_disseny(calents_c, LAGS).reindex(columns=beta_c.index, fill_value=0.0)
pred = Xc.values @ beta_c.values
contrib_calor = sum(beta_c[f"cdd{k}"] * calents_c[f"cdd{k}"].values for k in range(LAGS + 1))

diag = pd.DataFrame({
    "data": calents_c["data"].dt.date.values,
    "dia": calents_c["data"].dt.day_name().str[:3].values,
    "tmax": calents_c["tmax"].round(1).values,
    "tmean": calents_c["tmean"].round(1).values,
    "calor_predita_%": (100 * contrib_calor).round(1),
    "residu_%": (100 * (np.log(calents_c["kwh"].values) - pred)).round(1),
    "files": files_dia.reindex(calents_c["data"]).values,
})
print(diag.to_string(index=False))

# Mateix residu separant agost de la resta
es_agost = calents_c["data"].dt.month.values == 8
res = np.log(calents_c["kwh"].values) - pred
if es_agost.any():
    print(f"\nResidu mitjà dies d'onada a l'AGOST:     {100*res[es_agost].mean():+.1f} %  ({es_agost.sum()} dies)")
if (~es_agost).any():
    print(f"Residu mitjà dies d'onada FORA d'agost: {100*res[~es_agost].mean():+.1f} %  ({(~es_agost).sum()} dies)")
print(f"Efecte estacional usat: {EFECTE_ESTACIONAL}")
```

## 21. Renda i calor antropogènica per CP

**Què fa:** prepara dues variables per CP que fan servir les seccions 22g, 23f i 27: la **renda** (mitjana dels barris de cada CP, llegida de la memòria cau `renda_cp.csv`) i la **calor antropogènica elèctrica** $Q_F$ (W/m²) de l'estiu per sector. Abans es calculaven a les seccions A3 i A4, que ara són a l'annex amb els seus tests i gràfics.

**Què cal mirar:** que surtin ~40 CP amb renda i que $Q_F$ total tingui una mediana de l'ordre de 7-9 W/m².

```python
# ============ 21. Renda i calor antropogènica per CP (per a les seccions 22g, 23f i 27) ============
# Calor antropogènica: tota l'electricitat que es consumeix acaba convertida en calor.
# Consum mitjà d'estiu (kWh/dia) / àrea del CP (m²)  ->  flux de calor Q_F (W/m²).   kWh/dia × 1000/24 = W
s_all = df[df["data"].isin(ciutat["data"]) & df["data"].dt.month.isin(MESOS_ESTIU)]
per_sector = (s_all.groupby(["cp", "sector", "data"])["kwh"].sum()        # diari per CP i sector
                   .groupby(["cp", "sector"]).mean().unstack(fill_value=0))  # mitjana d'estiu
qf = pd.DataFrame({
    "qf_residencial": per_sector[sector_triat] * 1000 / 24 / area_cp,
    "qf_no_residencial": per_sector.drop(columns=[sector_triat]).sum(axis=1) * 1000 / 24 / area_cp,
}).dropna()
qf["qf_total"] = qf.sum(axis=1)
print("Flux de calor antropogènica elèctric a l'estiu (W/m², mitjana del CP):")
print(qf.describe().loc[["min", "50%", "max"]].round(2).to_string())

# Renda per CP: de la memòria cau (la calcula la secció A3 de l'annex a partir de les adreces i els barris)
try:
    renda_cp = pd.read_csv(f"{CARPETA_LOCAL}/renda_cp.csv", dtype={"cp": str}).set_index("cp")["renda"]
    print(f"\n[OK] Renda per CP llegida de la memòria cau: {len(renda_cp)} CP, "
          f"de {renda_cp.min():,.0f} a {renda_cp.max():,.0f} €")
except Exception as e:
    print(f"\n[i] No hi ha renda_cp.csv a {CARPETA_LOCAL} ({e}). "
          "Les seccions 22g i 23f funcionaran sense renda; per calcular-la, executa la secció A3 de l'annex.")
```

## 22. Codis postals i tendència de la susceptibilitat (v0.9)

**Què hem après amb la v0.6:**

3. El test "suma de tots els veïns" tenia poca potència: amb 6-11 veïns, qualsevol salt d'un CP petit semblava conservat. Ara fem un **balanç en kWh/dia**.

**Canvi de criteri (documentat):** el criteri fixat a la v0.6 suposava que $\chi^0$ no canvia amb els anys. La secció 22d va mostrar que sí que canvia, i per això el substituïm per comparacions dins de cada any. Aquest canvi es fa **després** d'haver vist els resultats i així s'ha de dir a l'article.

**Seccions:**

  Dades extra: es baixen soles; si falla, es llegeixen de `CARPETA_LOCAL` (`humitat_era5.csv`, `preus_electricitat.csv`, `hut_comunicacio_opendata.csv`).

```python
# ============ 22a. El CSV porta el nombre de comptadors? ============
print("Columnes del CSV original:", list(elec_raw.columns))
paraules = ["compt", "contad", "contract", "suministr", "subministr", "punt", "cups", "num", "n_"]
candidates = [c for c in elec_raw.columns if any(p in c.lower() for p in paraules)]
print("Columnes que podrien ser nombre de comptadors:", candidates if candidates else "cap")
```

```python
# ============ 22b. Balanç en kWh de cada salt ============
# Per a cada salt de la secció 12:
#   Δ_cp   = canvi del consum del CP (kWh/dia), mesurat com a canvi de QUOTA × consum de la ciutat
#   Δ_v    = el mateix per a cada veí v
#   Δ_nr   = el mateix per als sectors NO residencials del mateix CP
# Si un veí té un canvi de signe contrari i de mida semblant (balanç entre 0,5 i 2), els comptadors
# han anat a aquell veí. Si és el no residencial del mateix CP, és una reclassificació de sector.
geo = pol.set_index("cp").geometry   # polígons dels CP (abans es definia a la secció 19, eliminada)
BALANC = (0.5, 2.0)   # -Δ_soci / Δ_cp ha de caure en aquest interval
W = pd.Timedelta(days=FINESTRA)

serie_res = {cp: g.groupby("data")["kwh"].sum().sort_index() for cp, g in diari_brut.groupby("cp")}
no_res = (df[(df["sector"] != sector_triat) & df["cp"].isin(serie_res)]
            .groupby(["cp", "data"])["kwh"].sum())
TOTAL_NR = no_res.groupby("data").sum()
# Referència estable: la suma dels CP SENSE cap salt. Si un CP veí fa un salt, no contamina la quota dels altres.
REF = sum(s for cp, s in serie_res.items() if cp not in set(salts_df["cp"]))

def delta_kwh(s, d, total):
    # Canvi de nivell (kWh/dia) d'una sèrie a la data d: canvi de quota sobre 'total' × nivell de 'total'
    q = s / total.reindex(s.index)
    abans   = q[(q.index > d - W) & (q.index <= d)].median()
    despres = q[(q.index > d) & (q.index <= d + W)].median()
    nivell  = total[(total.index > d - W) & (total.index <= d + W)].median()
    return (despres - abans) * nivell

def veins_de(cp):
    if cp not in geo.index:
        return []
    return [v for v in geo.index if v != cp and v in serie_res and geo[cp].buffer(20).intersects(geo[v])]

def dins(b):
    return np.isfinite(b) and BALANC[0] <= b <= BALANC[1]

files = []
for _, r in salts_df.iterrows():
    cp, d = r["cp"], pd.Timestamp(r["data_salt"])
    if cp not in serie_res:
        continue
    d_cp = delta_kwh(serie_res[cp], d, REF)
    dv = {v: delta_kwh(serie_res[v], d, REF) for v in veins_de(cp)}
    oposats = {v: x for v, x in dv.items() if np.sign(x) == -np.sign(d_cp)}
    soci = max(oposats, key=lambda v: abs(oposats[v])) if oposats else None
    b_soci = -oposats[soci] / d_cp if soci else np.nan
    b_tots = -sum(oposats.values()) / d_cp if oposats else np.nan
    d_nr = delta_kwh(no_res.loc[cp].sort_index(), d, TOTAL_NR) if cp in no_res.index.get_level_values(0) else np.nan
    b_nr = -d_nr / d_cp if np.isfinite(d_nr) else np.nan
    if dins(b_soci):
        tipus = "reassignació a un veí"
    elif dins(b_nr):
        tipus = "reclassificació de sector"
    elif dins(b_tots):
        tipus = "reassignació repartida"
    else:
        tipus = "no es conserva"
    files.append({"cp": cp, "data_salt": d.date(), "factor": r["factor"], "dkwh_cp": round(d_cp),
                  "soci": soci, "balanc_soci": round(b_soci, 2), "balanc_veins": round(b_tots, 2),
                  "balanc_no_res": round(b_nr, 2), "tipus": tipus})
classif = pd.DataFrame(files)
print(f"Salts: {len(classif)}")
print(classif["tipus"].value_counts().to_string())
print()
print(classif.sort_values(["data_salt", "cp"]).to_string(index=False))
classif.to_csv("classificacio_salts_v12.csv", index=False)

# --- Salts sense ningú que els compensi: entrades o sortides netes de comptadors a la ciutat ---
nets = classif[classif["tipus"] == "no es conserva"]
print(f"\nCanvi net de consum residencial que ningú compensa: {nets['dkwh_cp'].sum()/1e3:+.0f} MWh/dia "
      f"({100*nets['dkwh_cp'].sum()/TOTAL.median():+.1f} % del total de la ciutat)")
print("  per any:", (nets.groupby(pd.to_datetime(nets["data_salt"]).dt.year)["dkwh_cp"].sum() / 1e3).round(0).to_dict(), "MWh/dia")

# --- Unitats de conservació noves: només CP units amb el seu soci directe ---
pare = {cp: cp for cp in serie_res}
def arrel(x):
    while pare[x] != x:
        x = pare[x]
    return x
for _, r in classif[classif["tipus"] == "reassignació a un veí"].iterrows():
    pare[arrel(r["soci"])] = arrel(r["cp"])
unitats = {}
for cp in serie_res:
    unitats.setdefault(arrel(cp), []).append(cp)
unitats = [sorted(u) for u in unitats.values() if len(u) > 1]
print(f"\nUnitats de conservació: {len(unitats)}")
for u in unitats:
    print(f"  {len(u)} CP: {', '.join(u)}")
if any(len(u) > 6 for u in unitats):
    print("  [!] Hi ha unitats de més de 6 CP: mira-les al mapa abans de fer-les servir")
```

```python
# ============ 22c. χ⁰ per CP en diverses versions ============
# Funcions comunes de la secció 22. Les errors es calculen de dues maneres:
#   - MQO (com fins ara)
#   - per CLÚSTERS de setmana: els residus de dies seguits s'assemblen i el MQO subestima l'error.
def disseny_v07(taula, lags, per_any=False, amb_trams=False):
    X = matriu_disseny(taula, lags).reset_index(drop=True)
    if per_any:
        for a in ANYS:
            m = (taula["data"].dt.year == a).values.astype(float)
            for k in range(lags + 1):
                X[f"cdd{k}_{a}"] = X[f"cdd{k}"].values * m
        X = X.drop(columns=[f"cdd{k}" for k in range(lags + 1)])
    if amb_trams and "tram" in taula.columns:
        dt = pd.get_dummies(taula["tram"].astype(str), prefix="tram", drop_first=True).astype(float)
        X = pd.concat([X, dt.reset_index(drop=True)], axis=1)
    return X.loc[:, (X != 0).any(axis=0)]

def ajusta_cl(taula, X, y=None):
    # MQO + error per clústers de setmana ISO (fórmula "sandwich")
    y = np.log(taula["kwh"].values) if y is None else y      # y: per defecte el logaritme del consum
    Xv = X.values
    beta, *_ = np.linalg.lstsq(Xv, y, rcond=None)
    u = y - Xv @ beta
    n, p = len(y), np.linalg.matrix_rank(Xv)
    pinv = np.linalg.pinv(Xv.T @ Xv)
    cov_ols = (u @ u / (n - p)) * pinv
    iso = taula["data"].dt.isocalendar()
    g = (iso["year"].astype(int) * 100 + iso["week"].astype(int)).values
    meat = np.zeros((Xv.shape[1], Xv.shape[1]))
    for gi in np.unique(g):
        s = Xv[g == gi].T @ u[g == gi]
        meat += np.outer(s, s)
    G = len(np.unique(g))
    cov_cl = pinv @ meat @ pinv * (G / (G - 1)) * ((n - 1) / (n - p))
    cols = X.columns
    return (pd.Series(beta, index=cols), pd.DataFrame(cov_ols, index=cols, columns=cols),
            pd.DataFrame(cov_cl, index=cols, columns=cols), u)

def chi0_de(beta, cov, sufix=""):
    c = [f"cdd{k}{sufix}" for k in range(LAGS + 1)]
    if not all(x in beta.index for x in c):
        return np.nan, np.nan
    return 100 * beta[c].sum(), 100 * np.sqrt(cov.loc[c, c].values.sum())

def creixement(beta, cov, a1, a2):
    # χ⁰(a2) − χ⁰(a1) i el seu error, tenint en compte la covariància entre els dos anys
    c1 = [f"cdd{k}_{a1}" for k in range(LAGS + 1)]
    c2 = [f"cdd{k}_{a2}" for k in range(LAGS + 1)]
    if not all(x in beta.index for x in c1 + c2):
        return np.nan, np.nan
    w = pd.Series(0.0, index=beta.index); w[c2] = 1.0; w[c1] = -1.0
    return 100 * float(w @ beta), 100 * float(np.sqrt(w @ cov @ w))

def afegeix_trams(taula, cp):
    dates = np.sort(pd.to_datetime(salts_df.loc[salts_df["cp"] == cp, "data_salt"]).values) if len(salts_df) else np.array([], dtype="datetime64[ns]")
    taula = taula.copy()
    taula["tram"] = np.searchsorted(dates, taula["data"].values, side="left")
    return taula

def taula_de(serie, Tc=None, tsrc=None, onades=False, anys=None, cp=None):
    # Taula de regressió d'una sèrie diària (columnes data, kwh): estius, CDD, i opcionalment trams
    taula = prepara(serie[["data", "kwh"]], TC if Tc is None else Tc, tsrc)
    taula = taula[taula["kwh"] > 0]          # dies amb 0 kWh (franges buides): el logaritme no existeix
    if anys is not None:
        taula = taula[taula["data"].dt.year.isin(anys)]
    if not onades:
        taula = taula[~taula["onada"]]
    taula = taula.reset_index(drop=True)
    if cp is not None:
        taula = afegeix_trams(taula, cp)
    return taula

def chi0_versio(serie, cp=None, anys=None, amb_trams=False):
    taula = taula_de(serie, anys=anys, cp=cp if amb_trams else None)
    if len(taula) < 100:
        return np.nan, np.nan
    beta, _, cov, _ = ajusta_cl(taula, disseny_v07(taula, LAGS, amb_trams=amb_trams))
    return chi0_de(beta, cov)

amb_salts = set(salts_df["cp"]) if len(salts_df) else set()
files = []
for cp, g in diari_brut.groupby("cp"):
    s_brut = g[["data", "kwh"]]
    fila = {"cp": cp, "te_salts": cp in amb_salts}
    fila["chi_net"],       fila["se_net"]       = chi0_versio(s_brut, anys=[2021, 2022])
    fila["chi_trams2122"], fila["se_trams2122"] = chi0_versio(s_brut, cp, anys=[2021, 2022], amb_trams=True)
    fila["chi_trams"],     fila["se_trams"]     = chi0_versio(s_brut, cp, amb_trams=True)
    files.append(fila)
versions = pd.DataFrame(files).set_index("cp")
versions["chi_unitat"], versions["se_unitat"] = versions["chi_trams"], versions["se_trams"]
for u in unitats:
    suma = diari_brut[diari_brut["cp"].isin(u)].groupby("data", as_index=False)["kwh"].sum()
    versions.loc[versions.index.isin(u), ["chi_unitat", "se_unitat"]] = chi0_versio(suma)
print("χ⁰ (%/K) per CP · errors per clústers de setmana")
print(versions.round(2).sort_values("chi_trams", ascending=False).to_string())
```

```python
# ============ 22d. χ⁰ per a cada estiu: ciutat i CP ============
def chi_per_any(taula, extres=None, cluster=True):
    # Una funció de resposta per any; 'extres' = columnes addicionals (control de preu, terme quadràtic...)
    X = disseny_v07(taula, LAGS, per_any=True, amb_trams="tram" in taula.columns)
    if extres is not None:
        X = pd.concat([X, extres.reset_index(drop=True)], axis=1)
    beta, cov_ols, cov_cl, u = ajusta_cl(taula, X)
    cov = cov_cl if cluster else cov_ols
    fila = {}
    for a in ANYS:
        fila[a], fila[f"se_{a}"] = chi0_de(beta, cov, f"_{a}")
    fila["creix_21_24"], fila["se_creix"] = creixement(beta, cov, 2021, 2024)
    return fila, beta, cov

# v0.9: la ciutat només amb els dies en què hi ha TOTS els CP (la 5 treu alguns parells CP-dia anòmals,
# i en aquells dies la suma de la ciutat sortiria artificialment baixa)
n_cp_dia = diari_brut.groupby("data")["cp"].nunique()
# v0.11: un CP sense dades en un trimestre (buit, secció 4) ja no s'espera en aquell trimestre
vius = diari_brut.groupby(diari_brut["data"].dt.to_period("Q"))["cp"].nunique()
esperats = pd.Series(vius.reindex(n_cp_dia.index.to_period("Q")).values, index=n_cp_dia.index)
dies_complets = n_cp_dia.index[n_cp_dia >= esperats]
ciutat_brut = diari_brut[diari_brut["data"].isin(dies_complets)].groupby("data", as_index=False)["kwh"].sum()
print(f"Dies de ciutat complets: {len(dies_complets)} de {len(n_cp_dia)}")
base, beta_base, cov_base = chi_per_any(taula_de(ciutat_brut))
base_ols, _, _ = chi_per_any(taula_de(ciutat_brut), cluster=False)
print("Ciutat · χ⁰ per estiu (%/K): error MQO → error per clústers de setmana")
for a in ANYS:
    print(f"  {a}: {base[a]:5.2f}  ±{base_ols[f'se_{a}']:.2f} → ±{base[f'se_{a}']:.2f}")
print(f"  Creixement 2021→2024: {base['creix_21_24']:+.2f} ± {base['se_creix']:.2f} %/K")

files = []
for cp, g in diari_brut.groupby("cp"):
    try:
        f, _, _ = chi_per_any(taula_de(g, cp=cp))
        files.append({"cp": cp, **{a: f[a] for a in ANYS}, **{f"se_{a}": f[f"se_{a}"] for a in ANYS}})
    except Exception as e:
        print(f"  [!] {cp}: {e}")
per_any_cp = pd.DataFrame(files).set_index("cp")
per_any_cp.to_csv("chi_per_estiu_v12.csv")
```

```python
# ============ 22e. Qualitat del mapa, any per any ============
# Criteri nou (vegeu la capçalera de la secció 22):
#   1. Els salts no distorsionen la tendència: el creixement de χ⁰ (mitjana 2023-24 menys mitjana 2021-22)
#      dels CP AMB salts no és diferent del dels CP SENSE salts (Mann-Whitney, p > 0,05).
#   2. La relació amb la illa de calor no depèn de l'any: ρ(χ⁰, ΔLST) té el mateix signe cada any.
P = per_any_cp.copy()
# v0.11: 2023-24 i no 2024-25. Al juliol del 2025 hi ha una recodificació de CP a l'origen (alerta de dades 1)
P["creix"] = P[[2023, 2024]].mean(axis=1) - P[[2021, 2022]].mean(axis=1)
P["te_salts"] = P.index.isin(amb_salts)
P = P.join(illa[["dLST"]], how="left")

amb, sense = P.loc[P["te_salts"], "creix"].dropna(), P.loc[~P["te_salts"], "creix"].dropna()
mw = stats.mannwhitneyu(amb, sense)
cond1 = mw.pvalue > 0.05
print(f"Creixement de χ⁰ (2023-24 menys 2021-22): sense salts mediana {sense.median():+.2f} ({len(sense)} CP) · "
      f"amb salts {amb.median():+.2f} ({len(amb)} CP) · Mann-Whitney p = {mw.pvalue:.3f}")
print(f"Condició 1 → {'OK' if cond1 else 'NO'}")

print("\nχ⁰ contra illa de calor (ΔLST), any per any:")
signes = []
for a in ANYS:
    for nom, sub in [("tots", P), ("sense salts", P[~P["te_salts"]])]:
        d = sub.dropna(subset=[a, "dLST"])
        rho, pv = stats.spearmanr(d["dLST"], d[a])
        if nom == "tots":
            signes.append(np.sign(rho))
        print(f"  {a} · {nom:11s}: ρ = {rho:+.2f} (p = {pv:.3f}, {len(d)} CP)")
cond2 = len(set(signes)) == 1
print(f"Condició 2 → {'OK' if cond2 else 'NO'}")

print("\nOrdre dels CP entre estius (Spearman), només CP sense salts:")
print(P.loc[~P["te_salts"], ANYS].corr(method="spearman").round(2).to_string())

print("\n" + "=" * 70)
if cond1 and cond2:
    print("DECISIÓ: els mapes per any (amb trams) es poden fer servir.")
elif cond1:
    print("DECISIÓ: els salts no distorsionen la tendència, però la relació amb la illa de calor canvia amb l'any.\n"
          "          Cal entendre per què abans de publicar un mapa (secció 22g).")
else:
    print("DECISIÓ: els CP amb salts tenen una tendència diferent: mapa només amb CP sense salts o només 2021-2022.")
print("=" * 70)
```

```python
# ============ 22f. Proves de la tendència (v0.9) ============
# Cada prova repeteix χ⁰ per estiu de la CIUTAT (residencial) canviant UNA cosa.
# Si el creixement 2021→2024 desapareix amb un canvi, aquella és (part de) l'explicació.
# Totes les dades extra es baixen soles; si la baixada falla, es llegeixen de CARPETA_LOCAL:
#   humitat_era5.csv, preus_electricitat.csv, hut_comunicacio_opendata.csv (pisos turístics, Open Data BCN)
import requests
DIES_OK = set(diari_brut["data"])          # els mateixos dies que el residencial (ja sense dies dolents ni atípics)

def serie_df(sector=None, tram=None, cps=None):
    # Consum diari de ciutat a partir de 'df' (tots els sectors i franges), amb els mateixos dies vàlids
    d = df[df["cp"].isin(serie_res) & df["data"].isin(DIES_OK)]
    if "tram" in d.columns:
        d = d[~d["tram"].astype(str).str.contains("consta", case=False)]     # franja "No consta": buida
    if sector is not None:
        d = d[d["sector"].isin([sector] if isinstance(sector, str) else sector)]
    if tram is not None:
        d = d[d["tram"] == tram]
    if cps is not None:
        d = d[d["cp"].isin(cps)]
    s = d.groupby("data", as_index=False)["kwh"].sum()
    return s[s["kwh"] > 0]

def fila_creix(f):
    return {**{a: f[a] for a in ANYS}, **{f"se_{a}": f[f"se_{a}"] for a in ANYS},
            "creix_21_24": f["creix_21_24"], "se_creix": f["se_creix"]}

proves = {}
ciutat_res = ciutat_brut[ciutat_brut["kwh"] > 0]          # ciutat: només dies complets (22d)

# --- T0. Base ---
proves["T0 base"] = fila_creix(chi_per_any(taula_de(ciutat_res))[0])

# --- T1. Amb els dies d'onada ---
proves["T1 amb els dies d'onada"] = fila_creix(chi_per_any(taula_de(ciutat_res, onades=True))[0])

# --- T2. Tc comú diferent (18 a 23 °C): si el creixement depèn de Tc, varia aquí ---
for Tc_ in [18, 23]:
    proves[f"T2 Tc = {Tc_} °C"] = fila_creix(chi_per_any(taula_de(ciutat_res, Tc=Tc_))[0])

# --- T3. Terme quadràtic comú a tots els anys ---
t_ = taula_de(ciutat_res)
f_, b_, _ = chi_per_any(t_, extres=pd.DataFrame({"cdd_quad": t_["cdd0"] ** 2}))
proves["T3 terme quadràtic comú"] = fila_creix(f_)
print(f"T3 · terme quadràtic: {100*b_['cdd_quad']:+.3f} %/K² (positiu = la resposta es corba cap amunt)")

# --- T4. Temperatura de les estacions del Meteocat ---
proves["T4 estacions del Meteocat"] = fila_creix(chi_per_any(taula_de(ciutat_res, Tc=TC_EST, tsrc=temp_est_ciutat))[0])

# --- T5. Només els CP sense cap salt ---
sense_salts = [cp for cp in serie_res if cp not in amb_salts]
proves[f"T5 només CP sense salts ({len(sense_salts)})"] = fila_creix(chi_per_any(taula_de(serie_df(sector_triat, cps=sense_salts)))[0])

# --- T6. Resposta ABSOLUTA (MWh/dia per K): si només baixés el consum de base, aquí no creixeria ---
t_ = taula_de(ciutat_res)
X_ = disseny_v07(t_, LAGS, per_any=True)
b_abs, *_ = np.linalg.lstsq(X_.values, t_["kwh"].values / 1e3, rcond=None)
b_abs = pd.Series(b_abs, index=X_.columns)
abs_any = {a: b_abs[[f"cdd{k}_{a}" for k in range(LAGS + 1)]].sum() for a in ANYS if f"cdd0_{a}" in b_abs.index}
nivell_base = t_[t_["cdd0"] == 0].groupby(t_["data"].dt.year)["kwh"].mean() / 1e3
print("\nT6 · resposta absoluta (MWh/dia per K) i nivell dels dies sense calor (MWh/dia):")
for a in abs_any:
    print(f"  {a}: {abs_any[a]:6.0f}   ·   nivell base {nivell_base.get(a, np.nan):6.0f}")

# --- T7. Humitat: punt de rosada (control i interacció) i temperatura aparent ---
def baixa_humitat():
    url = ("https://archive-api.open-meteo.com/v1/archive"
           f"?latitude={LAT}&longitude={LON}&start_date={min(ANYS)}-01-01&end_date={max(ANYS)}-12-31"
           "&daily=dew_point_2m_mean,apparent_temperature_mean&timezone=Europe%2FMadrid")
    h = pd.DataFrame(requests.get(url, timeout=120).json()["daily"]).rename(columns={"time": "data"})
    h["data"] = pd.to_datetime(h["data"])
    return h
hum = de_la_cau("humitat_era5.csv", baixa_humitat)
td_mitja = hum.loc[hum["data"].dt.month.isin(MESOS_ESTIU), "dew_point_2m_mean"].mean()
t_ = taula_de(ciutat_res)
td = t_[["data"]].merge(hum[["data", "dew_point_2m_mean"]], on="data", how="left")["dew_point_2m_mean"].fillna(td_mitja).values - td_mitja
f_, b_, c_ = chi_per_any(t_, extres=pd.DataFrame({"td": td, "cdd0_x_td": t_["cdd0"].values * td}))
proves["T7a control del punt de rosada"] = fila_creix(f_)
print(f"\nT7 · interacció CDD × punt de rosada: {100*b_['cdd0_x_td']:+.3f} ± {100*np.sqrt(c_.loc['cdd0_x_td','cdd0_x_td']):.3f} %/K per grau de rosada")
ta = temp[["data", "onada"]].merge(hum[["data", "apparent_temperature_mean"]].rename(columns={"apparent_temperature_mean": "tmean"}), on="data")
millor = min(np.arange(18, 28, 0.5), key=lambda Tc_: np.mean(ajusta_cl(taula_de(ciutat_res, Tc=Tc_, tsrc=ta), disseny_v07(taula_de(ciutat_res, Tc=Tc_, tsrc=ta), LAGS, per_any=True))[3] ** 2))
proves[f"T7b temperatura aparent (Tc = {millor} °C)"] = fila_creix(chi_per_any(taula_de(ciutat_res, Tc=millor, tsrc=ta))[0])

# --- T8. Preu de l'electricitat (Red Eléctrica: PVPC), interacció amb els CDD ---
def baixa_preus_ree():
    files = []
    for a in ANYS:
        for m in MESOS_ESTIU:
            ini = pd.Timestamp(a, m, 1); fi = ini + pd.offsets.MonthEnd(1)
            url = ("https://apidatos.ree.es/es/datos/mercados/precios-mercados-tiempo-real"
                   f"?start_date={ini:%Y-%m-%d}T00:00&end_date={fi:%Y-%m-%d}T23:59&time_trunc=hour")
            for sr in requests.get(url, timeout=120).json()["included"]:
                for v in sr["attributes"]["values"]:
                    files.append({"data": v["datetime"][:10], "serie": sr["attributes"]["title"], "eur_mwh": v["value"]})
    p = pd.DataFrame(files); p["data"] = pd.to_datetime(p["data"])
    return p.pivot_table(index="data", columns="serie", values="eur_mwh", aggfunc="mean").reset_index()
preus = de_la_cau("preus_electricitat.csv", baixa_preus_ree)
ps = preus[preus["data"].dt.month.isin(MESOS_ESTIU)]
print("\nT8 · PVPC mitjà d'estiu (€/MWh):", ps.groupby(ps["data"].dt.year)["PVPC"].mean().round(0).to_dict())
mu, sd = ps["PVPC"].mean(), ps["PVPC"].std()
def ext_preu(t):
    z = ((t[["data"]].merge(preus[["data", "PVPC"]], on="data", how="left")["PVPC"] - mu) / sd).fillna(0).values
    return pd.DataFrame({f"cdd{k}_x_p": t[f"cdd{k}"].values * z for k in range(LAGS + 1)})
for sec_, nom in [(sector_triat, "residencial"), ("Serveis", "serveis")]:
    t_ = taula_de(ciutat_res if sec_ == sector_triat else serie_df(sec_))
    f_, b_, c_ = chi_per_any(t_, extres=ext_preu(t_))
    cc = [f"cdd{k}_x_p" for k in range(LAGS + 1)]
    print(f"    {nom}: efecte del preu {100*b_[cc].sum():+.2f} ± {100*np.sqrt(c_.loc[cc, cc].values.sum()):.2f} %/K per cada {sd:.0f} €/MWh")
    if sec_ == sector_triat:
        proves["T8 control del preu (PVPC)"] = fila_creix(f_)

# --- T9. Per sector (tots els sectors, sense la franja 'No consta' ni dies amb 0 kWh) ---
print("\nT9 · χ⁰ per sector i estiu (%/K):")
per_sector = {}
for sec_ in ["Residencial", "Serveis", "Indústria", None]:
    try:
        s_ = ciutat_res if sec_ == sector_triat else serie_df(sec_)
        per_sector[sec_ or "Tots"] = fila_creix(chi_per_any(taula_de(s_))[0])
    except Exception as e:
        print(f"  {sec_}: {e}")
per_sector = pd.DataFrame(per_sector).T
print(per_sector[ANYS + ["creix_21_24", "se_creix"]].astype(float).round(2).to_string())

# --- T10. Per franja horària (residencial) ---
print("\nT10 · χ⁰ residencial per franja i estiu (%/K):")
per_franja = pd.DataFrame({tr: fila_creix(chi_per_any(taula_de(serie_df(sector_triat, tram=tr)))[0])
                           for tr in sorted(t for t in df["tram"].dropna().unique() if "consta" not in str(t).lower())}).T
print(per_franja[ANYS + ["creix_21_24", "se_creix"]].astype(float).round(2).to_string())

# --- T11. Per mesos: juny + setembre contra juliol + agost ---
MESOS_ORIG = MESOS_ESTIU
for mesos, nom in [([6, 9], "juny + setembre"), ([7, 8], "juliol + agost")]:
    MESOS_ESTIU = mesos
    try:
        proves[f"T11 només {nom}"] = fila_creix(chi_per_any(taula_de(ciutat_res))[0])
    finally:
        MESOS_ESTIU = MESOS_ORIG

# --- T12. Turisme: el creixement per CP és més gran on hi ha més pisos turístics? ---
try:
    hut = llegeix_csv_flexible(f"{CARPETA_LOCAL}/hut_comunicacio_opendata.csv").dropna(subset=["LONGITUD_X", "LATITUD_Y"])
    pts_h = gpd.GeoDataFrame(hut, geometry=gpd.points_from_xy(hut["LONGITUD_X"], hut["LATITUD_Y"]), crs="EPSG:4326").to_crs(pol.crs)
    hut_cp = gpd.sjoin(pts_h, pol[["cp", "geometry"]], predicate="within").groupby("cp").agg(n_hut=("cp", "size"), places=("NUMERO_PLACES", "sum"))
    Ht = P[["creix", "te_salts"]].join(hut_cp).fillna({"n_hut": 0, "places": 0})
    Ht["places_per_MWh"] = Ht["places"] / (diari_brut.groupby("cp")["kwh"].mean() / 1e3)
    rho, pv = stats.spearmanr(Ht["places_per_MWh"], Ht["creix"], nan_policy="omit")
    print(f"\nT12 · {int(Ht['n_hut'].sum())} pisos turístics assignats · creixement vs places per MWh/dia: ρ = {rho:+.2f} (p = {pv:.3f})")
    Ht["terc"] = pd.qcut(Ht["places_per_MWh"], 3, labels=["pocs", "mitjans", "molts"])
    print("    Mediana del creixement per tercils:", Ht.groupby("terc", observed=True)["creix"].median().round(2).to_dict())
except Exception as e:
    print(f"\nT12 · [!] Sense dades de pisos turístics ({e}). Baixa el CSV d'Open Data BCN (habitatges-us-turistic) a CARPETA_LOCAL")

# --- T13. Corba de resposta de cada estiu, sense forma imposada (bins d'1 °C) ---
t3 = ciutat_res.merge(temp, on="data")
t3 = t3[t3["data"].dt.month.isin(MESOS_ESTIU) & t3["data"].dt.year.isin(ANYS)].reset_index(drop=True)
t3["tm3"] = t3["data"].map(temp.set_index("data")["tmean"].rolling(3).mean())
t3["bin"] = np.clip(np.floor(t3["tm3"]), 20, 28).astype(int)
cat_ = lambda v, c: pd.Categorical(v, categories=c)
Xb = [pd.get_dummies(pd.DataFrame({"setm": cat_(t3["data"].dt.isocalendar().week.astype(int).values, list(range(1, 54))),
                                   "dia": cat_(t3["data"].dt.dayofweek, list(range(7))),
                                   "fest": cat_(t3["data"].isin(pd.to_datetime(list(festius.keys()))), [False, True])}), drop_first=True).astype(float)]
BINS = list(range(21, 29))
for a in ANYS:
    m = (t3["data"].dt.year == a).values
    Xb.append(pd.DataFrame({f"any{a}": m.astype(float)}))
    Xb.append(pd.DataFrame({f"b{bb}_{a}": (m & (t3["bin"] == bb).values).astype(float) for bb in BINS}))
Xb = pd.concat(Xb, axis=1); Xb = Xb.loc[:, (Xb != 0).any()]
bb_, _, cb_, _ = ajusta_cl(t3, Xb)
fig, ax = plt.subplots(figsize=(8, 5))
corbes = {}
for a in ANYS:
    xs = [20] + [bb for bb in BINS if f"b{bb}_{a}" in bb_.index]
    ys = [0] + [100 * (np.exp(bb_[f"b{bb}_{a}"]) - 1) for bb in BINS if f"b{bb}_{a}" in bb_.index]
    es = [0] + [200 * np.sqrt(cb_.loc[f"b{bb}_{a}", f"b{bb}_{a}"]) for bb in BINS if f"b{bb}_{a}" in bb_.index]
    corbes[a] = dict(zip(xs, np.round(ys, 1)))
    ax.errorbar(np.array(xs) + 0.06 * (a - 2023), ys, es, fmt="o-", capsize=2, label=str(a))
ax.set_xlabel("Temperatura mitjana dels 3 últims dies, ERA5 (°C; 20 = ≤20, 28 = ≥28)")
ax.set_ylabel("Consum residencial respecte de ≤20 °C (%)")
ax.set_title("T13 · Corba de resposta de cada estiu, sense forma imposada (±2 e.e.)")
ax.legend(); plt.tight_layout(); plt.savefig("corbes_per_estiu_v12.png", dpi=200); plt.show()
print("T13 · consum respecte de ≤20 °C (%):"); print(pd.DataFrame(corbes).T.to_string())

# --- T14. Residus dels dies d'onada per estiu i T15. Cobertura de primavera ---
t_all = taula_de(ciutat_res, onades=True)
nrm = t_all[~t_all["onada"]].reset_index(drop=True)
X_n = disseny_v07(nrm, LAGS, per_any=True)
b_n, *_ = np.linalg.lstsq(X_n.values, np.log(nrm["kwh"].values), rcond=None)
cal = t_all[t_all["onada"]].reset_index(drop=True)
if len(cal):
    X_c = disseny_v07(cal, LAGS, per_any=True).reindex(columns=X_n.columns, fill_value=0.0)
    cal["residu"] = 100 * (np.log(cal["kwh"].values) - X_c.values @ b_n)
    print("\nT14 · residu mitjà dels dies d'onada (%; positiu = més del que prediu la llei lineal):")
    print(cal.groupby(cal["data"].dt.year)["residu"].agg(["mean", "count"]).round(1).rename(columns={"mean": "residu_%", "count": "dies"}).to_string())
prim = ciutat_res[ciutat_res["data"].dt.month.isin([4, 5])]
nv = prim.groupby(prim["data"].dt.year)["kwh"].mean()
print("\nT15 · consum residencial d'abril-maig (MWh/dia) i canvi respecte al primer any:",
      {a: f"{v/1e3:.0f} ({100*(v/nv.iloc[0]-1):+.1f} %)" for a, v in nv.items()})
```

```python
# ============ 22g. On creix χ⁰? ============
# Creixement de cada CP (mitjana 2023-24 menys mitjana 2021-22) contra les seves característiques.
# Si l'aire condicionat s'instal·la on hi ha renda, el creixement hauria d'anar amb la renda;
# si el que creix és la calor, hauria d'anar amb la illa de calor.
G = P[["creix", "te_salts", "dLST"]].copy()
G["chi_2122"] = P[[2021, 2022]].mean(axis=1)
if "renda_cp" in globals():
    G = G.join(renda_cp.rename("renda"))
if "qf" in globals():
    G = G.join(np.log(qf["qf_no_residencial"]).rename("ln_qf"))
G = G.join(np.log(diari_brut.groupby("cp")["kwh"].mean()).rename("ln_consum"))
G = G[~G["te_salts"]]   # només CP sense salts: el creixement és net
print(f"Creixement de χ⁰ per CP (sense salts, {len(G)} CP): mediana {G['creix'].median():+.2f} %/K, "
      f"P10-P90 {G['creix'].quantile(0.1):+.2f} a {G['creix'].quantile(0.9):+.2f}")
for col in [c for c in ["dLST", "renda", "ln_qf", "chi_2122", "ln_consum"] if c in G.columns]:
    d = G.dropna(subset=[col, "creix"])
    rho, pv = stats.spearmanr(d[col], d["creix"])
    print(f"  creixement vs {col:10s}: ρ = {rho:+.2f} (p = {pv:.3f}, {len(d)} CP)")
print("  (compte: si el creixement va amb chi_2122 negatiu, pot ser regressió a la mitjana)")

if "pol" in globals():
    gg = pol.merge(P[["creix", "te_salts"]].reset_index(), on="cp").to_crs(epsg=3857)
    fig, ax = plt.subplots(figsize=(9, 9))
    gg.plot(ax=ax, column="creix", cmap="RdBu_r", vmin=-gg["creix"].abs().quantile(0.9), vmax=gg["creix"].abs().quantile(0.9),
            edgecolor="white", linewidth=0.8, legend=True, legend_kwds={"label": "Δχ⁰ (2023-24 menys 2021-22, %/K)", "shrink": 0.6})
    for _, r in gg.iterrows():
        p_ = r.geometry.representative_point()
        ax.annotate(r["cp"][-2:] + ("*" if r["te_salts"] else ""), (p_.x, p_.y), ha="center", va="center", fontsize=8)
    ax.set_axis_off(); ax.set_title("On ha crescut la susceptibilitat? (*: CP amb salts, corregits amb trams)")
    plt.tight_layout(); plt.savefig("mapa_creixement_chi_v12.png", dpi=200); plt.show()
```

```python
# ============ 22i. Fitxa de control de tot el notebook ============
# Un sol lloc amb els números clau de cada secció, per fer-ne una captura.
def g_(nom, f=lambda x: x):
    try:
        return f(globals()[nom])
    except Exception:
        return "—"
fitxa = {
    "3 · CP analitzats": g_("diari", lambda d: d["cp"].nunique()),
    "3 · CP petits exclosos": g_("petits", lambda x: ", ".join(x) if len(x) else "cap"),
    "5 · dies exclosos (ciutat)": g_("dies_dolents", len),
    "11 · dies atípics exclosos": g_("dies_atipics", len),
    "12 · salts de nivell": g_("salts_df", len),
    "6 · dies d'onada": g_("onades", len),
    "7 · Tc (ERA5 / estacions, °C)": f"{g_('TC')} / {g_('TC_EST')}",
    "8 · χ⁰ ciutat, tots els anys (%/K)": g_("chi_c", lambda c: round(100 * c.sum(), 2)),
    "10 · residu d'onada de la ciutat (%)": g_("rc", lambda r: round(100 * r.mean(), 1)),
    "22b · salts per tipus": g_("classif", lambda c: c["tipus"].value_counts().to_dict()),
    "22b · unitats de conservació": g_("unitats", lambda u: [len(x) for x in u]),
    "22d · creixement 2021→2024 (%/K)": f"{base['creix_21_24']:+.2f} ± {base['se_creix']:.2f}",
    "22e · condicions 1 i 2": f"{'OK' if cond1 else 'NO'} / {'OK' if cond2 else 'NO'}",
}
for k, v_ in fitxa.items():
    print(f"{k:42s} {v_}")
```

## 22j. Calor antropogènica total i bucle G

**Per què:** la secció A4 fa servir només la calor dels edificis **no** residencials, perquè allà intenta *explicar* la χ residencial i posar-hi el consum residencial seria circular. Per al **bucle de retroalimentació** cal **tota** la calor antropogènica elèctrica i, sobretot, **quant creix per cada grau**:

$$G = \frac{\partial Q_F/\partial T}{h_\text{eff}}, \qquad \chi_\text{eff} = \frac{\chi_0}{1-G}$$

**Què fa:** per a cada sector i estiu calcula la resposta **absoluta** (MWh/dia per K), la passa a W m⁻² K⁻¹ dividint per l'àrea dels CP, i calcula $G$ per a tres valors de $h_\text{eff}$ (**10, 25 i 50 W m⁻² K⁻¹: rang orientatiu, pendent de verificar a la bibliografia**). Després fa el mateix per CP (estius 2024-2025) i en dibuixa el mapa.

**Compte:** només és calor **elèctrica** (hi falten gas, trànsit i metabolisme). Un aire condicionat expulsa al carrer $E(1+\mathrm{COP})$ mentre funciona, però el balanç net diari és $E$: la $G$ local i instantània (de nit) pot ser fins a $1+\mathrm{COP} \approx 4$ vegades més gran.

```python
# ============ 22j. Calor antropogènica total i bucle G ============
H_EFF = [10, 25, 50]      # W m-2 K-1: rang orientatiu, PER VERIFICAR a la bibliografia
COP = 3.0                 # coeficient d'eficiència típic d'un aire condicionat (per a la cota de la G instantània)
area_tot = area_cp.reindex(list(serie_res)).sum()          # m², CP analitzats
kwhdia_a_wm2 = lambda x: x * 1000 / 24 / area_tot           # kWh/dia (ciutat) -> W/m²

def resposta_absoluta(s, cp=None):
    # Resposta absoluta per estiu (kWh/dia per K): mateix model, però en nivells i no en logaritme
    t_ = taula_de(s, cp=cp)
    X_ = disseny_v07(t_, LAGS, per_any=True, amb_trams="tram" in t_.columns)
    b_, _, c_, _ = ajusta_cl(t_, X_, y=t_["kwh"].values)
    out = {}
    for a in ANYS:
        cc = [f"cdd{k}_{a}" for k in range(LAGS + 1)]
        if all(x in b_.index for x in cc):
            out[a] = (b_[cc].sum(), np.sqrt(c_.loc[cc, cc].values.sum()))
    return out

files = []
for sec_, nom in [(sector_triat, "Residencial"), ("Serveis", "Serveis"), ("Indústria", "Indústria"), (None, "Tots els sectors")]:
    s_ = ciutat_res if sec_ == sector_triat else serie_df(sec_)
    qf_estiu = kwhdia_a_wm2(s_.loc[s_["data"].dt.month.isin(MESOS_ESTIU), "kwh"].mean())
    for a, (v, e) in resposta_absoluta(s_).items():
        files.append({"sector": nom, "any": a, "Q_F_estiu_Wm2": qf_estiu,
                      "dE_dT_MWh_dia_K": v / 1e3, "dQF_dT_Wm2K": kwhdia_a_wm2(v), "se_dQF_dT": kwhdia_a_wm2(e),
                      **{f"G_h{h}": kwhdia_a_wm2(v) / h for h in H_EFF}})
qf_taula = pd.DataFrame(files)
print(f"Àrea dels {len(serie_res)} CP: {area_tot/1e6:.1f} km²")
print(qf_taula.pivot(index="sector", columns="any", values="dQF_dT_Wm2K").round(3).to_string())
print("\nQ_F d'estiu (W/m²):", qf_taula.groupby("sector")["Q_F_estiu_Wm2"].first().round(2).to_dict())
tot_ = qf_taula[qf_taula["sector"] == "Tots els sectors"].set_index("any")
print("\nG de la ciutat (tots els sectors), segons h_eff:")
print(tot_[[f"G_h{h}" for h in H_EFF]].round(4).to_string())
print(f"Cota de la G local i instantània (×(1+COP) = ×{1+COP:.0f}) el 2024: "
      f"{(1+COP)*tot_.loc[2024, 'G_h50']:.3f} a {(1+COP)*tot_.loc[2024, 'G_h10']:.3f}")
qf_taula.to_csv("calor_antropogenica_G_v12.csv", index=False)

# --- Per CP: ∂Q_F/∂T de tots els sectors, mitjana 2023-2024 (v0.11: el 2025 té la recodificació de CP) (amb trams) ---
files = []
for cp in serie_res:
    s_ = serie_df(cps=[cp])
    try:
        r_ = resposta_absoluta(s_, cp=cp)
        v = np.mean([r_[a][0] for a in [2023, 2024] if a in r_])
        files.append({"cp": cp, "dQF_dT_Wm2K": v * 1000 / 24 / area_cp[cp],
                      "QF_estiu_Wm2": s_.loc[s_["data"].dt.month.isin(MESOS_ESTIU), "kwh"].mean() * 1000 / 24 / area_cp[cp]})
    except Exception as e:
        print(f"  [!] {cp}: {e}")
qf_cp = pd.DataFrame(files).set_index("cp")
qf_cp["G_h25"] = qf_cp["dQF_dT_Wm2K"] / 25
print("\nCP amb més ∂Q_F/∂T (W m⁻² K⁻¹, 2023-24):")
print(qf_cp.sort_values("dQF_dT_Wm2K", ascending=False).head(8).round(3).to_string())
if "pol" in globals():
    g3 = pol.merge(qf_cp.reset_index(), on="cp").to_crs(epsg=3857)
    fig, ax = plt.subplots(figsize=(9, 9))
    g3.plot(ax=ax, column="dQF_dT_Wm2K", cmap="OrRd", legend=True, edgecolor="white", linewidth=0.8,
            legend_kwds={"label": "∂Q_F/∂T (W m⁻² K⁻¹), tots els sectors, 2023-24", "shrink": 0.6})
    for _, r in g3.iterrows():
        p_ = r.geometry.representative_point(); ax.annotate(r["cp"][-2:], (p_.x, p_.y), ha="center", fontsize=8)
    ax.set_axis_off(); ax.set_title("Calor antropogènica addicional per cada grau de calor")
    plt.tight_layout(); plt.savefig("mapa_dQF_dT_v12.png", dpi=200); plt.show()
```

## 22k. Model de llindars: les corbes de cada estiu tenen la mateixa forma?

**Hipòtesi:** cada llar encén l'aire a partir d'un llindar propi. La resposta de la ciutat és $\Delta E(T) \propto \phi\,F(T)$, amb $\phi$ la fracció de llars amb aire i $F(T)$ la fracció de llindars superats. Si només creix $\phi$, **les corbes de cada estiu són la mateixa forma multiplicada per un factor** $\lambda_a = \phi_a/\phi_{2021}$.

**Què fa:** amb les corbes de la T13 (22f), (1) comprova si per sota de 24 °C les corbes són iguals entre estius (χ² respecte de la mitjana ponderada), i (2) per sobre de 24 °C ajusta $y_{a,b} = \lambda_a\,g_b$ (forma comuna $g_b$, un factor per estiu) i mira si l'ajust és acceptable (χ²/graus de llibertat ≈ 1).

```python
# ============ 22k. Model de llindars ============
from scipy import stats as _st
Y = pd.DataFrame({a: {bb: 100 * bb_[f"b{bb}_{a}"] for bb in BINS if f"b{bb}_{a}" in bb_.index} for a in ANYS})   # log·100 ≈ %
S = pd.DataFrame({a: {bb: 100 * np.sqrt(cb_.loc[f"b{bb}_{a}", f"b{bb}_{a}"]) for bb in BINS if f"b{bb}_{a}" in bb_.index} for a in ANYS})

# (1) Per sota de 24 °C: són iguals?
baix = [bb for bb in BINS if bb < 24]
chi2_b, dof_b = 0.0, 0
for bb in baix:
    y, s = Y.loc[bb].dropna(), S.loc[bb].dropna()
    w = 1 / s ** 2; m = (w * y).sum() / w.sum()
    chi2_b += (((y - m) / s) ** 2).sum(); dof_b += len(y) - 1
print(f"Per sota de 24 °C, corbes iguals entre estius: χ² = {chi2_b:.1f} amb {dof_b} g.ll. (p = {1-_st.chi2.cdf(chi2_b, dof_b):.3f})")

# (2) Per sobre de 24 °C: forma comuna × factor per estiu (mínims quadrats ponderats, per alternança)
alt = [bb for bb in BINS if bb >= 24]
y, s = Y.loc[alt], S.loc[alt]
lam = pd.Series(1.0, index=ANYS)
for _ in range(200):
    g = ((y * lam / s ** 2).sum(axis=1) / ((lam ** 2) / s ** 2).sum(axis=1))
    lam_nou = ((y.mul(g, axis=0) / s ** 2).sum() / ((s ** -2).mul(g ** 2, axis=0)).sum())
    lam_nou = lam_nou / lam_nou[2021]                       # normalitzat al 2021
    if np.allclose(lam_nou, lam, atol=1e-8): break
    lam = lam_nou
g = ((y * lam / s ** 2).sum(axis=1) / ((lam ** 2) / s ** 2).sum(axis=1))
pred = pd.DataFrame(np.outer(g, lam), index=alt, columns=ANYS)
chi2_a = ((((y - pred) / s) ** 2).sum()).sum()
dof_a = int(y.notna().sum().sum()) - len(alt) - (len(ANYS) - 1)
se_lam = 1 / np.sqrt(((s ** -2).mul(g ** 2, axis=0)).sum())
print(f"Per sobre de 24 °C, forma comuna × factor: χ² = {chi2_a:.1f} amb {dof_a} g.ll. (χ²/g.ll. = {chi2_a/dof_a:.2f}; p = {1-_st.chi2.cdf(chi2_a, dof_a):.3f})")
print("Factor de cada estiu λ (2021 = 1), proporcional a la fracció de llars amb aire si la hipòtesi és certa:")
print(pd.DataFrame({"lambda": lam.round(2), "se": se_lam.round(2)}).T.to_string())
lam.to_csv("factor_llindars_v12.csv")
```

## 22l. Comprovació per correlacions (fluctuació-resposta)

**Idea:** si el sistema és lineal i el forçament és la temperatura, la funció de resposta es pot obtenir **també** de les correlacions entre fluctuacions: $C_{ET}(k) = \sum_j \chi(j)\,C_{TT}(k-j)$, i per tant $\chi = C_{TT}^{-1} C_{ET}$ (un estimador de tipus Yule-Walker). Les fluctuacions són els residus després de treure els efectes fixos. Si els dos estimadors coincideixen, el model lineal és consistent; si no, hi ha no linealitats o variables amagades. **És una comprovació de consistència, no un teorema de fluctuació-dissipació en sentit estricte.**

```python
# ============ 22l. Comprovació per correlacions ============
t_ = taula_de(ciutat_res).sort_values("data").reset_index(drop=True)
Xfe = disseny_v07(t_, LAGS).drop(columns=[f"cdd{k}" for k in range(LAGS + 1)])
proj = lambda v: v - Xfe.values @ np.linalg.lstsq(Xfe.values, v, rcond=None)[0]
rE = pd.Series(proj(np.log(t_["kwh"].values)), index=t_["data"])
rT = pd.Series(proj(t_["cdd0"].values.astype(float)), index=t_["data"])
def cov_desfasada(a, b, k):
    # Cov(a(t), b(t-k)) només amb parells de dies que existeixen tots dos
    bb = b.copy(); bb.index = bb.index + pd.Timedelta(days=k)
    j = pd.concat([a, bb], axis=1, join="inner").dropna()
    return (j.iloc[:, 0] * j.iloc[:, 1]).mean()
CTT = np.array([[cov_desfasada(rT, rT, abs(k - j)) for j in range(LAGS + 1)] for k in range(LAGS + 1)])
CET = np.array([cov_desfasada(rE, rT, k) for k in range(LAGS + 1)])
chi_corr = np.linalg.solve(CTT, CET)
b_reg, _, _, _ = ajusta_cl(t_, disseny_v07(t_, LAGS))
chi_reg = np.array([b_reg[f"cdd{k}"] for k in range(LAGS + 1)])
print("χ(k) (%/K):   regressió   correlacions")
for k in range(LAGS + 1):
    print(f"  k = {k}:      {100*chi_reg[k]:6.2f}      {100*chi_corr[k]:6.2f}")
print(f"  χ⁰:         {100*chi_reg.sum():6.2f}      {100*chi_corr.sum():6.2f}")
```

## 23. Ritmes dia–nit de la resposta a la calor (v0.10, revisada a la v0.11)

**Per què:** la secció 22 diu que la susceptibilitat creix, sobretot al vespre i a la nit. L'article 1 ara gira al voltant
d'una sola frase: *Barcelona està desplaçant la seva resposta elèctrica a la calor cap al vespre i la nit, a les llars i als

Totes les dades noves es baixen soles; si la baixada falla, es llegeixen de `CARPETA_LOCAL` (`temp_horaria.csv`, `hotels.csv`).
Si tampoc hi són, la secció continua amb una **aproximació** (avisa en pantalla) o salta aquella prova.

```python
# ============ 23a. Dades: franges horàries i temperatura hora a hora ============
import re, requests

# --- Paràmetres de la secció 23 (també es guarden a l'especificació congelada, 23i) ---
TC_DIA = 27.0      # llindar per a la temperatura de DIA (mitjana de 12 a 18 h), °C
TC_NIT = 20.0      # llindar per a la temperatura de NIT (mitjana de 0 a 6 h), °C
ANY_REF, ANY_FI = 2021, 2024   # anys del creixement principal (com a la secció 22)

# --- Les quatre franges del CSV, amb un nom curt ("00-06", ...) ---
FRANGES = sorted(t for t in df["tram"].astype(str).unique() if not re.search("consta", t, re.I))
def nom_franja(tr):
    h = re.findall(r"(\d{1,2}):", str(tr))          # "De 00:00:00 a 05:59:59 h" -> ["00", "05"]
    return f"{int(h[0]):02d}-{int(h[0]) + 6:02d}" if h else str(tr)
NOM = {tr: nom_franja(tr) for tr in FRANGES}
print("Franges:", list(NOM.values()))

# --- Sectors que analitzem ---
sector_serveis = [s for s in df["sector"].unique() if "serv" in str(s).lower()]
SECTORS = {"Residencial": sector_triat}
if sector_serveis:
    SECTORS["Serveis"] = sector_serveis[0]
print("Sectors:", SECTORS)

# --- Temperatura, radiació i capa límit, hora a hora (ERA5 via Open-Meteo, hora local) ---
def baixa_horaria(variables):
    url = ("https://archive-api.open-meteo.com/v1/archive"
           f"?latitude={LAT}&longitude={LON}&start_date={min(ANYS)}-01-01&end_date={max(ANYS)}-12-31"
           f"&hourly={variables}&timezone=Europe%2FMadrid")
    r = requests.get(url, timeout=180); r.raise_for_status()
    h = pd.DataFrame(r.json()["hourly"]).rename(columns={"time": "hora"})
    h["hora"] = pd.to_datetime(h["hora"])
    return h

APROX_HORARIA = False
def baixa_horaria_completa():
    try:
        return baixa_horaria("temperature_2m,shortwave_radiation,boundary_layer_height")
    except Exception as e:
        print(f"[!] Sense alçada de capa límit ({e}); baixo només temperatura i radiació")
        return baixa_horaria("temperature_2m,shortwave_radiation")
try:
    hor = de_la_cau("temp_horaria.csv", baixa_horaria_completa, col_data="hora")
except Exception as e:
    print(f"[!] Temperatura horària no baixada ({e}); provo el fitxer local")
    try:
        hor = pd.read_csv(f"{CARPETA_LOCAL}/temp_horaria.csv", parse_dates=["hora"])
    except Exception as e2:
        hor = None
        print(f"[!] Tampoc hi ha {CARPETA_LOCAL}/temp_horaria.csv ({e2}).")

if hor is not None:
    hor["data"] = hor["hora"].dt.normalize()
    hor["franja"] = (hor["hora"].dt.hour // 6) * 6                 # 0, 6, 12, 18
    tf = hor.pivot_table(index="data", columns="franja", values="temperature_2m", aggfunc="mean")
    tf.columns = [f"t_{c:02d}" for c in tf.columns]
    tf["tmin_h"] = hor.groupby("data")["temperature_2m"].min()
    tf["tmax_h"] = hor.groupby("data")["temperature_2m"].max()
    tf["rad"] = hor.groupby("data")["shortwave_radiation"].sum() / 1000          # kWh/m² al dia
    if "boundary_layer_height" in hor.columns:
        blh = hor.pivot_table(index="data", columns="franja", values="boundary_layer_height", aggfunc="mean")
        blh.columns = [f"blh_{c:02d}" for c in blh.columns]
        tf = tf.join(blh)
    tf = tf.reset_index()
else:
    # APROXIMACIÓ: sense dades horàries, fem servir la màxima com a temperatura de dia
    # i 2·mitjana − màxima com a estimació de la mínima. Avisem a tot arreu.
    APROX_HORARIA = True
    print("[!] APROXIMACIÓ: temperatura de dia = Tmax; de nit = 2·Tmean − Tmax. Els resultats 23c són provisionals.")
    tf = temp[["data", "tmax", "tmean"]].copy()
    tf["t_12"] = tf["tmax"] - 1.0
    tf["t_00"] = 2 * tf["tmean"] - tf["tmax"] + 1.0
    tf["t_06"], tf["t_18"] = tf["tmean"], tf["tmean"]
    tf["tmin_h"], tf["tmax_h"] = 2 * tf["tmean"] - tf["tmax"], tf["tmax"]
    tf = tf.drop(columns=["tmax", "tmean"])

# Temperatura de DIA = mitjana de 12 a 18 h; de NIT = mitjana de 0 a 6 h (la nit que acaba aquell dia)
tf = tf.sort_values("data").reset_index(drop=True)
tf["t_dia"], tf["t_nit"] = tf["t_12"], tf["t_00"]
for nom_, col, tc in [("dia", "t_dia", TC_DIA), ("nit", "t_nit", TC_NIT)]:
    tf[f"cdd_{nom_}0"] = np.maximum(tf[col] - tc, 0)
    tf[f"cdd_{nom_}1"] = tf[f"cdd_{nom_}0"].shift(1)                 # el dia abans (memòria d'un dia)
tf.to_csv("temp_franges_v12.csv", index=False)

# --- v0.11 · ALERTA DE DADES 2: règim de les franges horàries ---
# Entre el 2021 i el 2023, el pes de cada franja és molt estable. Des del 2024, alguns mesos una part del consum
# de 18-24 h apareix a 06-12 h (canvi a l'origen de les dades, no físic). Marquem cada dia amb un indicador:
# anomalia de la ràtio 18-24 / (18-24 + 06-12) respecte de la mateixa setmana del 2021-2023.
P_ = (df[(df["sector"] == sector_triat) & df["tram"].isin(FRANGES)]
      .pivot_table(index="data", columns="tram", values="kwh", aggfunc="sum")[FRANGES])
rat = P_[FRANGES[3]] / (P_[FRANGES[3]] + P_[FRANGES[1]])
ref_ = rat[rat.index.year <= 2023]
clim = ref_.groupby(ref_.index.isocalendar().week.values).median()
anom = 100 * (rat - clim.reindex(rat.index.isocalendar().week.values).values)
LLINDAR_REGIM = -2.5          # punts de ràtio (×100): per sota, el dia és del règim B
REGIM_B = set(anom.index[anom < LLINDAR_REGIM])
print("Dies del règim B (franges desplaçades) per any:", pd.Series([d.year for d in REGIM_B]).value_counts().sort_index().to_dict())
print("Anomalia mediana per mes (punts):")
print(anom.groupby([anom.index.year, anom.index.month]).median().unstack().round(1).to_string())

def ext_regim(t):
    # Columna extra per a les regressions per franja: 1 si el dia és del règim B
    return pd.DataFrame({"regim_B": t["data"].isin(REGIM_B).astype(float).values})
estiu_tf = tf[tf["data"].dt.month.isin(MESOS_ESTIU)]
print(f"[OK] Temperatura per franges: {len(tf)} dies · estiu: T dia {estiu_tf['t_dia'].mean():.1f} °C, "
      f"T nit {estiu_tf['t_nit'].mean():.1f} °C · correlació dia-nit {estiu_tf['t_dia'].corr(estiu_tf['t_nit']):.2f}")
if "blh_00" in tf.columns:
    print("Alçada mitjana de la capa límit a l'estiu (m):",
          {c[-2:] + " h": round(estiu_tf[c].mean()) for c in tf.columns if c.startswith("blh_")})
```

```python
# ============ 23b. χ⁰ per franja, estiu i sector ============
def serie23(sec, tram=None):
    # Sèrie de ciutat d'un sector (i franja) amb els mateixos CP i només els dies complets de la 22d
    s = serie_df(sec, tram=tram)
    return s[s["data"].isin(dies_complets)]

def resum_anys(taula, extres=None):
    # χ⁰ per estiu i tres mesures del canvi (amb la covariància completa entre anys)
    f, beta, cov = chi_per_any(taula, extres=extres)
    out = {a: f[a] for a in ANYS} | {f"se_{a}": f[f"se_{a}"] for a in ANYS}
    out["creix_21_24"], out["se_21_24"] = creixement(beta, cov, ANY_REF, ANY_FI)
    out["creix_22_24"], out["se_22_24"] = creixement(beta, cov, 2022, ANY_FI)
    # Pendent lineal: combinació de χ⁰ dels cinc anys amb pesos (a − ā) / Σ(a − ā)²
    anys_ok = [a for a in ANYS if all(f"cdd{k}_{a}" in beta.index for k in range(LAGS + 1))]
    aa = np.array(anys_ok, float); pes = (aa - aa.mean()) / ((aa - aa.mean()) ** 2).sum()
    w = pd.Series(0.0, index=beta.index)
    for a, p in zip(anys_ok, pes):
        for k in range(LAGS + 1):
            w[f"cdd{k}_{a}"] = p
    out["pendent"], out["se_pendent"] = 100 * float(w @ beta), 100 * float(np.sqrt(w @ cov @ w))
    return out

R23b = {}
for nom_s, sec in SECTORS.items():
    R23b[(nom_s, "total")] = resum_anys(taula_de(serie23(sec)))          # el total diari no depèn del règim
    for tr in FRANGES:
        t_ = taula_de(serie23(sec, tram=tr))
        R23b[(nom_s, NOM[tr])] = resum_anys(t_, extres=ext_regim(t_))     # v0.11: control del règim de franges
R23b = pd.DataFrame(R23b).T
R23b.index.names = ["sector", "franja"]
R23b.to_csv("chi_franja_sector_v12.csv")

# Robustesa: només els dies del règim A (franges com al 2021-2023)
R23b_A = {}
for nom_s, sec in SECTORS.items():
    for tr in FRANGES:
        t_ = taula_de(serie23(sec, tram=tr))
        t_ = t_[~t_["data"].isin(REGIM_B)].reset_index(drop=True)
        r_ = resum_anys(t_)
        R23b_A[(nom_s, NOM[tr])] = {"creix_21_24": r_["creix_21_24"], "se_21_24": r_["se_21_24"]}
R23b_A = pd.DataFrame(R23b_A).T
print("\nRobustesa · creixement 2021→2024 per franja només amb dies del règim A:")
print(R23b_A.round(2).to_string())
cols = ANYS + ["creix_21_24", "se_21_24", "creix_22_24", "se_22_24", "pendent", "se_pendent"]
print("χ⁰ (%/K) per franja i estiu · canvi 2021→2024, 2022→2024 i pendent (%/K per any)")
print(R23b[cols].round(2).to_string())

fig, axs = plt.subplots(1, len(SECTORS), figsize=(6 * len(SECTORS), 4), sharey=True, squeeze=False)
for ax, nom_s in zip(axs[0], SECTORS):
    for fr in [NOM[t] for t in FRANGES]:
        r = R23b.loc[(nom_s, fr)]
        ax.errorbar(ANYS, [r[a] for a in ANYS], yerr=[r[f"se_{a}"] for a in ANYS], marker="o", capsize=3, label=fr)
    ax.set_title(nom_s); ax.set_xticks(ANYS); ax.grid(alpha=.3)
axs[0][0].set_ylabel("χ⁰ (% per K)"); axs[0][0].legend(title="Franja")
plt.tight_layout(); plt.savefig("chi_franja_sector_v12.png", dpi=150); plt.show()
```

```python
# ============ 23c. Matriu dia–nit: a quina calor respon cada franja? ============
# Substituïm els graus-dia de la temperatura mitjana per dos forçaments:
#   calor de DIA (T de 12 a 18 h per sobre de TC_DIA) i calor de NIT (T de 0 a 6 h per sobre de TC_NIT),
# cadascun amb el mateix dia i el dia abans. La resposta a cada forçament és la suma dels dos coeficients.
DN = ["cdd_dia0", "cdd_dia1", "cdd_nit0", "cdd_nit1"]

def taula_dn(serie):
    t = taula_de(serie).merge(tf[["data"] + DN], on="data", how="left").dropna(subset=DN).reset_index(drop=True)
    return t

def ajust_dn(t, per_any=False):
    X = disseny_v07(t, LAGS).drop(columns=[f"cdd{k}" for k in range(LAGS + 1)], errors="ignore")
    if per_any:
        for a in ANYS:
            m = (t["data"].dt.year == a).values.astype(float)
            for c in DN:
                X[f"{c}_{a}"] = t[c].values * m
    else:
        for c in DN:
            X[c] = t[c].values
    X["regim_B"] = t["data"].isin(REGIM_B).astype(float).values      # v0.11: control del règim de franges
    X = X.loc[:, (X != 0).any(axis=0)]
    beta, _, cov, _ = ajusta_cl(t, X)
    return beta, cov

def suma(beta, cov, cols):
    cols = [c for c in cols if c in beta.index]
    return 100 * beta[cols].sum(), 100 * np.sqrt(cov.loc[cols, cols].values.sum())

files = []
for nom_s, sec in SECTORS.items():
    for tr in [None] + FRANGES:
        t = taula_dn(serie23(sec, tram=tr))
        b, c = ajust_dn(t)
        bd, cd = ajust_dn(t, per_any=True)
        fila = {"sector": nom_s, "franja": "total" if tr is None else NOM[tr]}
        fila["chi_dia"], fila["se_dia"] = suma(b, c, ["cdd_dia0", "cdd_dia1"])
        fila["chi_nit"], fila["se_nit"] = suma(b, c, ["cdd_nit0", "cdd_nit1"])
        # Canvi 2021→2024 de cada resposta
        for nom_, cc in [("dia", ["cdd_dia0", "cdd_dia1"]), ("nit", ["cdd_nit0", "cdd_nit1"])]:
            w = pd.Series(0.0, index=bd.index)
            for x in cc:
                if f"{x}_{ANY_FI}" in w.index: w[f"{x}_{ANY_FI}"] = 1.0
                if f"{x}_{ANY_REF}" in w.index: w[f"{x}_{ANY_REF}"] = -1.0
            fila[f"creix_{nom_}"], fila[f"se_creix_{nom_}"] = 100 * float(w @ bd), 100 * float(np.sqrt(w @ cd @ w))
        files.append(fila)
R23c = pd.DataFrame(files).set_index(["sector", "franja"])
R23c.to_csv("matriu_dia_nit_v12.csv")
print(f"Resposta (%/K) a la calor de DIA (T 12-18 h > {TC_DIA:.0f} °C) i de NIT (T 0-6 h > {TC_NIT:.0f} °C)"
      + ("  [APROXIMACIÓ sense dades horàries]" if APROX_HORARIA else ""))
print(R23c.round(2).to_string())

# Sensibilitat als dos llindars (residencial, franges de nit i de tarda)
print("\nSensibilitat als llindars · residencial (resposta a la nit / resposta al dia):")
_TCD, _TCN = TC_DIA, TC_NIT
for tcd, tcn in [(26, 19), (27, 20), (28, 21), (27, 19), (27, 21)]:
    for nom_, col, tc in [("dia", "t_dia", tcd), ("nit", "t_nit", tcn)]:
        tf[f"cdd_{nom_}0"] = np.maximum(tf[col] - tc, 0); tf[f"cdd_{nom_}1"] = tf[f"cdd_{nom_}0"].shift(1)
    txt = []
    for tr in [FRANGES[0], FRANGES[2]]:
        b, c = ajust_dn(taula_dn(serie23(SECTORS["Residencial"], tram=tr)))
        txt.append(f"{NOM[tr]}: nit {suma(b, c, DN[2:])[0]:.2f} / dia {suma(b, c, DN[:2])[0]:.2f}")
    print(f"  TC_DIA {tcd}, TC_NIT {tcn} ·  " + "  ·  ".join(txt))
for nom_, col, tc in [("dia", "t_dia", _TCD), ("nit", "t_nit", _TCN)]:       # tornem als llindars triats
    tf[f"cdd_{nom_}0"] = np.maximum(tf[col] - tc, 0); tf[f"cdd_{nom_}1"] = tf[f"cdd_{nom_}0"].shift(1)
```

```python
# ============ 23d. Deformació del cicle diari ============
# v0.11: AVÍS. La v0.10 donava un desplaçament del pic de 18 min/K el 2024 que era un ARTEFACTE del canvi de
# règim de franges (alerta de dades 2). Ara controlem el règim i dibuixem només els dies del règim A.
# Cada dia, el consum de les 4 franges dibuixa un cicle. En mesurem:
#   - el PES de cada franja (fracció del consum del dia)
#   - l'HORA DEL PIC del cicle (primer harmònic: fase de Σ E_b·e^{2πi·h_b/24}, amb h_b = 3, 9, 15, 21 h)
CENTRE = {tr: 6 * i + 3 for i, tr in enumerate(FRANGES)}

def cicle(sec):
    d = df[df["cp"].isin(serie_res) & df["data"].isin(DIES_OK) & (df["sector"] == sec) & df["tram"].isin(FRANGES)]
    P = d.pivot_table(index="data", columns="tram", values="kwh", aggfunc="sum")[FRANGES]
    P = P[(P > 0).all(axis=1)]
    z = sum(P[tr] * np.exp(2j * np.pi * CENTRE[tr] / 24) for tr in FRANGES)
    out = (P.div(P.sum(axis=1), axis=0)).rename(columns=NOM)
    out["hora_pic"] = (np.angle(z) % (2 * np.pi)) * 24 / (2 * np.pi)
    out["amplitud"] = np.abs(z) / P.sum(axis=1)
    return out.reset_index()

R23d, fig_cicle = {}, {}
for nom_s, sec in SECTORS.items():
    c_ = cicle(sec).merge(temp[["data", "tmean"]], on="data")
    c_ = c_[c_["data"].dt.month.isin(MESOS_ESTIU) & c_["data"].dt.year.isin(ANYS)]
    c_["vespre_nit"] = c_[NOM[FRANGES[3]]] + c_[NOM[FRANGES[0]]]
    fig_cicle[nom_s] = c_
    # Pendent de l'hora del pic i del pes vespre+nit amb la temperatura, per estiu (efectes fixos de dia de la setmana)
    for var in ["hora_pic", "vespre_nit"]:
        t = taula_de(c_[["data"]].assign(kwh=1.0)).merge(c_[["data", var]], on="data")
        X = disseny_v07(t, LAGS, per_any=True)
        X["regim_B"] = t["data"].isin(REGIM_B).astype(float).values       # v0.11: control del règim de franges
        beta, _, cov, _ = ajusta_cl(t, X, y=t[var].values)
        esc = 60 if var == "hora_pic" else 100          # minuts per K, o punts percentuals per K
        for a in ANYS:
            v, e = chi0_de(beta, cov, f"_{a}")          # chi0_de multiplica per 100: ho desfem
            R23d[(nom_s, var, a)] = (esc * v / 100, esc * e / 100)
R23d = pd.Series({k: v[0] for k, v in R23d.items()}).unstack(), pd.Series({k: v[1] for k, v in R23d.items()}).unstack()
print("Per cada grau de calor (CDD): desplaçament de l'hora del pic (minuts/K) i pes de vespre+nit (punts %/K), per estiu")
print(R23d[0].round(2).to_string()); print("errors:"); print(R23d[1].round(2).to_string())
R23d[0].to_csv("cicle_diari_v12.csv")

fig, axs = plt.subplots(1, len(SECTORS), figsize=(6 * len(SECTORS), 4), squeeze=False)
for ax, nom_s in zip(axs[0], SECTORS):
    c_ = fig_cicle[nom_s]
    for a in ANYS:
        g = c_[(c_["data"].dt.year == a) & ~c_["data"].isin(REGIM_B)]       # només règim A
        b = g.groupby(np.floor(g["tmean"]))["vespre_nit"].agg(["mean", "size"])
        b = b[b["size"] >= 5]
        ax.plot(b.index, 100 * b["mean"], marker="o", label=str(a))
    ax.set_title(nom_s); ax.set_xlabel("T mitjana (°C)"); ax.grid(alpha=.3)
axs[0][0].set_ylabel("Pes de 18 a 6 h (%), dies del règim A"); axs[0][0].legend()
plt.tight_layout(); plt.savefig("cicle_diari_v12.png", dpi=150); plt.show()
```

```python
# ============ 23e. Qui és a casa? Laborables, caps de setmana, juliol i agost ============
es_festiu = lambda d: d.dt.dayofweek.ge(5) | d.isin(pd.to_datetime(list(festius.keys())))
SUBS = {
    "laborables": lambda t: t[~es_festiu(t["data"])],
    "caps de setmana i festius": lambda t: t[es_festiu(t["data"])],
    "juliol": lambda t: t[t["data"].dt.month == 7],
    "agost": lambda t: t[t["data"].dt.month == 8],
}
files = []
for nom_s, sec in SECTORS.items():
    for tr in [FRANGES[0], FRANGES[3], None]:
        base_t = taula_de(serie23(sec, tram=tr))
        for nom_sub, f in SUBS.items():
            try:
                sub = f(base_t).reset_index(drop=True)
                r = resum_anys(sub, extres=ext_regim(sub) if tr is not None else None)
                files.append({"sector": nom_s, "franja": "total" if tr is None else NOM[tr], "subconjunt": nom_sub,
                              **{a: r[a] for a in ANYS}, "creix_21_24": r["creix_21_24"], "se": r["se_21_24"]})
            except Exception as e:
                print(f"  [!] {nom_s} {tr} {nom_sub}: {e}")
R23e = pd.DataFrame(files).set_index(["sector", "franja", "subconjunt"])
R23e.to_csv("qui_es_a_casa_v12.csv")
print("χ⁰ (%/K) per subconjunt de dies")
print(R23e.round(2).to_string())
```

```python
# ============ 23f. Barris: on creix la resposta nocturna? ============
# Per a cada CP (o unitat de conservació de la 22b): χ⁰ per estiu de la franja de NIT (0-6 h) i de VESPRE (18-24 h),
# amb trams entre salts. Creixement = mitjana 2023-24 menys mitjana 2021-22.
def serie_cp(cps, tr):
    d = df[df["cp"].isin(cps) & df["data"].isin(DIES_OK) & (df["sector"] == sector_triat) & (df["tram"] == tr)]
    s = d.groupby("data", as_index=False)["kwh"].sum()
    return s[s["kwh"] > 0]

grups = [[cp] for cp in serie_res if not any(cp in u for u in unitats)] + [list(u) for u in unitats]
files = []
for g in grups:
    fila = {"cp": g[0] if len(g) == 1 else "+".join(g), "n_cp": len(g)}
    for tr, nom_ in [(FRANGES[0], "nit"), (FRANGES[3], "vespre")]:
        try:
            t = taula_de(serie_cp(g, tr), cp=g[0] if len(g) == 1 else None)
            f, _, _ = chi_per_any(t, extres=ext_regim(t))
            fila[f"creix_{nom_}"] = np.nanmean([f.get(2023), f.get(2024)]) - np.nanmean([f.get(2021), f.get(2022)])
            fila[f"chi21_{nom_}"] = f.get(2021)
        except Exception as e:
            fila[f"creix_{nom_}"] = np.nan
    files.append(fila)
R23f = pd.DataFrame(files).set_index("cp")

# Variables explicatives per CP (només per als CP individuals)
expl = pd.DataFrame(index=R23f.index)
if "renda_cp" in globals(): expl["renda"] = renda_cp
if "illa" in globals() and "dLST" in illa.columns: expl["dLST"] = illa["dLST"]
if "qf" in globals():
    for c in qf.columns:
        if qf[c].dtype.kind in "fi": expl[f"qf_{c}"] = qf[c]
if "hut_cp" in globals(): expl["places_hut"] = hut_cp["places"]
R23f = R23f.join(expl)
R23f.to_csv("barris_resposta_nocturna_v12.csv")
print(f"Creixement de χ⁰ nocturn per CP: mediana {R23f['creix_nit'].median():+.2f} %/K, "
      f"{(R23f['creix_nit'] > 0).mean():.0%} dels CP creixen")
print(R23f.sort_values("creix_nit", ascending=False)[["n_cp", "creix_nit", "creix_vespre"]].round(2).head(10).to_string())
print("\nCorrelacions de Spearman amb el creixement nocturn (CP individuals):")
for c in expl.columns:
    m = R23f[["creix_nit", c]].dropna()
    if len(m) > 8:
        rho, p = stats.spearmanr(m["creix_nit"], m[c])
        print(f"  {c:20s} ρ = {rho:+.2f} (p = {p:.3f}, n = {len(m)})")
try:
    # v0.11: les unitats fusionades ("08001+08004+08039") es pinten a tots els seus CP
    valor_cp = {cp: v for idx, v in R23f["creix_nit"].items() for cp in str(idx).split("+")}
    mapa_n = pol.assign(creix_nit=pol["cp"].map(valor_cp))
    ax = mapa_n.plot(column="creix_nit", cmap="RdBu_r", legend=True, figsize=(7, 7),
                     missing_kwds={"color": "lightgrey"}, vmin=-6, vmax=6, edgecolor="white", linewidth=.3)
    ax.set_title("Creixement de χ⁰ nocturn (0-6 h), 2021-22 → 2023-24 (%/K)"); ax.set_axis_off()
    plt.tight_layout(); plt.savefig("mapa_creixement_nit_v12.png", dpi=150); plt.show()
except Exception as e:
    print(f"[i] Sense mapa ({e})")
```

```python
# ============ 23g. Serveis i hotels ============
# v0.11: si la baixada automàtica falla, cal desar a CARPETA_LOCAL un fitxer hotels.csv (Open Data BCN, cerca "hotels",
# el CSV que tingui columnes de latitud i longitud).
# Hipòtesi: la resposta NOCTURNA dels serveis ve sobretot d'hotels (i hospitals). Si és així, ha de ser més gran
# als CP amb més places hoteleres per unitat de consum de serveis.
def baixa_hotels():
    q = requests.get("https://opendata-ajuntament.barcelona.cat/data/api/3/action/package_search?q=hotels&rows=20",
                     timeout=60).json()["result"]["results"]
    for p in q:
        for r in p.get("resources", []):
            if str(r.get("format", "")).lower() == "csv":
                try:
                    h = llegeix_csv_flexible(r["url"])
                except Exception:
                    continue
                cx = [c for c in h.columns if re.search("lon|_x$|longitud", c, re.I)]
                cy = [c for c in h.columns if re.search("lat|_y$|latitud", c, re.I)]
                if cx and cy:
                    print(f"[OK] Hotels: '{p['title']}' · {r.get('name')} · {len(h)} files")
                    return h.rename(columns={cx[0]: "lon", cy[0]: "lat"})
    raise ValueError("cap CSV d'hotels amb coordenades")

R23g = None
try:
    try:
        hot = baixa_hotels()
    except Exception as e:
        print(f"[!] Hotels no baixats ({e}); provo {CARPETA_LOCAL}/hotels.csv")
        hot = llegeix_csv_flexible(f"{CARPETA_LOCAL}/hotels.csv")
        cx = [c for c in hot.columns if re.search("lon|_x$|longitud", c, re.I)][0]
        cy = [c for c in hot.columns if re.search("lat|_y$|latitud", c, re.I)][0]
        hot = hot.rename(columns={cx: "lon", cy: "lat"})
    hot["lon"] = pd.to_numeric(hot["lon"].astype(str).str.replace(",", "."), errors="coerce")
    hot["lat"] = pd.to_numeric(hot["lat"].astype(str).str.replace(",", "."), errors="coerce")
    hot = hot.dropna(subset=["lon", "lat"])
    pts = gpd.GeoDataFrame(hot, geometry=gpd.points_from_xy(hot["lon"], hot["lat"]), crs="EPSG:4326").to_crs(pol.crs)
    n_hot = gpd.sjoin(pts, pol[["cp", "geometry"]], predicate="within").groupby("cp").size().rename("n_hotels")
    files = []
    for cp in serie_res:
        fila = {"cp": cp}
        for tr, nom_ in [(FRANGES[0], "nit"), (FRANGES[2], "tarda")]:
            d = df[(df["cp"] == cp) & df["data"].isin(DIES_OK) & (df["sector"] == SECTORS["Serveis"]) & (df["tram"] == tr)]
            s = d.groupby("data", as_index=False)["kwh"].sum()
            try:
                t_ = taula_de(s[s["kwh"] > 0], cp=cp)                       # amb trams entre salts
                beta_, _, cov_, _ = ajusta_cl(t_, disseny_v07(t_, LAGS, amb_trams=True))
                fila[f"chi_{nom_}"], _ = chi0_de(beta_, cov_)
            except Exception:
                fila[f"chi_{nom_}"] = np.nan
        files.append(fila)
    R23g = pd.DataFrame(files).set_index("cp").join(n_hot).fillna({"n_hotels": 0})
    mwh_serveis = df[df["sector"] == SECTORS["Serveis"]].groupby("cp")["kwh"].sum() / 1e3
    R23g["hotels_per_GWh"] = R23g["n_hotels"] / (mwh_serveis.reindex(R23g.index) / 1e3)
    R23g["nit_menys_tarda"] = R23g["chi_nit"] - R23g["chi_tarda"]
    R23g.to_csv("serveis_hotels_v12.csv")
    for c in ["chi_nit", "nit_menys_tarda"]:
        m = R23g[[c, "hotels_per_GWh"]].dropna()
        rho, p = stats.spearmanr(m[c], m["hotels_per_GWh"])
        print(f"Serveis · {c} vs hotels per GWh de serveis: ρ = {rho:+.2f} (p = {p:.3f}, n = {len(m)})")
except Exception as e:
    print(f"[i] 23g saltada: {e}")
```

```python
# ============ 23h. Explicacions alternatives ============
# (i) AUTOCONSUM FOTOVOLTAIC: les plaques redueixen el consum que veu la xarxa al migdia, i més els dies de sol.
#     Si això frena la χ del migdia, la interacció calor × radiació a la franja 12-18 hauria de fer-se més NEGATIVA cada any.
if "rad" in tf.columns:
    for nom_s, sec in SECTORS.items():
        t = taula_de(serie23(sec, tram=FRANGES[2])).merge(tf[["data", "rad"]], on="data", how="left")
        t["rad"] = t["rad"].fillna(t["rad"].mean())
        rz = ((t["rad"] - t["rad"].mean()) / t["rad"].std()).values
        ext = pd.DataFrame({"rad": rz, "regim_B": t["data"].isin(REGIM_B).astype(float).values})
        for a in ANYS:
            ext[f"cdd0_x_rad_{a}"] = t["cdd0"].values * rz * (t["data"].dt.year == a).values
        f, beta, cov = chi_per_any(t, extres=ext)
        txt = ", ".join(f"{a}: {100*beta[f'cdd0_x_rad_{a}']:+.2f}±{100*np.sqrt(cov.loc[f'cdd0_x_rad_{a}', f'cdd0_x_rad_{a}']):.2f}"
                        for a in ANYS if f"cdd0_x_rad_{a}" in beta.index)
        print(f"(i) {nom_s} 12-18 h · interacció calor × radiació (%/K per σ de radiació): {txt}")
        print(f"    χ⁰ de la franja 12-18 controlant la radiació: creixement 2021→2024 {f['creix_21_24']:+.2f} ± {f['se_creix']:.2f}")
else:
    print("(i) Sense radiació: prova fotovoltaica saltada")

# (ii) COTXE ELÈCTRIC: augmenta el NIVELL nocturn, però no hauria de dependre de la temperatura.
#      Comparem el nivell de la franja 0-6 h els dies sense calor amb la χ⁰ nocturna.
t = taula_de(serie23(sector_triat, tram=FRANGES[0]))
nivell = t[t["cdd0"] == 0].groupby(t["data"].dt.year)["kwh"].mean() / 1e3
print("\n(ii) Residencial 0-6 h · nivell dels dies sense calor (MWh) i χ⁰ (%/K):")
for a in ANYS:
    print(f"   {a}: nivell {nivell.get(a, np.nan):6.0f}   χ⁰ {R23b.loc[('Residencial', NOM[FRANGES[0]]), a]:5.2f}")

# (iii) COVID 2021: el canvi 2022→2024 i el pendent dels cinc anys ja són a la taula 23b.
print("\n(iii) Canvi sense l'estiu 2021 (2022→2024) i pendent dels cinc estius, per franja:")
print(R23b[["creix_21_24", "creix_22_24", "se_22_24", "pendent", "se_pendent"]].round(2).to_string())

# (iv) CAPA LÍMIT (per a l'article 2): quantes vegades més prima és de nit que de dia?
if "blh_00" in tf.columns:
    e_ = tf[tf["data"].dt.month.isin(MESOS_ESTIU)]
    r_ = (e_["blh_12"] / e_["blh_00"]).median()
    print(f"\n(iv) Alçada de la capa límit a l'estiu: nit (0-6 h) {e_['blh_00'].median():.0f} m · "
          f"dia (12-18 h) {e_['blh_12'].median():.0f} m · quocient dia/nit (mediana) {r_:.1f}")
```

## 24. La branca de fred: quant consumim per cada grau de fred, i com és de simètrica la V?

**Pregunta.** Fins ara només hem mirat la branca d'estiu de la V. A l'hivern, el consum elèctric també puja quan fa fred. Quant puja per cada grau, i ha canviat entre el 2021 i el 2025?

**Model (per a cada hivern, novembre-març):**
$$\ln E(d) = \sum_{k=0}^{3} \chi_h(k)\,\mathrm{HDD}(d-k) + \text{hivern} + \text{setmana ISO} + \text{dia de la setmana} + \text{festiu} + \text{Nadal} + \varepsilon, \qquad \mathrm{HDD}=\max(T_h-T,0)$$

**Per què aquests efectes fixos:**
- **Setmana ISO:** absorbeix el cicle de la **llum**. A l'hivern s'encén més llum, però a la mateixa setmana de l'any les hores de sol són pràcticament iguals cada any. Per tant, $\chi_h$ s'identifica amb les **anomalies de temperatura** (una setmana de gener més freda que la del mateix gener d'un altre any), no amb el cicle de l'any. Comprovat amb dades sintètiques: un cicle de llum sense resposta al fred dona $\chi_h \approx 0$.
- **Nadal** (24/12 a 6/1): les vacances no cauen sempre a la mateixa setmana ISO.

**Què esperar.** A Barcelona la calefacció és sobretot de **gas**: l'electricitat només veu la part electrificada (bombes de calor, radiadors elèctrics, aire condicionat reversible). Per tant, esperem $\chi_h$ més petita que $\chi_c$. El quocient
$$R = \frac{\chi_h}{\chi_c}$$
mesura la **simetria de la V**: $R=1$ vol dir que cada grau de fred mou l'electricitat tant com cada grau de calor.

**La lectura que ens interessa:** el mateix aparell (una bomba de calor reversible) fa fred a l'estiu i calor a l'hivern. Si la tendència d'estiu ve de més aparells, $\chi_h$ també hauria de créixer i $R$ s'acostaria a 1. Si la tendència d'estiu és ús més intens de l'aire condicionat, $\chi_h$ no té per què moure's. **És una prova creuada del resultat principal i un indicador d'electrificació de la calefacció** (connecta amb les bombes de calor de l'Eva).

**Confusions a vigilar:** el gas no el veiem (no es pot tancar el balanç); els hiverns del 2022-23 van coincidir amb la crisi de preus (l'estalvi forçat podria abaixar $\chi_h$); salts de nivell de les dades dins d'un hivern (robustesa R5).

```python
# ============ 24·0. Codi del paquet consum_temperatura (v0.2.0) ============
# Còpia EXACTA dels mòduls del repositori power_temperature_urban (cel·la generada automàticament).
# No cal llegir-la: la documentació és al repositori i a les cel·les de text de les seccions 24 i 25.
# L'escrivim en una carpeta local i la importem com a 'ct'. Al final comprovem amb dades sintètiques
# que els estimadors recuperen paràmetres coneguts (regla de la tesi: res amb dades reals sense això).
import os, sys, importlib
FITXERS_PAQUET = {
    '__init__.py': r'''"""consum_temperatura: response of urban electricity consumption to temperature.

Code of article 1 of Pau Buchaca's PhD thesis (ETSAB-UPC): how Barcelona's
electricity consumption responds to heat, why the response grows mostly at night,
and how much of the growth comes from more air conditioners versus more use.
"""
from .forcings import cooling_degree_days, heating_degree_days, heatwaves, band_temperatures
from .response import (design_matrix, fit_cluster, prepare, static_susceptibility, fit_city,
                       choose_tc, response_by_year, growth, threshold_factor, extensive_share)
from .data import from_cache, era5_daily, era5_hourly, daily_consumption
from .alerts import band_regime, complete_days, level_jumps
from .seasonal import (winter_label, christmas, prepare_winter, winter_design, fit_winter, choose_th,
                       response_by_winter, asymmetry, falling_days, flexible_memory, hysteresis_test, summer_phase,
                       response_by_phase)
from . import synthetic

__version__ = "0.2.0"
''',
    'forcings.py': r'''"""Forcings: turn temperature series into the regressors of the response model.

Comments in Catalan follow the style of the thesis notebooks.
"""
import numpy as np
import pandas as pd


def cooling_degree_days(temp, tc, lags=3, col="tmean"):
    """Add cooling degree days with lags: cdd_k(d) = max(T(d-k) - tc, 0), k = 0..lags.

    temp: DataFrame with a 'data' (date) column and a temperature column `col`.
    """
    t = temp.sort_values("data").copy()
    for k in range(lags + 1):
        # graus-dia de refrigeració del dia d-k
        t[f"cdd{k}"] = np.maximum(t[col].shift(k) - tc, 0)
    return t


def heating_degree_days(temp, th, lags=3, col="tmean"):
    """Add heating degree days with lags: hdd_k(d) = max(th - T(d-k), 0)."""
    t = temp.sort_values("data").copy()
    for k in range(lags + 1):
        t[f"hdd{k}"] = np.maximum(th - t[col].shift(k), 0)
    return t


def heatwaves(temp, percentile=0.95, min_days=3, months=(6, 7, 8, 9), col="tmax"):
    """Boolean Series: days inside a run of >= min_days with tmax above the summer percentile."""
    summer = temp[temp["data"].dt.month.isin(months)]
    threshold = summer[col].quantile(percentile)
    hot = temp[col] >= threshold
    run_id = (hot != hot.shift()).cumsum()          # numera les ratxes consecutives
    run_len = hot.groupby(run_id).transform("sum")
    return hot & (run_len >= min_days)


def band_temperatures(hourly, bands=None):
    """Daily mean temperature in hour bands (e.g. night 0-6 h, day 12-18 h).

    hourly: DataFrame with 'hora' (timestamp) and 'temp'.
    bands: dict name -> (first_hour, last_hour_exclusive).
    """
    if bands is None:
        bands = {"nit": (0, 6), "dia": (12, 18)}
    h = hourly.copy()
    h["data"] = h["hora"].dt.normalize()
    h["h"] = h["hora"].dt.hour
    out = None
    for name, (a, b) in bands.items():
        m = h[(h["h"] >= a) & (h["h"] < b)].groupby("data", as_index=False)["temp"].mean()
        m = m.rename(columns={"temp": f"t_{name}"})
        out = m if out is None else out.merge(m, on="data", how="outer")
    return out.sort_values("data").reset_index(drop=True)
''',
    'response.py': r'''"""Linear response of log electricity consumption to temperature forcing.

Model (per spatial unit, summer days):

    ln E(d) = sum_k chi_k * CDD(d - k) + fixed effects (year, ISO week, weekday, holiday) + eps

The static susceptibility chi0 = sum_k chi_k is the % change of consumption per
degree-day (x100). Standard errors are cluster-robust by ISO week, because
residuals of consecutive days are correlated.
"""
import numpy as np
import pandas as pd

from .forcings import cooling_degree_days


def design_matrix(table, lags=3, years=None, holidays=(), by_year=False):
    """Regressors: CDD lags (optionally interacted with year) + fixed effects.

    table must contain 'data' and cdd0..cdd{lags}. holidays: iterable of dates.
    """
    years = sorted(table["data"].dt.year.unique()) if years is None else list(years)
    cdd_cols = [f"cdd{k}" for k in range(lags + 1)]
    if by_year:
        # una funció de resposta per estiu: CDD x indicador d'any
        parts = {}
        yr = table["data"].dt.year.values
        for a in years:
            for c in cdd_cols:
                parts[f"{c}_{a}"] = table[c].values * (yr == a)
        X_cdd = pd.DataFrame(parts)
    else:
        X_cdd = table[cdd_cols].reset_index(drop=True)
    cat = lambda v, cats: pd.Categorical(v, categories=cats)
    hol = pd.to_datetime(pd.Series(list(holidays))) if len(holidays) else pd.Series([], dtype="datetime64[ns]")
    fe = pd.get_dummies(pd.DataFrame({
        "any": cat(table["data"].dt.year.values, years),
        "setmana": cat(table["data"].dt.isocalendar().week.astype(int).values, list(range(1, 54))),
        "dia": cat(table["data"].dt.dayofweek.values, list(range(7))),
        "festiu": cat(table["data"].isin(hol).values, [False, True]),
    }), drop_first=True).astype(float)
    X = pd.concat([X_cdd.reset_index(drop=True), fe.reset_index(drop=True)], axis=1)
    X = X.loc[:, X.abs().sum() > 0]              # treu columnes buides (efectes sense dades)
    X.insert(0, "const", 1.0)
    return X


def fit_cluster(table, X, y=None):
    """OLS with cluster-robust covariance by ISO week. Returns (beta, cov, residuals)."""
    if y is None:
        y = np.log(table["kwh"].values)
    Xv = X.values.astype(float)
    beta, *_ = np.linalg.lstsq(Xv, y, rcond=None)
    res = y - Xv @ beta
    n, p = Xv.shape
    pinv = np.linalg.pinv(Xv.T @ Xv)
    iso = table["data"].dt.isocalendar()
    g = (iso["year"].astype(int) * 100 + iso["week"].astype(int)).values
    groups = np.unique(g)
    if len(groups) > 1:
        meat = np.zeros((p, p))
        for gi in groups:
            s = Xv[g == gi].T @ res[g == gi]
            meat += np.outer(s, s)
        G = len(groups)
        cov = pinv @ meat @ pinv * (G / (G - 1)) * ((n - 1) / max(n - p, 1))
    else:
        cov = (res @ res / max(n - p, 1)) * pinv
    return pd.Series(beta, index=X.columns), pd.DataFrame(cov, index=X.columns, columns=X.columns), res


def prepare(consumption, temp, tc, lags=3, months=(6, 7, 8, 9), years=None):
    """Merge daily consumption ('data','kwh') with CDD lags; keep summer days of the chosen years."""
    t = cooling_degree_days(temp, tc, lags)
    table = consumption.merge(t, on="data")
    table = table[table["data"].dt.month.isin(months)]
    if years is not None:
        table = table[table["data"].dt.year.isin(years)]
    return table.dropna().reset_index(drop=True)


def static_susceptibility(beta, cov, lags=3, suffix="", prefix="cdd"):
    """chi0 = 100 * sum_k chi_k (% per degree-day) and its standard error.

    prefix: "cdd" (heat branch) or "hdd" (cold branch).
    """
    names = [f"{prefix}{k}{suffix}" for k in range(lags + 1)]
    names = [n for n in names if n in beta.index]
    w = np.ones(len(names))
    chi = 100 * float(beta[names].values @ w)
    se = 100 * float(np.sqrt(w @ cov.loc[names, names].values @ w))
    return chi, se


def fit_city(consumption, temp, tc, lags=3, holidays=(), exclude=None, **kw):
    """Static response of a whole series. exclude: iterable of dates to drop (e.g. heatwave days)."""
    table = prepare(consumption, temp, tc, lags, **kw)
    if exclude is not None:
        table = table[~table["data"].isin(pd.to_datetime(list(exclude)))].reset_index(drop=True)
    X = design_matrix(table, lags, holidays=holidays)
    beta, cov, res = fit_cluster(table, X)
    chi, se = static_susceptibility(beta, cov, lags)
    return {"chi0": chi, "se": se, "beta": beta, "cov": cov, "n": len(table), "mse": float(np.mean(res ** 2))}


def choose_tc(consumption, temp, candidates=np.arange(18.0, 25.5, 0.5), lags=3, holidays=(), **kw):
    """Comfort temperature that minimises the in-sample mean squared error."""
    rows = []
    for tc in candidates:
        r = fit_city(consumption, temp, tc, lags, holidays, **kw)
        rows.append((tc, r["mse"]))
    prof = pd.DataFrame(rows, columns=["tc", "mse"])
    return float(prof.loc[prof["mse"].idxmin(), "tc"]), prof


def response_by_year(consumption, temp, tc, lags=3, years=None, holidays=(), exclude=None, **kw):
    """chi0 for each summer (CDD x year interaction) and growth between two years.

    exclude: iterable of dates to drop (e.g. heatwave days, as in section 22d of notebook v0.12.1).
    """
    table = prepare(consumption, temp, tc, lags, years=years, **kw)
    if exclude is not None:
        table = table[~table["data"].isin(pd.to_datetime(list(exclude)))].reset_index(drop=True)
    years = sorted(table["data"].dt.year.unique()) if years is None else list(years)
    X = design_matrix(table, lags, years=years, holidays=holidays, by_year=True)
    beta, cov, _ = fit_cluster(table, X)
    out = {}
    for a in years:
        out[a] = static_susceptibility(beta, cov, lags, suffix=f"_{a}")
    return pd.DataFrame(out, index=["chi0", "se"]).T, beta, cov


def growth(beta, cov, y0, y1, lags=3, prefix="cdd"):
    """Difference chi0(y1) - chi0(y0) with its standard error (covariances included).

    prefix: "cdd" for summers (labels = years), "hdd" for winters (labels = starting year).
    """
    n0 = [f"{prefix}{k}_{y0}" for k in range(lags + 1)]
    n1 = [f"{prefix}{k}_{y1}" for k in range(lags + 1)]
    names = n0 + n1
    w = np.r_[-np.ones(len(n0)), np.ones(len(n1))]
    d = 100 * float(beta[names].values @ w)
    se = 100 * float(np.sqrt(w @ cov.loc[names, names].values @ w))
    return d, se


def threshold_factor(curves, errors, ref=None, max_iter=200):
    """Common shape x factor per year: curves[b, year] ~ g(b) * lambda(year), lambda(ref) = 1.

    curves, errors: DataFrames (index = temperature bins above the threshold, columns = years).
    Weighted alternating least squares, as in section 22k of notebook v0.12.1 (20k in v0.11).
    Returns (lambda Series, se Series, shape g Series, chi2, dof).
    """
    y, s = curves, errors
    ref = y.columns[0] if ref is None else ref
    lam = pd.Series(1.0, index=y.columns)
    for _ in range(max_iter):
        g = (y * lam / s ** 2).sum(axis=1) / ((lam ** 2) / s ** 2).sum(axis=1)
        new = (y.mul(g, axis=0) / s ** 2).sum() / ((s ** -2).mul(g ** 2, axis=0)).sum()
        new = new / new[ref]
        if np.allclose(new, lam, atol=1e-10):
            lam = new
            break
        lam = new
    g = (y * lam / s ** 2).sum(axis=1) / ((lam ** 2) / s ** 2).sum(axis=1)
    pred = pd.DataFrame(np.outer(g, lam), index=y.index, columns=y.columns)
    chi2 = float((((y - pred) / s) ** 2).sum().sum())
    dof = int(y.notna().sum().sum()) - len(y.index) - (len(y.columns) - 1)
    se = 1 / np.sqrt(((s ** -2).mul(g ** 2, axis=0)).sum())
    return lam, se, g, chi2, dof


def extensive_share(n0, n1, lam0, lam1):
    """Share of the log-growth of the response explained by more appliances.

    Minimal model chi ∝ n_AC · u  =>  ln(lam1/lam0) = ln(n1/n0) + ln(u1/u0).
    """
    return float(np.log(n1 / n0) / np.log(lam1 / lam0))
''',
    'seasonal.py': r'''"""The whole year: the cold branch of the V and the hysteresis test.

Two new lines of article 1 (notebook v0.12, sections 24 and 25):

1. Cold branch. Winter model (November to March):

       ln E(d) = sum_k chi_h,k * HDD(d - k) + fixed effects (winter, ISO week, weekday,
                 holiday, Christmas) + eps,      HDD = max(Th - T, 0)

   The ISO-week fixed effect absorbs the daylight cycle (hours of light are almost the
   same in the same week every year), so chi_h is identified from weather anomalies
   within the same week of different years. The ratio chi_h / chi_c measures how
   symmetric the V is. In Barcelona heating is mostly gas: electricity only sees the
   electrified part (heat pumps, electric radiators). If chi_h grows over the years,
   that is a signature of electrification of heating.

2. Hysteresis test ("fronts" test). At the same temperature, do we consume more when
   the heat is leaving (temperature falling) than when it is arriving (rising), once the
   linear memory (lags) is accounted for?

       ln E(d) = sum_k chi_k CDD(d - k) + gamma * CDD(d) * 1[T(d) < T(d-1)] + FE + eps

   A linear response with memory predicts gamma = 0 (the lags already describe the
   difference between rising and falling days). Relays with hysteresis (an air
   conditioner switched on above T_on and off below T_off < T_on) predict gamma > 0.
   This is the C3 criterion: linear memory closes the loop, true hysteresis does not.

Comments in Catalan follow the style of the thesis notebooks.
"""
import numpy as np
import pandas as pd

from .forcings import cooling_degree_days, heating_degree_days
from .response import fit_cluster, prepare, design_matrix, static_susceptibility

WINTER_MONTHS = (11, 12, 1, 2, 3)


# ---------------------------------------------------------------------------
# 1. Branca de fred
# ---------------------------------------------------------------------------

def winter_label(dates):
    """Winter of each date, labelled by the year in which it starts (Nov 2021 - Mar 2022 -> 2021)."""
    d = pd.to_datetime(pd.Series(dates))
    return np.where(d.dt.month.values >= 7, d.dt.year.values, d.dt.year.values - 1)


def christmas(dates):
    """True from 24 December to 6 January (Christmas holidays: the ISO week does not capture them)."""
    d = pd.to_datetime(pd.Series(dates))
    m, day = d.dt.month.values, d.dt.day.values
    return ((m == 12) & (day >= 24)) | ((m == 1) & (day <= 6))


def prepare_winter(consumption, temp, th, lags=3, months=WINTER_MONTHS, winters=None):
    """Merge daily consumption ('data', 'kwh') with HDD lags; keep winter days.

    winters: list of winter labels to keep (e.g. [2021, 2022, 2023, 2024]).
    """
    t = heating_degree_days(temp, th, lags)
    table = consumption.merge(t, on="data")
    table = table[table["data"].dt.month.isin(months)].copy()
    table["hivern"] = winter_label(table["data"])
    if winters is not None:
        table = table[table["hivern"].isin(winters)]
    return table.dropna().reset_index(drop=True)


def winter_design(table, lags=3, holidays=(), by_winter=False, winter_month=False):
    """Regressors: HDD lags (optionally one set per winter) + fixed effects.

    Fixed effects: winter, ISO week (daylight cycle), weekday, holiday, Christmas.
    winter_month=True adds winter x month effects (absorbs level jumps of the data inside a winter).
    """
    winters = sorted(table["hivern"].unique())
    hdd_cols = [f"hdd{k}" for k in range(lags + 1)]
    if by_winter:
        # una funció de resposta per hivern: HDD x indicador d'hivern
        parts = {}
        w = table["hivern"].values
        for a in winters:
            for c in hdd_cols:
                parts[f"{c}_{a}"] = table[c].values * (w == a)
        X_h = pd.DataFrame(parts)
    else:
        X_h = table[hdd_cols].reset_index(drop=True)
    cat = lambda v, cats: pd.Categorical(v, categories=cats)
    hol = pd.to_datetime(pd.Series(list(holidays))) if len(holidays) else pd.Series([], dtype="datetime64[ns]")
    fe = pd.get_dummies(pd.DataFrame({
        "hivern": cat(table["hivern"].values, winters),
        "setmana": cat(table["data"].dt.isocalendar().week.astype(int).values, list(range(1, 54))),
        "dia": cat(table["data"].dt.dayofweek.values, list(range(7))),
        "festiu": cat(table["data"].isin(hol).values, [False, True]),
        "nadal": cat(christmas(table["data"]), [False, True]),
    }), drop_first=True).astype(float)
    if winter_month:
        # efectes fixos hivern x mes: un nivell propi per a cada mes de cada hivern
        wm = pd.Series(table["hivern"].astype(str).values + "_" + table["data"].dt.month.astype(str).values)
        fe = pd.concat([fe, pd.get_dummies(wm, prefix="hm", drop_first=True).astype(float)], axis=1)
    X = pd.concat([X_h.reset_index(drop=True), fe.reset_index(drop=True)], axis=1)
    X = X.loc[:, X.abs().sum() > 0]              # treu columnes buides
    X.insert(0, "const", 1.0)
    return X


def fit_winter(consumption, temp, th, lags=3, holidays=(), winters=None, exclude=None,
               months=WINTER_MONTHS, winter_month=False):
    """Static cold response chi_h (% per degree-day below th) of a whole series."""
    table = prepare_winter(consumption, temp, th, lags, months=months, winters=winters)
    if exclude is not None:
        table = table[~table["data"].isin(pd.to_datetime(list(exclude)))].reset_index(drop=True)
    X = winter_design(table, lags, holidays, winter_month=winter_month)
    beta, cov, res = fit_cluster(table, X)
    chi, se = static_susceptibility(beta, cov, lags, prefix="hdd")
    return {"chi_h": chi, "se": se, "beta": beta, "cov": cov, "n": len(table),
            "mse": float(np.mean(res ** 2))}


def choose_th(consumption, temp, candidates=np.arange(10.0, 19.5, 0.5), lags=3, holidays=(), winters=None,
              exclude=None):
    """Heating threshold that minimises the in-sample mean squared error (same rule as choose_tc)."""
    rows = [(th, fit_winter(consumption, temp, th, lags, holidays, winters, exclude)["mse"]) for th in candidates]
    prof = pd.DataFrame(rows, columns=["th", "mse"])
    return float(prof.loc[prof["mse"].idxmin(), "th"]), prof


def response_by_winter(consumption, temp, th, lags=3, winters=None, holidays=(), exclude=None,
                       months=WINTER_MONTHS, winter_month=False):
    """chi_h for each winter (HDD x winter interaction). Returns (table, beta, cov)."""
    table = prepare_winter(consumption, temp, th, lags, months=months, winters=winters)
    if exclude is not None:
        table = table[~table["data"].isin(pd.to_datetime(list(exclude)))].reset_index(drop=True)
    X = winter_design(table, lags, holidays, by_winter=True, winter_month=winter_month)
    beta, cov, _ = fit_cluster(table, X)
    out = {a: static_susceptibility(beta, cov, lags, suffix=f"_{a}", prefix="hdd")
           for a in sorted(table["hivern"].unique())}
    return pd.DataFrame(out, index=["chi_h", "se"]).T, beta, cov


def asymmetry(chi_h, se_h, chi_c, se_c):
    """Ratio R = chi_h / chi_c and its standard error (delta method, independent estimates).

    R = 1: symmetric V (each degree of cold moves electricity as much as each degree of heat).
    R << 1: heating mostly not electric.
    """
    r = chi_h / chi_c
    se = abs(r) * np.sqrt((se_h / chi_h) ** 2 + (se_c / chi_c) ** 2)
    return float(r), float(se)


# ---------------------------------------------------------------------------
# 2. Histèresi
# ---------------------------------------------------------------------------

def falling_days(temp, col="tmean"):
    """True on days whose mean temperature is lower than the day before (the heat is leaving)."""
    t = temp.sort_values("data")
    return pd.Series((t[col] < t[col].shift(1)).values, index=t["data"].values)


def flexible_memory(table, temp, tc, lags=3, knots=(0, 2, 4, 6), col="tmean"):
    """Additive, curved memory: hinge functions max(T(d-k) - (tc + knot), 0) for every lag k.

    Any additive response sum_k f_k(T(d-k)), linear or curved, is well approximated by these
    columns. Hysteresis is NOT additive (the state depends on the path), so it cannot hide here.
    """
    t = temp.sort_values("data").set_index("data")[col]
    cols = {}
    for k in range(lags + 1):
        tk = t.shift(k).reindex(table["data"]).values
        for kn in knots:
            cols[f"m{k}_{kn:g}"] = np.maximum(tk - (tc + kn), 0)
    return pd.DataFrame(cols)


def hysteresis_test(consumption, temp, tc, lags=3, holidays=(), months=(6, 7, 8, 9), years=None,
                    exclude=None, knots=(0, 2, 4, 6), linear=False):
    """Fronts test: extra response on falling days at the same temperature, net of memory.

        ln E(d) = memory(T(d), T(d-1), ..., T(d-lags)) + gamma * CDD(d) * 1[T(d) < T(d-1)] + FE + eps

    memory = flexible additive hinge basis (default) or plain CDD lags (linear=True, kept only
    to show why it fails: with a curved response it gives false positives).
    Synthetic checks (tests/test_seasonal.py): gamma ~ 0 for a linear response with memory and
    for a curved response without hysteresis; gamma > 0 for relays with a hysteresis band of
    1-2 K. A band of 3 K or more looks like memory in daily data and is NOT detected.
    Returns dict with gamma (% per degree-day), se, z, n and the share of hot days that are falling.
    """
    table = prepare(consumption, temp, tc, lags, months=months, years=years)
    if exclude is not None:
        table = table[~table["data"].isin(pd.to_datetime(list(exclude)))].reset_index(drop=True)
    fall = falling_days(temp).reindex(table["data"].values).fillna(False).values.astype(bool)
    X = design_matrix(table, lags, years=years, holidays=holidays)
    if not linear:
        # memòria flexible: substituïm els CDD lineals per frontisses a diverses temperatures
        X = X.drop(columns=[f"cdd{k}" for k in range(lags + 1)])
        X = pd.concat([X, flexible_memory(table, temp, tc, lags, knots)], axis=1)
    # gamma: CDD del dia x indicador "la temperatura baixa respecte d'ahir"
    X["cdd0_baixa"] = table["cdd0"].values * fall
    beta, cov, _ = fit_cluster(table, X)
    g = 100 * float(beta["cdd0_baixa"])
    se = 100 * float(np.sqrt(cov.loc["cdd0_baixa", "cdd0_baixa"]))
    hot = table["cdd0"].values > 0
    return {"gamma": g, "se": se, "z": g / se if se > 0 else np.nan, "n": len(table),
            "frac_baixa": float(np.mean(fall[hot])) if hot.any() else np.nan}


PHASES = {"arribada": ((6, 1), (7, 15)), "plena": ((7, 16), (8, 15)), "sortida": ((8, 16), (9, 30))}


def summer_phase(dates):
    """Phase of the summer: 'arribada' (1 Jun - 15 Jul), 'plena' (16 Jul - 15 Aug), 'sortida' (16 Aug - 30 Sep)."""
    d = pd.to_datetime(pd.Series(dates))
    md = d.dt.month.values * 100 + d.dt.day.values
    out = np.full(len(d), None, dtype=object)
    for name, ((m0, d0), (m1, d1)) in PHASES.items():
        out[(md >= m0 * 100 + d0) & (md <= m1 * 100 + d1)] = name
    return out


def response_by_phase(consumption, temp, tc, lags=3, holidays=(), years=None, exclude=None):
    """chi0 in each phase of the summer (arrival, full, departure) with week fixed effects.

    Returns (table phase -> chi0, se; difference departure - arrival with se).
    Caution: August occupancy also changes the slope; this is a descriptive complement
    to the fronts test, not a test of hysteresis by itself.
    """
    table = prepare(consumption, temp, tc, lags, months=(6, 7, 8, 9), years=years)
    if exclude is not None:
        table = table[~table["data"].isin(pd.to_datetime(list(exclude)))].reset_index(drop=True)
    ph = summer_phase(table["data"])
    X = design_matrix(table, lags, years=years, holidays=holidays)
    cdd_cols = [f"cdd{k}" for k in range(lags + 1)]
    X = X.drop(columns=cdd_cols)
    for name in PHASES:
        for c in cdd_cols:
            X[f"{c}_{name}"] = table[c].values * (ph == name)
    beta, cov, _ = fit_cluster(table, X)
    out = {name: static_susceptibility(beta, cov, lags, suffix=f"_{name}") for name in PHASES}
    n0 = [f"cdd{k}_arribada" for k in range(lags + 1)]
    n1 = [f"cdd{k}_sortida" for k in range(lags + 1)]
    w = np.r_[-np.ones(len(n0)), np.ones(len(n1))]
    d = 100 * float(beta[n0 + n1].values @ w)
    se = 100 * float(np.sqrt(w @ cov.loc[n0 + n1, n0 + n1].values @ w))
    return pd.DataFrame(out, index=["chi0", "se"]).T, (d, se)
''',
    'synthetic.py': r'''"""Synthetic data with a known response, to check that the estimators recover it.

Rule of the thesis: no method is used on real data before it recovers known
parameters on synthetic data.
"""
import numpy as np
import pandas as pd


def summer_temperature(years=(2021, 2022, 2023, 2024, 2025), seed=0):
    """Daily temperature with a seasonal cycle and AR(1) weather noise (all year)."""
    rng = np.random.default_rng(seed)
    days = pd.date_range(f"{min(years)}-01-01", f"{max(years)}-12-31", freq="D")
    doy = days.dayofyear.values
    seasonal = 16 + 8 * np.sin(2 * np.pi * (doy - 110) / 365)
    noise = np.zeros(len(days))
    for i in range(1, len(days)):
        noise[i] = 0.7 * noise[i - 1] + rng.normal(0, 1.5)
    tmean = seasonal + noise
    return pd.DataFrame({"data": days, "tmean": tmean, "tmax": tmean + 5})


def consumption(temp, tc=22.0, chi_by_year=None, lag_weights=(0.5, 0.3, 0.15, 0.05),
                base=1e5, sigma=0.02, seed=1):
    """Daily kWh with ln E = ln base + chi(year)/100 * sum_k w_k CDD(d-k) + weekday effect + noise.

    chi_by_year: dict year -> static susceptibility (% per degree-day).
    """
    rng = np.random.default_rng(seed)
    t = temp.sort_values("data").reset_index(drop=True)
    cdd = np.maximum(t["tmean"] - tc, 0)
    if chi_by_year is None:
        chi_by_year = {a: 3.0 for a in t["data"].dt.year.unique()}
    chi = t["data"].dt.year.map(chi_by_year).values / 100
    resp = sum(w * cdd.shift(k).fillna(0).values for k, w in enumerate(lag_weights))
    weekday = np.where(t["data"].dt.dayofweek.values >= 5, -0.08, 0.0)
    ln_e = np.log(base) + chi * resp + weekday + rng.normal(0, sigma, len(t))
    return pd.DataFrame({"data": t["data"], "kwh": np.exp(ln_e)})


def annual_consumption(temp, tc=22.0, chi_c_by_year=None, th=15.0, chi_h_by_winter=None,
                       lag_weights=(0.5, 0.3, 0.15, 0.05), base=1e5, light=0.10,
                       christmas_effect=0.06, sigma=0.02, seed=2):
    """Whole-year daily kWh with a heat branch and a cold branch (the V).

    ln E = ln base + chi_c(year)/100 * sum_k w_k CDD(d-k) + chi_h(winter)/100 * sum_k w_k HDD(d-k)
           + light * cos(day of year) + weekday + Christmas + noise

    chi_c_by_year: dict year -> % per degree-day above tc.
    chi_h_by_winter: dict winter (starting year) -> % per degree-day below th.
    The deterministic 'light' cycle mimics the hours of darkness (more lighting in winter):
    the ISO-week fixed effect must absorb it.
    """
    from .seasonal import winter_label, christmas
    rng = np.random.default_rng(seed)
    t = temp.sort_values("data").reset_index(drop=True)
    years = t["data"].dt.year
    winters = pd.Series(winter_label(t["data"]))
    if chi_c_by_year is None:
        chi_c_by_year = {a: 5.0 for a in years.unique()}
    if chi_h_by_winter is None:
        chi_h_by_winter = {w: 1.5 for w in winters.unique()}
    cdd = np.maximum(t["tmean"] - tc, 0)
    hdd = np.maximum(th - t["tmean"], 0)
    resp_c = sum(w * cdd.shift(k).fillna(0).values for k, w in enumerate(lag_weights))
    resp_h = sum(w * hdd.shift(k).fillna(0).values for k, w in enumerate(lag_weights))
    chi_c = years.map(chi_c_by_year).fillna(0).values / 100
    chi_h = winters.map(chi_h_by_winter).fillna(0).values / 100
    doy = t["data"].dt.dayofyear.values
    ln_e = (np.log(base) + chi_c * resp_c + chi_h * resp_h
            + light * np.cos(2 * np.pi * doy / 365.25)                   # més llum encesa a l'hivern
            + np.where(t["data"].dt.dayofweek.values >= 5, -0.08, 0.0)
            + christmas_effect * christmas(t["data"])
            + rng.normal(0, sigma, len(t)))
    return pd.DataFrame({"data": t["data"], "kwh": np.exp(ln_e)})


def relay_consumption(temp, on_mean=23.0, on_sd=1.5, width=2.0, n_units=4000, amplitude=0.30,
                      base=1e5, sigma=0.015, seed=5):
    """Daily kWh from a population of air conditioners that behave as relays (Preisach model).

    Unit i switches ON when T > a_i and OFF when T < a_i - width; between the two it keeps
    its state. width > 0 gives true hysteresis; width = 0 gives a smooth, non-linear
    response WITHOUT hysteresis (heterogeneous thresholds only): the null of the fronts test.
    ln E = ln base + amplitude * (fraction of units ON) + weekday + noise.
    """
    rng = np.random.default_rng(seed)
    t = temp.sort_values("data").reset_index(drop=True)
    a = rng.normal(on_mean, on_sd, n_units)       # llindar d'encesa de cada aparell
    b = a - width                                  # llindar d'apagada
    state = np.zeros(n_units, dtype=bool)
    frac = np.empty(len(t))
    for i, T in enumerate(t["tmean"].values):
        state = np.where(T > a, True, np.where(T < b, False, state))
        frac[i] = state.mean()
    ln_e = (np.log(base) + amplitude * frac
            + np.where(t["data"].dt.dayofweek.values >= 5, -0.08, 0.0)
            + rng.normal(0, sigma, len(t)))
    return pd.DataFrame({"data": t["data"], "kwh": np.exp(ln_e), "frac_on": frac})
''',
    'data.py': r'''"""Data access with a local cache (Google Drive in Colab, any folder elsewhere).

Every download is saved to `folder`; the next run reads it from there, so a
server outage or rate limit does not stop the analysis. Delete a file to force
a fresh download.
"""
import os
import pandas as pd


def from_cache(folder, filename, download, date_col="data", until=None, **read_kw):
    """Read folder/filename if it exists and covers `until`; otherwise call download(), save and return."""
    path = os.path.join(folder, filename)
    if os.path.exists(path):
        d = pd.read_csv(path, parse_dates=[date_col], **read_kw)
        if until is None or d[date_col].max() >= pd.Timestamp(until) - pd.Timedelta(days=7):
            print(f"[cau] {filename} llegit ({len(d):,} files)")
            return d
        print(f"[cau] {filename} no arriba fins a {until}: el baixo de nou")
    d = download()
    try:
        os.makedirs(folder, exist_ok=True)
        d.to_csv(path, index=False)
        print(f"[cau] {filename} baixat i desat")
    except OSError as e:
        print(f"[cau] {filename} baixat però no desat ({e})")
    return d


def era5_daily(lat, lon, start, end, timeout=120):
    """Daily ERA5 max and mean 2 m temperature from the Open-Meteo archive API."""
    import requests
    url = ("https://archive-api.open-meteo.com/v1/archive"
           f"?latitude={lat}&longitude={lon}&start_date={start}&end_date={end}"
           "&daily=temperature_2m_max,temperature_2m_mean&timezone=Europe%2FMadrid")
    r = requests.get(url, timeout=timeout)
    r.raise_for_status()
    d = r.json()["daily"]
    return pd.DataFrame({"data": pd.to_datetime(d["time"]),
                         "tmax": d["temperature_2m_max"], "tmean": d["temperature_2m_mean"]})


def era5_hourly(lat, lon, start, end, timeout=180):
    """Hourly ERA5 2 m temperature and boundary-layer height from Open-Meteo."""
    import requests
    url = ("https://archive-api.open-meteo.com/v1/archive"
           f"?latitude={lat}&longitude={lon}&start_date={start}&end_date={end}"
           "&hourly=temperature_2m,boundary_layer_height&timezone=Europe%2FMadrid")
    r = requests.get(url, timeout=timeout)
    r.raise_for_status()
    h = r.json()["hourly"]
    return pd.DataFrame({"hora": pd.to_datetime(h["time"]), "temp": h["temperature_2m"],
                         "blh": h.get("boundary_layer_height")})


def find_column(df, words):
    """First column whose normalised name contains any of `words` (robust to portal renamings)."""
    for c in df.columns:
        name = c.lower().replace("_", "").replace(" ", "")
        if any(w in name for w in words):
            return c
    return None


def daily_consumption(raw):
    """Open Data BCN / Datadis electricity file -> daily kWh by postcode, sector and time band.

    Returns columns: data, cp, sector, tram (if present), kwh.
    """
    col_date = find_column(raw, ["data", "fecha", "date"])
    col_cp = find_column(raw, ["codipostal", "codigopostal", "postal"])
    col_sec = find_column(raw, ["sector", "tipus", "tipo"])
    col_band = find_column(raw, ["tram", "franja", "horari"])
    col_val = find_column(raw, ["valor", "consum", "kwh", "energia"])
    if None in (col_date, col_cp, col_val):
        raise ValueError(f"No trobo les columnes bàsiques: {list(raw.columns)}")
    ren = {col_date: "data", col_cp: "cp", col_val: "kwh"}
    if col_sec:
        ren[col_sec] = "sector"
    if col_band:
        ren[col_band] = "tram"
    d = raw.rename(columns=ren)
    d["data"] = pd.to_datetime(d["data"])
    d["cp"] = d["cp"].astype(str).str.extract(r"(\d{5})")[0]
    keys = [k for k in ["data", "cp", "sector", "tram"] if k in d.columns]
    return d.groupby(keys, as_index=False)["kwh"].sum()
''',
    'alerts.py': r'''"""Data-quality alerts found while building article 1 (see docs/article1.md).

Alert 1: postcode recoding (meters moving between postcodes, zero-sum at city level).
Alert 2: change of regime in the time-band split of consumption since 2024.
"""
import numpy as np
import pandas as pd


def band_regime(daily_bands, evening="18-24", morning="06-12", ref_until=2023, threshold=-2.5):
    """Days whose evening share departs from the reference-weeks climatology.

    daily_bands: DataFrame index = date, columns include the two band names (kWh).
    Share = evening / (evening + morning). Anomaly in percentage points against the
    median of the same ISO week in years <= ref_until. Returns (anomaly Series, set of regime-B days).
    """
    share = daily_bands[evening] / (daily_bands[evening] + daily_bands[morning])
    ref = share[share.index.year <= ref_until]
    clim = ref.groupby(ref.index.isocalendar().week.values).median()
    anom = 100 * (share - clim.reindex(share.index.isocalendar().week.values).values)
    return anom, set(anom.index[anom < threshold])


def complete_days(daily_cp):
    """Days on which every postcode alive in that quarter reports data (city sums are then comparable)."""
    n_day = daily_cp.groupby("data")["cp"].nunique()
    alive = daily_cp.groupby(daily_cp["data"].dt.to_period("Q"))["cp"].nunique()
    expected = pd.Series(alive.reindex(n_day.index.to_period("Q")).values, index=n_day.index)
    return n_day.index[n_day >= expected]


def level_jumps(series, window=60, threshold=0.15):
    """Persistent level shifts of a daily series: log-ratio of window means after/before each date.

    Returns a DataFrame of candidate jump dates with the log-ratio, |log-ratio| > threshold.
    """
    s = np.log(series.sort_index())
    # mitjanes de les finestres anterior (fins al dia d-1) i posterior (des del dia d):
    # el contrast és màxim just al dia del salt
    before = s.rolling(window, min_periods=window).mean().shift(1)
    after = s[::-1].rolling(window, min_periods=window).mean()[::-1]
    jump = (after - before).dropna()
    cand = jump[jump.abs() > threshold]
    if cand.empty:
        return pd.DataFrame(columns=["data", "log_ratio"])
    # una data per salt: el màxim absolut de cada grup de dates seguides
    grp = (cand.index.to_series().diff().dt.days.fillna(1) > 1).cumsum()
    rows = [(g.abs().idxmax(), float(g.loc[g.abs().idxmax()])) for _, g in cand.groupby(grp.values)]
    return pd.DataFrame(rows, columns=["data", "log_ratio"])
''',
}
CARPETA_PAQUET = os.path.join("/content" if os.path.exists("/content") else ".", "ct_paquet", "consum_temperatura")
os.makedirs(CARPETA_PAQUET, exist_ok=True)
for nom, codi in FITXERS_PAQUET.items():
    with open(os.path.join(CARPETA_PAQUET, nom), "w", encoding="utf-8") as f:
        f.write(codi)
sys.path.insert(0, os.path.dirname(CARPETA_PAQUET))
for m in [m for m in sys.modules if m.startswith("consum_temperatura")]:
    del sys.modules[m]                       # si es torna a executar, carrega la versió nova
import consum_temperatura as ct
from consum_temperatura import synthetic as sy
print(f"[OK] Paquet consum_temperatura {ct.__version__}")

# --- Comprovació amb dades sintètiques (paràmetres coneguts) ---
t_s = sy.summer_temperature(seed=0)
c_s = sy.annual_consumption(t_s, th=15.0, chi_h_by_winter={2020: 1.0, 2021: 1.0, 2022: 1.5, 2023: 2.0, 2024: 2.5})
tab_s, _, _ = ct.response_by_winter(c_s, t_s, 15.0, winters=[2021, 2022, 2023, 2024])
print("Sintètic · χ_h per hivern (veritat 1,0 / 1,5 / 2,0 / 2,5):", tab_s["chi_h"].round(2).tolist())
# Test de fronts: mediana de 5 simulacions (una sola pot caure en el ~3 % de falsos positius)
z0, z1 = [], []
for k in range(5):
    t_k = sy.summer_temperature(seed=100 + k)
    z0.append(ct.hysteresis_test(sy.relay_consumption(t_k, width=0.0, n_units=2000, seed=200 + k)[["data", "kwh"]], t_k, 20.0)["z"])
    z1.append(ct.hysteresis_test(sy.relay_consumption(t_k, width=1.0, n_units=2000, seed=200 + k)[["data", "kwh"]], t_k, 20.0)["z"])
z0, z1 = np.median(z0), np.median(z1)
print(f"Sintètic · test de fronts (mediana de 5): sense histèresi z = {z0:+.1f} (ha de ser < 2) · amb histèresi d'1 K z = {z1:+.1f} (ha de ser > 2)")
assert np.abs(tab_s["chi_h"].values - np.array([1.0, 1.5, 2.0, 2.5])).max() < 0.4 and z0 < 2 < z1, \
    "La comprovació sintètica falla: no continuïs i avisa Claude"
print("[OK] Els estimadors recuperen els paràmetres coneguts")
```

### 24a. La V de tot l'any, any per any (mirada descriptiva)

**Què fa:** dibuixa el consum diari de la ciutat contra la temperatura mitjana, després de treure només el dia de la setmana, els festius i el nivell de cada any. Una línia per any, residencial i serveis.

**Què cal mirar:** la forma de la V i si les dues branques s'obren o es tanquen d'un any a l'altre.

**Compte:** és només una mirada. La branca freda barreja temperatura, **llum** (l'hivern és fosc) i Nadal; la calenta barreja temperatura i vacances d'agost. El model de la 24c ho separa.

```python
# ============ 24a. La V de tot l'any, any per any (descriptiu) ============
FESTIUS = sorted(pd.to_datetime(list(festius.keys())))
ONADES = temp.loc[temp["onada"], "data"]
SERIES_24 = {nom: (ciutat_res if nom == "Residencial" else serie_df(sec)) for nom, sec in SECTORS.items()}

def ln_net(serie):
    # ln E sense l'efecte del dia de la setmana, els festius i el nivell de cada any
    v = serie.merge(temp[["data", "tmean"]], on="data")
    v = v[v["kwh"] > 0].copy()
    fe = pd.get_dummies(pd.DataFrame({"dia": v["data"].dt.dayofweek.astype(str),
                                      "festiu": v["data"].isin(FESTIUS).astype(str),
                                      "any": v["data"].dt.year.astype(str)}), drop_first=True).astype(float)
    X = np.column_stack([np.ones(len(v)), fe.values])
    y = np.log(v["kwh"].values)
    b, *_ = np.linalg.lstsq(X, y, rcond=None)
    v["ln_net"] = y - X @ b + b[0]
    return v

fig, ax = plt.subplots(1, len(SERIES_24), figsize=(6.5 * len(SERIES_24), 4.5), squeeze=False)
for a_, (nom, s) in zip(ax[0], SERIES_24.items()):
    v = ln_net(s)
    v["bin"] = np.floor(v["tmean"]).astype(int) + 0.5
    ref = v.loc[v["tmean"].between(16, 19), "ln_net"].mean()      # zona plana de la V
    for a in ANYS:
        m = v[v["data"].dt.year == a].groupby("bin")["ln_net"].agg(["mean", "count"])
        m = m[m["count"] >= 5]
        a_.plot(m.index, 100 * (m["mean"] - ref), "o-", ms=3, label=str(a))
    a_.axhline(0, c="grey", lw=0.7)
    a_.set_xlabel("Temperatura mitjana diària (°C)"); a_.set_ylabel("Consum respecte a 16-19 °C (%)")
    a_.set_title(f"{nom}: la V de cada any"); a_.legend(fontsize=8)
plt.tight_layout(); plt.savefig("V_tot_l_any_v12.png", dpi=200); plt.show()
```

### 24b. Tria del llindar de calefacció $T_h$

**Què fa:** prova llindars de 10 a 19 °C i es queda amb el que minimitza l'error del model d'hivern (la mateixa regla que per a $T_c$ a la secció 15). Un llindar per sector.

**Per què l'error quadràtic mitjà:** el model té una frontissa a $T_h$. Per a cada llindar candidat ajustem el model i mirem quant s'equivoca; el llindar que s'equivoca menys és l'estimació de mínims quadrats del colze de la V (amb errors gaussians, és el de màxima versemblança).

**Què cal mirar:**
- Que el mínim no caigui a la vora (10 o 19 °C). Si hi cau, cal ampliar `TH_CANDIDATS`.
- **Com de profund és el mínim.** Si l'error gairebé no canvia d'un llindar a l'altre (perfil pla), les dades no saben on és el colze: passa quan el sector gairebé no respon al fred. Llavors el mínim és soroll, i fem servir el llindar del residencial (`PERFIL_PLA`).

```python
# ============ 24b. Tria del llindar de calefacció Th ============
PERFIL_PLA = 0.02   # si l'error varia menys d'un 2 % entre tots els llindars, el llindar no està identificat
TH = {}
fig, ax = plt.subplots(1, len(SERIES_24), figsize=(5.5 * len(SERIES_24), 3.3), squeeze=False)
for a_, (nom, s) in zip(ax[0], SERIES_24.items()):     # el residencial va primer
    th_min, prof = ct.choose_th(s, temp, TH_CANDIDATS, LAGS, FESTIUS, HIVERNS)
    variacio = prof["mse"].max() / prof["mse"].min() - 1
    if variacio < PERFIL_PLA and nom != "Residencial":
        TH[nom] = TH["Residencial"]
        print(f"[!] {nom}: perfil pla (l'error només varia un {100 * variacio:.1f} %): el llindar no està identificat. "
              f"Faig servir el del residencial, {TH[nom]} °C (el mínim era {th_min} °C)")
    else:
        TH[nom] = th_min
        print(f"{nom}: l'error varia un {100 * variacio:.0f} % entre llindars → Th = {th_min} °C")
    a_.plot(prof["th"], prof["mse"], "o-"); a_.axvline(TH[nom], ls="--", c="grey")
    a_.set_xlabel("$T_h$ (°C)"); a_.set_ylabel("Error quadràtic mitjà"); a_.set_title(f"{nom}: $T_h$ = {TH[nom]} °C")
    if th_min in (TH_CANDIDATS.min(), TH_CANDIDATS.max()):
        print(f"[!] {nom}: el mínim és a la vora de la graella; amplia TH_CANDIDATS")
plt.tight_layout(); plt.show()
print("Llindars de calefacció (°C):", TH, "· llindar de refrigeració de la ciutat (secció 15): Tc =", TC, "°C")
```

### 24c. La resposta al fred, hivern per hivern

**Què fa:** estima $\chi^0_h$ per a cada hivern (2021-22 a 2024-25) i el canvi entre el primer i l'últim, amb error per clústers de setmana. També dibuixa la funció de resposta $\chi_h(k)$ amb tots els hiverns junts: quina part de la resposta arriba el mateix dia i quina els dies següents.

**Què cal mirar:**
- El **signe**: $\chi_h > 0$ vol dir que el consum puja amb el fred.
- Si el **canvi** entre hiverns és significatiu ($|z| > 2$).
- La **memòria**: a l'hivern, l'edifici triga més a refredar-se que a escalfar-se? Si $\chi_h(1)$ és gran, sí.

```python
# ============ 24c. χ_h per hivern i funció de resposta al fred ============
RES_H = {}
etiqueta = lambda w: f"{w}-{str(w + 1)[-2:]}"
for nom, s in SERIES_24.items():
    tab, b, c = ct.response_by_winter(s, temp, TH[nom], LAGS, HIVERNS, FESTIUS)
    d, se = ct.growth(b, c, HIVERNS[0], HIVERNS[-1], LAGS, prefix="hdd")
    pool = ct.fit_winter(s, temp, TH[nom], LAGS, FESTIUS, HIVERNS)
    RES_H[nom] = {"taula": tab, "creix": (d, se), "pool": pool}
    print(f"\n{nom} (Th = {TH[nom]} °C) · χ_h per hivern (% de consum per grau-dia per sota de Th):")
    for w, r in tab.iterrows():
        print(f"  hivern {etiqueta(w)}: {r.chi_h:5.2f} ± {r.se:.2f}")
    print(f"  Tots junts: {pool['chi_h']:.2f} ± {pool['se']:.2f}  ({pool['n']} dies)")
    print(f"  Canvi {etiqueta(HIVERNS[0])} → {etiqueta(HIVERNS[-1])}: {d:+.2f} ± {se:.2f} %/K  (z = {d / se:+.1f})")

fig, ax = plt.subplots(1, 2, figsize=(12, 4))
for i, (nom, r) in enumerate(RES_H.items()):
    tab = r["taula"]
    ax[0].errorbar(tab.index + 0.05 * i, tab["chi_h"], yerr=2 * tab["se"], fmt="o-", capsize=3, label=nom)
    b, c = r["pool"]["beta"], r["pool"]["cov"]
    k = np.arange(LAGS + 1)
    chi_k = np.array([100 * b[f"hdd{j}"] for j in k])
    se_k = np.array([100 * np.sqrt(c.loc[f"hdd{j}", f"hdd{j}"]) for j in k])
    ax[1].errorbar(k + 0.05 * i, chi_k, yerr=2 * se_k, fmt="o-", capsize=3, label=nom)
ax[0].set_xticks(HIVERNS); ax[0].set_xticklabels([etiqueta(w) for w in HIVERNS])
ax[0].set_ylabel("$\\chi^0_h$ (% / K)"); ax[0].set_title("Resposta al fred per hivern (±2 e.e.)")
ax[1].set_xlabel("Retard k (dies)"); ax[1].set_ylabel("$\\chi_h(k)$ (% / K)"); ax[1].set_title("Funció de resposta al fred (tots els hiverns)")
for a_ in ax:
    a_.axhline(0, c="grey", lw=0.7); a_.legend(fontsize=8)
plt.tight_layout(); plt.savefig("resposta_fred_v12.png", dpi=200); plt.show()
```

### 24d. Simetria de la V: el quocient fred/calor any per any

**Què fa:** per a cada any $y$, compara la resposta a la calor de l'estiu $y$ ($\chi^0_c$, el mateix model de la secció 22d, ara calculat amb el paquet) amb la resposta al fred de l'hivern que comença el novembre de $y$ ($\chi^0_h$), i en calcula el quocient $R=\chi^0_h/\chi^0_c$ amb el seu error.

**Comprovació de consistència:** la $\chi^0_c$ residencial ha de coincidir (a la segona xifra decimal, aproximadament) amb la de la secció 22d. Si no coincideix, avisa: vol dir que el paquet i el notebook no fan el mateix.

**Què cal mirar:** si $R$ creix (la V es torna més simètrica: més electrificació de la calefacció) o es manté (la tendència d'estiu és específica de la refrigeració).

```python
# ============ 24d. Simetria de la V: R = χ_h / χ_c ============
RES_C = {}
for nom, s in SERIES_24.items():
    RES_C[nom], _, _ = ct.response_by_year(s, temp, TC, LAGS, years=ANYS, holidays=FESTIUS, exclude=ONADES)

if "base" in globals():
    print("Consistència · residencial, χ_c per estiu: paquet vs secció 22d")
    for a in ANYS:
        print(f"  {a}: {RES_C['Residencial'].loc[a, 'chi0']:5.2f} vs {base[a]:5.2f}")

files = []
for nom in SERIES_24:
    for w in HIVERNS:
        if w in RES_C[nom].index and w in RES_H[nom]["taula"].index:
            ch, sh = RES_H[nom]["taula"].loc[w, ["chi_h", "se"]]
            cc, sc = RES_C[nom].loc[w, ["chi0", "se"]]
            R_v, sR_v = ct.asymmetry(ch, sh, cc, sc)
            files.append({"sector": nom, "any": w, "chi_c_estiu": cc, "se_c": sc,
                          "chi_h_hivern": ch, "se_h": sh, "R": R_v, "se_R": sR_v})
ASIM = pd.DataFrame(files)
print("\nχ_c de l'estiu y, χ_h de l'hivern que comença el novembre de y, i R = χ_h / χ_c:")
print(ASIM.round(2).to_string(index=False))

fig, ax = plt.subplots(1, 2, figsize=(12, 4))
for i, (nom, g) in enumerate(ASIM.groupby("sector")):
    ax[0].errorbar(g["any"] + 0.05 * i, g["chi_c_estiu"], yerr=2 * g["se_c"], fmt="o-", capsize=3, label=f"{nom}: calor")
    ax[0].errorbar(g["any"] + 0.05 * i, g["chi_h_hivern"], yerr=2 * g["se_h"], fmt="s--", capsize=3, label=f"{nom}: fred")
    ax[1].errorbar(g["any"] + 0.05 * i, g["R"], yerr=2 * g["se_R"], fmt="o-", capsize=3, label=nom)
ax[0].set_ylabel("% de consum per grau-dia"); ax[0].set_title("Les dues branques de la V")
ax[1].axhline(1, c="grey", ls=":", lw=0.8); ax[1].set_ylabel("R = χ_h / χ_c"); ax[1].set_title("Simetria de la V (1 = simètrica)")
for a_ in ax:
    a_.set_xticks(HIVERNS); a_.set_xlabel("Any (estiu y · hivern y-y+1)"); a_.legend(fontsize=8)
plt.tight_layout(); plt.savefig("simetria_V_v12.png", dpi=200); plt.show()
```

### 24e. Robustesa del canvi de la resposta al fred

**Què fa:** repeteix el canvi de $\chi^0_h$ entre el primer i l'últim hivern canviant **una cosa** cada vegada:

| Prova | Què canvia | Què descarta |
|---|---|---|
| R1, R2 | $T_h$ un grau més baix o més alt | Que el resultat depengui del llindar |
| R3 | Sense els dies de Nadal | Les vacances |
| R4 | Només desembre-febrer | Els mesos de transició (novembre i març), amb poca calefacció |
| R5 | Efectes fixos hivern × mes | **Salts de nivell de les dades** dins d'un hivern (com els de la tardor del 2023) |
| R6 | Memòria de 6 dies | Una inèrcia tèrmica més llarga a l'hivern |
| R7 | Només CP sense salts (residencial) | Els CP amb recodificacions |

**Què cal mirar:** si el signe i l'ordre de magnitud del canvi es mantenen en totes les proves. Si una prova el fa desaparèixer, aquella és (part de) l'explicació.

```python
# ============ 24e. Robustesa del canvi de χ_h ============
def canvi_h(s, th, lags=LAGS, **kw):
    tab, b, c = ct.response_by_winter(s, temp, th, lags, HIVERNS, FESTIUS, **kw)
    d, se = ct.growth(b, c, HIVERNS[0], HIVERNS[-1], lags, prefix="hdd")
    return {"primer": tab["chi_h"].iloc[0], "últim": tab["chi_h"].iloc[-1], "canvi": d, "se": se, "z": d / se}

_dies = pd.date_range(f"{min(ANYS)}-01-01", f"{max(ANYS)}-12-31")
NADAL = _dies[ct.christmas(_dies)]
ROB_H = []
for nom, s in SERIES_24.items():
    th = TH[nom]
    proves_h = {"R0 base": canvi_h(s, th),
              f"R1 Th = {th - 1}": canvi_h(s, th - 1),
              f"R2 Th = {th + 1}": canvi_h(s, th + 1),
              "R3 sense Nadal": canvi_h(s, th, exclude=NADAL),
              "R4 només des.-feb.": canvi_h(s, th, months=(12, 1, 2)),
              "R5 efectes fixos hivern × mes": canvi_h(s, th, winter_month=True),
              "R6 memòria de 6 dies": canvi_h(s, th, lags=6)}
    if nom == "Residencial" and "sense_salts" in globals():
        proves_h[f"R7 només CP sense salts ({len(sense_salts)})"] = canvi_h(serie_df(sector_triat, cps=sense_salts), th)
    for p, r in proves_h.items():
        ROB_H.append({"sector": nom, "prova": p, **r})
ROB_H = pd.DataFrame(ROB_H)
print(f"Canvi de χ_h entre l'hivern {etiqueta(HIVERNS[0])} i el {etiqueta(HIVERNS[-1])} (%/K):")
print(ROB_H.round(2).to_string(index=False))
```

### 24f. A quina hora del dia respon el consum al fred?

**Què fa:** estima $\chi^0_h$ per a cada franja horària del CSV (00-06, 06-12, 12-18, 18-24 h), amb els hiverns 2021-22 i 2022-23 junts.

**Per què només aquests dos hiverns:** l'alerta 2 (canvi en el repartiment entre franges) comença el 2024. Amb els hiverns anteriors, les franges són fiables.

**Què cal mirar:** a l'estiu la resposta és sobretot de vespre i de nit. A l'hivern esperem un perfil diferent: matí (engegar la calefacció en llevar-se) i vespre (tornar a casa).

```python
# ============ 24f. χ_h per franja horària (hiverns abans de l'alerta 2) ============
HIVERNS_A = [w for w in HIVERNS if w <= 2022]
files = []
for nom, sec in SECTORS.items():
    for tr in FRANGES:
        s = serie_df(sec, tram=tr)
        try:
            r = ct.fit_winter(s, temp, TH[nom], LAGS, FESTIUS, HIVERNS_A)
            files.append({"sector": nom, "franja": NOM[tr], "chi_h": r["chi_h"], "se": r["se"], "dies": r["n"]})
        except Exception as e:
            print(f"  [!] {nom} {NOM[tr]}: {e}")
FRANJA_H = pd.DataFrame(files)
print(f"χ_h per franja (hiverns {', '.join(etiqueta(w) for w in HIVERNS_A)}), % per grau-dia:")
print(FRANJA_H.round(2).to_string(index=False))

fig, ax = plt.subplots(figsize=(7, 3.8))
for i, (nom, g) in enumerate(FRANJA_H.groupby("sector")):
    x = np.arange(len(g)) + 0.35 * i
    ax.bar(x, g["chi_h"], width=0.35, yerr=2 * g["se"], capsize=3, label=nom)
ax.set_xticks(np.arange(len(FRANGES)) + 0.175); ax.set_xticklabels([NOM[t] for t in FRANGES])
ax.axhline(0, c="grey", lw=0.7); ax.set_ylabel("$\\chi^0_h$ (% / K)"); ax.set_title("Resposta al fred per franja horària")
ax.legend(fontsize=8); plt.tight_layout(); plt.savefig("fred_per_franja_v12.png", dpi=200); plt.show()
```

### 24g. Com llegir la secció 24

**Lectures possibles del canvi de $\chi_h$** (proposta de Claude, per discutir):
- **$\chi_h$ creix i $R$ s'acosta a 1:** electrificació de la calefacció (bombes de calor). Seria coherent amb que una part del creixement d'estiu vingui de més aparells reversibles (marge extensiu).
- **$\chi_h$ estable i $\chi_c$ creix:** la tendència d'estiu és específica de la refrigeració: més ús de l'aire condicionat (marge intensiu) o aparells només de fred.
- **$\chi_h$ baixa el 2022-23:** compatible amb l'estalvi forçat per la crisi de preus (la secció 22f va trobar que el residencial d'estiu no respon al preu, però l'hivern pot ser diferent: cal mirar-ho abans d'afirmar-ho).

**Què NO es pot concloure:** quina fracció de la calefacció és elèctrica (no veiem el gas); que les bombes de calor siguin la causa (cal una dada de penetració; els certificats energètics del pilot tenen el sistema de calefacció declarat).

## 25. Histèresi: la ciutat recorda si la calor arriba o se'n va?

**Pregunta.** A la mateixa temperatura, consumim més quan la calor **se'n va** (la temperatura baixa respecte d'ahir) que quan **arriba** (puja)? Si sí, una part del sistema té **histèresi**: l'estat depèn del camí, no només de la temperatura i de la memòria lineal.

**Dues explicacions de per què podria passar:**
- **Memòria lineal (inèrcia tèrmica):** després de dies calents, l'edifici encara és calent i l'aire treballa més. Això ja ho descriu la funció de resposta amb retards $\chi(k)$: no és histèresi.
- **Histèresi veritable (relés):** un aparell s'encén quan la temperatura passa de $T_\text{on}$ i no s'apaga fins que baixa de $T_\text{off} < T_\text{on}$ (hàbit, termòstat, «ja que l'he encès…»). A la mateixa temperatura, quan la calor baixa hi ha més aparells encesos que quan puja.

**El test de fronts.** Ajustem
$$\ln E(d) = \underbrace{\sum_{k=0}^{3} f_k\big(T(d-k)\big)}_{\text{memòria flexible}} + \gamma\,\mathrm{CDD}(d)\,\mathbf{1}\big[T(d)<T(d-1)\big] + \text{efectes fixos} + \varepsilon$$
- $f_k$ són corbes (frontisses a $T_c$, $T_c+2$, $T_c+4$ i $T_c+6$ °C per a cada retard): qualsevol memòria **additiva**, lineal o corbada, hi cap.
- La histèresi **no** és additiva, i és l'única cosa que pot fer $\gamma>0$.

**Per què la memòria ha de ser flexible:** amb la memòria lineal de la resta del notebook, una resposta simplement **corbada** (sense histèresi) ja dona $\gamma>0$, perquè el model s'equivoca diferent els dies que baixa la temperatura. Ho vam comprovar amb dades sintètiques; la versió lineal es mostra només com a referència.

**Calibratge amb dades sintètiques (40 simulacions per cas, 5 estius):**

| Cas simulat | Probabilitat de $z>2$ |
|---|---|
| Resposta lineal amb memòria (sense histèresi) | 0 % |
| Resposta corbada sense histèresi | 3 % |
| Relés amb histèresi d'1 K | 100 % |
| Relés amb histèresi de 2 K | 62 % |
| Relés amb histèresi de 3 K | 10 % |

**Límit del test:** una histèresi ampla (3 K o més) es confon amb memòria amb dades diàries i **no es detecta**. Un resultat nul vol dir «no hi ha histèresi estreta», no «no hi ha histèresi». El test és d'una cua ($\gamma>0$), perquè és el que prediu la teoria dels relés. Un $\gamma<0$ clar seria una sorpresa interessant (per exemple, anticipar-se a la calor amb la previsió del temps).

### 25a. El test de fronts a la ciutat sencera

**Què fa:** aplica el test amb tots els estius (2021-2025), incloent-hi els dies d'onada (el final d'una onada és justament un front de baixada). Residencial i serveis.

**Què cal mirar:** $\gamma$ (en % per grau-dia de més, els dies que la temperatura baixa), $z$ i el valor $p$ d'una cua. La fila «memòria lineal» és la versió esbiaixada: si dona molt més que la flexible, confirma que la corba de resposta és corbada.

```python
# ============ 25a. Test de fronts (histèresi) ============
RES_25 = []
for nom, s in SERIES_24.items():
    for etiq, kw in [("memòria flexible", {}), ("memòria lineal (esbiaixada, referència)", {"linear": True})]:
        h = ct.hysteresis_test(s, temp, TC, LAGS, FESTIUS, MESOS_ESTIU, ANYS, **kw)
        RES_25.append({"sector": nom, "model": etiq, "gamma": h["gamma"], "se": h["se"], "z": h["z"],
                       "p_una_cua": 1 - stats.norm.cdf(h["z"]), "dies": h["n"], "frac_dies_calents_baixada": h["frac_baixa"]})
RES_25 = pd.DataFrame(RES_25)
print("γ: % de consum de més per grau-dia, els dies que la temperatura baixa respecte d'ahir")
print(RES_25.round(3).to_string(index=False))
```

### 25b. Robustesa del test (residencial) i canvi en el temps

**Què fa:** repeteix el test canviant una cosa cada vegada (frontisses, memòria de 5 dies, sense onades, sense el 2025, $T_c \pm 1$) i el fa per separat per als estius 2021-22 i 2023-24.

**Què cal mirar:** si $\gamma$ és estable. La comparació 2021-22 vs 2023-24 respon a una pregunta nova: **si hi ha histèresi i creix, encaixa amb més aparells que s'encenen i es deixen encesos** (la tendència de la secció 22).

```python
# ============ 25b. Robustesa del test de fronts i canvi en el temps ============
s = SERIES_24["Residencial"]
ANYS_SENSE_2025 = [a for a in ANYS if a != 2025]
proves_25 = {
    "P0 base": dict(),
    "P1 frontisses cada 1 K (Tc..Tc+6)": dict(knots=(0, 1, 2, 3, 4, 5, 6)),
    "P2 memòria de 5 dies": dict(lags=5),
    "P3 sense els dies d'onada": dict(exclude=ONADES),
    "P4 sense el 2025": dict(years=ANYS_SENSE_2025),
    f"P5 Tc = {TC - 1}": dict(tc=TC - 1),
    f"P6 Tc = {TC + 1}": dict(tc=TC + 1),
    "P7 estius 2021-2022": dict(years=[2021, 2022]),
    "P8 estius 2023-2024": dict(years=[2023, 2024]),
}
ROB_25 = []
for p, kw in proves_25.items():
    kw = dict(kw)
    tc_ = kw.pop("tc", TC); lags_ = kw.pop("lags", LAGS); anys_ = kw.pop("years", ANYS)
    h = ct.hysteresis_test(s, temp, tc_, lags_, FESTIUS, MESOS_ESTIU, anys_, **kw)
    ROB_25.append({"prova": p, "gamma": h["gamma"], "se": h["se"], "z": h["z"], "dies": h["n"]})
ROB_25 = pd.DataFrame(ROB_25)
print(ROB_25.round(3).to_string(index=False))
g0 = ROB_25.set_index("prova").loc["P7 estius 2021-2022"]; g1 = ROB_25.set_index("prova").loc["P8 estius 2023-2024"]
dg, sdg = g1["gamma"] - g0["gamma"], np.hypot(g0["se"], g1["se"])
print(f"\nCanvi de γ (2023-24 menys 2021-22): {dg:+.3f} ± {sdg:.3f} (z = {dg / sdg:+.1f})")
```

### 25c. Histèresi per franja horària

**Què fa:** el test de fronts per a cada franja horària del residencial, amb els estius 2021-2023 (abans de l'alerta 2).

**Què cal mirar:** si hi ha histèresi, on esperem trobar-la és **al vespre i a la nit** (l'aire del dormitori que es deixa encès), que és on creix la resposta (secció 23).

```python
# ============ 25c. Test de fronts per franja (residencial, estius 2021-2023) ============
ANYS_A = [a for a in ANYS if a <= 2023]
files = []
for tr in FRANGES:
    h = ct.hysteresis_test(serie_df(sector_triat, tram=tr), temp, TC, LAGS, FESTIUS, MESOS_ESTIU, ANYS_A)
    files.append({"franja": NOM[tr], "gamma": h["gamma"], "se": h["se"], "z": h["z"], "dies": h["n"]})
FRANJA_25 = pd.DataFrame(files)
print(FRANJA_25.round(3).to_string(index=False))
```

### 25d. Les tres fases de l'estiu (complement descriptiu)

**Què fa:** estima $\chi^0$ per separat a l'**arribada** (1 de juny a 15 de juliol), la **plena** (16 de juliol a 15 d'agost) i la **sortida** de l'estiu (16 d'agost a 30 de setembre), amb efectes fixos de setmana.

**Compte:** la diferència entre fases **no** és un test d'histèresi: l'agost canvia l'ocupació, i el setembre té menys hores de llum. Serveix per lligar-ho amb el resultat de la secció 22 («juny + setembre no creixen; juliol + agost sí»).

```python
# ============ 25d. χ⁰ per fase de l'estiu ============
for nom, s in SERIES_24.items():
    tab, (d, se) = ct.response_by_phase(s, temp, TC, LAGS, FESTIUS, ANYS, exclude=ONADES)
    print(f"\n{nom} · χ⁰ per fase de l'estiu (%/K):")
    for f_, r in tab.iterrows():
        print(f"  {f_:9s}: {r.chi0:5.2f} ± {r.se:.2f}")
    print(f"  Sortida menys arribada: {d:+.2f} ± {se:.2f} (z = {d / se:+.1f})")
```

### 25e. El bucle, vist

**Què fa:** treu del consum la memòria flexible (sense el terme $\gamma$) i dibuixa el residu mitjà per graus de temperatura, separant els dies en què la temperatura **puja** dels dies en què **baixa**.

**Què cal mirar:** si les dues corbes se separen per sobre de $T_c$ (la de baixada per sobre), és la histèresi que mesura $\gamma$. Si es creuen o se superposen, no n'hi ha (o és massa ampla per veure-la).

```python
# ============ 25e. El bucle: residu per temperatura, dies de pujada vs de baixada ============
from consum_temperatura.response import prepare as ct_prepare, design_matrix as ct_disseny, fit_cluster as ct_ajust
tab25 = ct_prepare(SERIES_24["Residencial"], temp, TC, LAGS, years=ANYS)
X25 = ct_disseny(tab25, LAGS, years=ANYS, holidays=FESTIUS).drop(columns=[f"cdd{k}" for k in range(LAGS + 1)])
X25 = pd.concat([X25, ct.flexible_memory(tab25, temp, TC, LAGS)], axis=1)
_, _, res25 = ct_ajust(tab25, X25)
baixa = ct.falling_days(temp).reindex(tab25["data"].values).fillna(False).values.astype(bool)
g = pd.DataFrame({"bin": np.floor(tab25["tmean"].values) + 0.5, "baixa": baixa, "res": 100 * res25})
m = g.groupby(["bin", "baixa"])["res"].agg(["mean", "sem", "count"]).reset_index()
m = m[m["count"] >= 8]
fig, ax = plt.subplots(figsize=(7, 4))
for val, lab, mk in [(False, "la temperatura puja", "o"), (True, "la temperatura baixa", "s")]:
    q = m[m["baixa"] == val]
    ax.errorbar(q["bin"], q["mean"], yerr=2 * q["sem"], fmt=mk + "-", capsize=3, label=lab)
ax.axhline(0, c="grey", lw=0.7); ax.axvline(TC, c="grey", ls="--", lw=0.8)
ax.set_xlabel("Temperatura mitjana del dia (°C)"); ax.set_ylabel("Residu de la memòria flexible (%)")
ax.set_title("Residencial: dies de pujada vs de baixada (±2 e.e.)"); ax.legend(fontsize=8)
plt.tight_layout(); plt.savefig("bucle_histeresi_v12.png", dpi=200); plt.show()
```

### 25f. Com llegir la secció 25

- **$\gamma \approx 0$ en totes les proves:** la resposta de la ciutat és una memòria (lineal o corbada) sense histèresi estreta. És coherent amb el resultat de les onades (residu ≈ 0) i reforça el model lineal de l'article 1. Tanca la pregunta de la consulta C3 en el sentit de «*crossover* sense salts».
- **$\gamma > 0$ estable, sobretot de nit:** hi ha aparells que es deixen encesos (relés amb histèresi). Seria un resultat nou: la resposta de la ciutat depèn del camí. Si a més creix entre 2021-22 i 2023-24, encaixa amb la tendència.
- **$\gamma < 0$:** inesperat. Primer cal descartar un artefacte (per exemple, l'efecte del dia de la setmana sobre els fronts) abans d'interpretar-lo.

**Què NO es pot concloure:** que no hi hagi histèresi ampla (el test no la veu); quin aparell concret la produeix.

```python
# ============ 25g. Desar els resultats de la secció 25 ============
RES_25.to_csv("histeresi_fronts_v12.csv", index=False)
ROB_25.to_csv("histeresi_robustesa_v12.csv", index=False)
FRANJA_25.to_csv("histeresi_per_franja_v12.csv", index=False)
print("[OK] Desats: histeresi_fronts_v12.csv, histeresi_robustesa_v12.csv, histeresi_per_franja_v12.csv")
```

## 27. Exportar les dades perquè Claude pugui iterar l'anàlisi

**Tesi:** física dels sistemes complexos aplicada a les ciutats · línia B: resposta de la ciutat a la calor

**Què fa aquest notebook**
1. Descarrega el consum elèctric **diari per codi postal** (Open Data BCN, font Datadis, 2020-2025).
2. Descarrega la temperatura diària: ERA5 (Open-Meteo, una sola sèrie per a tota la ciutat) i les estacions del Meteocat, amb les quals calcula una temperatura local per a cada CP.
3. Estima, per a cada codi postal, la **resposta lineal** del consum a la calor, amb memòria de fins a 3 dies:
$$\delta \ln E_i(d) = \sum_{k=0}^{L} \chi_i(k)\,\mathrm{CDD}(d-k) + \text{efectes fixos} + \varepsilon, \qquad \mathrm{CDD}(d)=\max\big(T(d)-T_c,\,0\big)$$
4. Ajusta el model **només amb dies d'estiu normals** i comprova si **prediu els dies d'onada de calor**. Si no els prediu, la resposta deixa de ser lineal: és el diagnòstic de la peça (c) de la tesi.

**Què NO fa (encara)**
- No hi ha termes creuats $\chi_{ij}$ ($i\neq j$): la calor d'un barri que afecta el consum del veí. Amb tres estacions, les temperatures dels CP estan gairebé perfectament correlacionades i no es poden separar.
- No estima el guany del bucle $G$ (cal separar la temperatura externa de la calor antropogènica).
- Els errors estàndard són aproximats (els residus diaris estan autocorrelacionats): són optimistes. Pas següent: *block bootstrap*.
- La memòria té 4 coeficients lliures (retards 0-3); encara no imposem una forma física (nucli exponencial $\propto e^{-\tau/\tau_m}$).

**Com usar-lo:** `Entorn d'execució → Executa-ho tot`. Si la descàrrega automàtica falla, mira la secció 2.

## A1. Mapa de la susceptibilitat $\chi^0$ per codi postal
OSM no té els límits dels codis postals de Barcelona: els **reconstruïm a partir de les adreces** (cada punt s'assigna a l'adreça més propera i s'ajunten les cel·les del mateix CP). Són límits aproximats. **Un CP no és un barri.**

```python
mapa = res_cp.copy()
mapa["chi_pct"] = 100 * mapa["chi_estatica"]
mapa["signif"] = (mapa["chi_estatica"] / mapa["chi_estatica_se"]).abs() > 2

# ---------- 4) Mapa ----------
g = pol.merge(mapa, on="cp").to_crs(epsg=3857)       # 3857: projecció dels mapes web
vmin, vmax = 0, g["chi_pct"].quantile(0.95)

fig, ax = plt.subplots(figsize=(11, 11))
g.plot(ax=ax, column="chi_pct", cmap="YlOrRd", vmin=vmin, vmax=vmax,
       edgecolor="white", linewidth=0.8, legend=True,
       legend_kwds={"label": "χ⁰  (% de consum per K)", "shrink": 0.55})
g[~g["signif"]].plot(ax=ax, facecolor="none", edgecolor="grey", hatch="///", linewidth=0.6)
bcn.to_crs(epsg=3857).boundary.plot(ax=ax, color="black", linewidth=0.8)
for _, r in g.iterrows():
    p = r.geometry.representative_point()
    ax.annotate(r["cp"][-2:], (p.x, p.y), ha="center", va="center", fontsize=8)

FONS = False   # True = provar el fons gris d'Esri. Els servidors d'OSM i CartoDB bloquegen Colab.
if FONS:
    try:
        cx.add_basemap(ax, source=cx.providers.Esri.WorldGrayCanvas, alpha=0.6)
    except Exception as e:
        print(f"[!] Sense mapa de fons ({e})")
ax.set_axis_off()
ax.set_title("Barcelona: susceptibilitat del consum elèctric residencial a la calor\n"
             f"χ⁰ per codi postal (estius 2021-2025, temperatura: {FONT_TEMP_CP}, Tc = {TC_CP} °C). Ratllat: no significatiu.\n"
             "Límits de CP aproximats a partir d'adreces d'OpenStreetMap", fontsize=11)
plt.tight_layout()
plt.savefig("mapa_chi_bcn_v12.png", dpi=200, bbox_inches="tight")
plt.show()

g.to_crs(epsg=4326).to_file("chi_per_cp_v12.geojson", driver="GeoJSON")
try:
    from google.colab import files
    files.download("mapa_chi_bcn_v12.png")
except Exception:
    pass
```

## A2. A quins barris corresponen els CP?
Creuem les adreces de cada CP amb els límits de barris i districtes (OSM). Comptem adreces, no superfície. Taula ordenada per $\chi^0$.

```python
# ============ A2. A quins barris corresponen els CP? Taula ordenada per χ⁰ ============
# Creuem les ADRECES de cada CP ('adr', ja a la memòria) amb els polígons dels barris (OSM o, si falla, el fitxer oficial).
# Comptem adreces i no superfície: així Montjuïc o el port no pesen per la seva mida.

from shapely import wkt

if adr is not None:
    # ---------- 1) Polígons de barris (admin_level=10) i districtes (admin_level=9) d'OSM ----------
    def limits_osm(nivell):
        f = ox.features_from_place(LLOC, tags={"admin_level": str(nivell)})
        f = f[(f["admin_level"] == str(nivell)) & f.geometry.geom_type.isin(["Polygon", "MultiPolygon"]) & f["name"].notna()]
        return f[["name", "geometry"]].reset_index(drop=True)

    barris = limits_osm(10)
    districtes = limits_osm(9)
    print(f"Barris trobats a OSM: {len(barris)} (Barcelona en té 73) · districtes: {len(districtes)} (en té 10)")

    # Alternativa oficial si OSM falla: baixa a mà el CSV de barris d'Open Data BCN
    # (conjunt "20170706-districtes-barris"), puja'l a dades_tesi i posa'n el nom aquí:
    FITXER_BARRIS = None   # p. ex. "/content/drive/MyDrive/dades_tesi/BarcelonaCiutat_Barris.csv"
    if FITXER_BARRIS:
        t = pd.read_csv(FITXER_BARRIS)
        col_geo = [c for c in t.columns if "wgs84" in c.lower()][0]
        col_nom = [c for c in t.columns if "nom_barri" in c.lower()][0]
        barris = gpd.GeoDataFrame({"name": t[col_nom]}, geometry=t[col_geo].apply(wkt.loads), crs="EPSG:4326")
        print(f"Barris del fitxer oficial: {len(barris)}")

    # ---------- 2) Cada adreça -> barri i districte ----------
    pts = adr[["cp", "geometry"]].to_crs(epsg=4326)
    pts = gpd.sjoin(pts, barris.to_crs(epsg=4326).rename(columns={"name": "barri"}), predicate="within", how="left").drop(columns="index_right")
    pts = gpd.sjoin(pts, districtes.to_crs(epsg=4326).rename(columns={"name": "districte"}), predicate="within", how="left").drop(columns="index_right")

    def resum(g, col, llindar=0.15):
        # Barris (o districtes) que tenen almenys el 15 % de les adreces del CP, amb el seu percentatge
        q = g[col].value_counts(normalize=True)
        q = q[q >= llindar]
        return ", ".join(f"{n} ({100*v:.0f}%)" for n, v in q.items())

    taula = (pts.groupby("cp")
                [["barri", "districte"]]
                .apply(lambda g: pd.Series({"barris_principals": resum(g, "barri"),
                                            "districte": resum(g, "districte", 0.3)}))
                .reset_index())
else:
    # v0.11: sense adreces (memòria cau sense adreces_cp.csv): taula de barris d'una execució anterior
    pts = None
    taula = pd.read_csv(f"{CARPETA_LOCAL}/barris_cp.csv", dtype={"cp": str})
    print(f"[i] Barris llegits de {CARPETA_LOCAL}/barris_cp.csv ({len(taula)} CP)")

# ---------- 3) Unim amb χ⁰ i ordenem ----------
taula = taula.merge(mapa[["cp", "chi_pct", "chi_estatica_se", "signif"]], on="cp")
taula["interval_95"] = taula.apply(lambda r: f"{r.chi_pct - 200*r.chi_estatica_se:.1f} a {r.chi_pct + 200*r.chi_estatica_se:.1f}", axis=1)
taula = taula.sort_values("chi_pct", ascending=False).reset_index(drop=True)
taula.index = taula.index + 1   # rànquing 1..42
taula["chi_pct"] = taula["chi_pct"].round(1)
taula["signif"] = taula["signif"].map({True: "sí", False: "NO"})

pd.set_option("display.max_colwidth", 80)
print(taula[["cp", "chi_pct", "interval_95", "signif", "districte", "barris_principals"]].to_string())
taula.to_csv("chi_per_cp_i_barris_v12.csv")
try:
    from google.colab import files
    files.download("chi_per_cp_i_barris_v12.csv")
except Exception:
    pass
```

## A3. Test de la hipòtesi: $\chi^0$ i renda
Hipòtesi de treball: $\chi^0 \sim$ capacitat de climatitzar (renda) $\times$ escalfament local (illa de calor). Primer test, només amb la renda. **Cal baixar a mà** el CSV de renda per càpita d'Open Data BCN i posar-ne el camí a `FITXER_RENDA`.

```python
# ============ A3. TEST DE LA HIPÒTESI: χ⁰ i renda per codi postal ============
# Hipòtesi: χ⁰ ~ capacitat de climatitzar (renda) × escalfament local (illa de calor).
# Primer test, només amb la renda: esperem que χ⁰ creixi amb la renda (amb excepcions com la Bonanova).
# Fa servir 'pts' (secció A2: cada adreça amb el seu barri) i 'mapa' (secció A1).
#
# DADES (cal baixar-les a mà, el portal bloqueja les descàrregues automàtiques):
#   Open Data BCN -> "Renda disponible de les llars per càpita" (renda-disponible-llars-bcn)
#   Baixa el CSV de l'any més recent i puja'l a la carpeta dades_tesi.
import unicodedata
from scipy import stats

FITXER_RENDA = "/content/drive/MyDrive/dades_tesi/2023_renda_disponible_llars_per_persona.csv"   # <- posa-hi el nom exacte del fitxer

def normalitza(nom):
    # minúscules, sense accents i amb un sol tipus d'apòstrof: així "l'Eixample" i "l’Eixample" coincideixen
    nom = str(nom).replace("’", "'").replace("`", "'")
    nom = unicodedata.normalize("NFKD", nom).encode("ascii", "ignore").decode()
    return " ".join(nom.lower().split())

col_valor = "renda per càpita"
if pts is not None:
    r = llegeix_csv_flexible(FITXER_RENDA)
    print("Columnes del fitxer:", list(r.columns))
    col_barri = [c for c in r.columns if "nom_barri" in c.lower()][0]
    col_valor = [c for c in r.columns if any(p in c.lower() for p in ["import", "euros", "valor", "renda"])
                 and pd.api.types.is_numeric_dtype(pd.to_numeric(r[c].astype(str).str.replace(",", "."), errors="coerce"))][-1]
    print(f"Faig servir: barri = '{col_barri}', renda = '{col_valor}'")
    r[col_valor] = pd.to_numeric(r[col_valor].astype(str).str.replace(",", "."), errors="coerce")

    # Si el fitxer ve per secció censal, en fem la mitjana per barri
    renda_barri = r.groupby(r[col_barri].map(normalitza))[col_valor].mean().rename("renda")
    print(f"Barris amb renda: {len(renda_barri)} · de {renda_barri.min():,.0f} a {renda_barri.max():,.0f} €")

    # ---------- Renda de cada CP = mitjana de la renda del barri de cada adreça ----------
    p = pd.DataFrame({"cp": pts["cp"].values, "barri": pts["barri"].map(normalitza).values})
    p["renda"] = p["barri"].map(renda_barri)
    sense = p.loc[p["renda"].isna() & p["barri"].ne("nan"), "barri"].unique()
    if len(sense):
        print("  [!] Barris d'OSM sense renda al fitxer (noms diferents?):", list(sense)[:15])
    renda_cp = p.groupby("cp")["renda"].mean()
else:
    # v0.11: sense adreces, renda per CP d'una execució anterior
    renda_cp = pd.read_csv(f"{CARPETA_LOCAL}/renda_cp.csv", dtype={"cp": str}).set_index("cp")["renda"]
    print(f"[i] Renda per CP llegida de {CARPETA_LOCAL}/renda_cp.csv ({len(renda_cp)} CP)")

t = mapa.set_index("cp")[["chi_pct", "chi_estatica_se", "signif"]].join(renda_cp).dropna()
t = t[t["signif"]]   # només CP amb χ⁰ significatiu
rho, p_rho = stats.spearmanr(t["renda"], t["chi_pct"])
reg = stats.linregress(t["renda"] / 1000, t["chi_pct"])
print(f"\nCP analitzats: {len(t)}")
print(f"Correlació de rangs (Spearman): ρ = {rho:.2f}  (p = {p_rho:.3f})")
print(f"Pendent: {reg.slope:.2f} %/K per cada 1.000 € de renda  (R² = {reg.rvalue**2:.2f})")

# ---------- Gràfic ----------
fig, ax = plt.subplots(figsize=(8, 6))
ax.errorbar(t["renda"], t["chi_pct"], yerr=200 * t["chi_estatica_se"], fmt="o", ms=5,
            alpha=0.7, capsize=2, elinewidth=0.7)
x = np.linspace(t["renda"].min(), t["renda"].max(), 50)
ax.plot(x, reg.intercept + reg.slope * x / 1000, "k--", lw=1, label=f"ajust lineal (ρ = {rho:.2f})")
for cp_, row in t.iterrows():
    ax.annotate(cp_[-2:], (row["renda"], row["chi_pct"]), fontsize=7, xytext=(3, 3), textcoords="offset points")
ax.set_xlabel(f"Renda ({col_valor}, mitjana dels barris del CP)")
ax.set_ylabel("χ⁰  (% de consum per K)")
ax.set_title("Susceptibilitat a la calor vs renda, per codi postal")
ax.legend(); plt.tight_layout()
plt.savefig("chi_vs_renda_v12.png", dpi=200); plt.show()

# CP que més s'aparten de la recta: els candidats a explicar amb l'illa de calor o el parc edificat
t["desviacio"] = t["chi_pct"] - (reg.intercept + reg.slope * t["renda"] / 1000)
print("\nMés susceptibles del que tocaria per renda:")
print(t.sort_values("desviacio", ascending=False).head(5)[["chi_pct", "renda", "desviacio"]].round(1).to_string())
print("\nMenys susceptibles del que tocaria per renda:")
print(t.sort_values("desviacio").head(5)[["chi_pct", "renda", "desviacio"]].round(1).to_string())
```

## A4. Calor antropogènica: els edificis no residencials escalfen els barris?

Tota l'electricitat que es consumeix acaba convertida en calor (aparells, il·luminació, i la calor que l'aire condicionat treu a fora). Amb el consum de **tots els sectors** per CP obtenim un flux de calor antropogènica $Q_F$ (W/m²). Si la calor dels comerços i oficines escalfa el barri, esperaríem $\chi^0$ residencial més alt on $Q_F^{\text{no res}}$ és alt.

**Compte:** és una correlació entre barris. Els barris densos tenen alhora més $Q_F$, més illa de calor per la morfologia i rendes diferents: no prova causalitat. L'àrea del CP inclou parcs i muntanya, i en aquests CP $Q_F$ surt més baix.

**v0.9:** aquí només es fa servir la calor dels edificis **no** residencials per *explicar* la χ residencial (posar-hi el residencial seria circular). La calor antropogènica **total** i la seva resposta a la temperatura, que són les que importen per al bucle G, són a la **secció 22j**.

```python
from scipy import stats
# Calor antropogènica dels edificis: tota l'electricitat que es consumeix acaba convertida en calor.
# Consum mitjà d'estiu (kWh/dia) / àrea del CP (m²)  ->  flux de calor Q_F (W/m²).   kWh/dia × 1000/24 = W
s_all = df[df["data"].isin(ciutat["data"]) & df["data"].dt.month.isin(MESOS_ESTIU)]
per_sector = (s_all.groupby(["cp", "sector", "data"])["kwh"].sum()      # diari per CP i sector
                   .groupby(["cp", "sector"]).mean().unstack(fill_value=0))  # mitjana d'estiu
print("Sectors:", list(per_sector.columns))

qf = pd.DataFrame({
    "qf_residencial": per_sector[sector_triat] * 1000 / 24 / area_cp,
    "qf_no_residencial": per_sector.drop(columns=[sector_triat]).sum(axis=1) * 1000 / 24 / area_cp,
}).dropna()
qf["qf_total"] = qf.sum(axis=1)
print("\nFlux de calor antropogènica elèctric a l'estiu (W/m², mitjana del CP):")
print(qf.describe().loc[["min", "50%", "max"]].round(2).to_string())

# Relació amb χ⁰ (només CP significatius)
q = mapa.set_index("cp")[["chi_pct", "signif"]].join(qf, how="inner")
q = q[q["signif"]]
if "renda_cp" in globals():
    q = q.join(renda_cp.rename("renda"))
rho, p_rho = stats.spearmanr(q["qf_no_residencial"], q["chi_pct"])
print(f"\nχ⁰ vs Q_F no residencial: Spearman ρ = {rho:.2f} (p = {p_rho:.3f}), {len(q)} CP")

# Si tenim la renda, mirem si Q_F aporta alguna cosa un cop descomptada la renda (regressió múltiple)
if "renda" in q.columns and q["renda"].notna().sum() > 10:
    qq = q.dropna(subset=["renda"])
    X = np.column_stack([np.ones(len(qq)), qq["renda"] / 1000, np.log(qq["qf_no_residencial"])])
    b, *_ = np.linalg.lstsq(X, qq["chi_pct"].values, rcond=None)
    res = qq["chi_pct"].values - X @ b
    s2 = res @ res / (len(qq) - 3)
    se = np.sqrt(np.diag(s2 * np.linalg.inv(X.T @ X)))
    print("Regressió χ⁰ = a + b·renda(k€) + c·ln Q_F,no res:")
    for nom, bi, si in zip(["a", "b (renda)", "c (ln Q_F)"], b, se):
        print(f"  {nom:12s} = {bi:7.3f} ± {si:.3f}   (t = {bi/si:5.2f})")

fig, ax = plt.subplots(1, 2, figsize=(13, 5))
ax[0].scatter(q["qf_no_residencial"], q["chi_pct"], s=20)
for cp_, r in q.iterrows():
    ax[0].annotate(cp_[-2:], (r["qf_no_residencial"], r["chi_pct"]), fontsize=7, xytext=(3, 3), textcoords="offset points")
ax[0].set_xscale("log"); ax[0].set_xlabel("Q_F no residencial (W/m²)"); ax[0].set_ylabel("χ⁰ (% / K)")
ax[0].set_title(f"Susceptibilitat residencial vs calor dels no residencials (ρ = {rho:.2f})")
g2 = pol.merge(qf.reset_index().rename(columns={"index": "cp"}), on="cp").to_crs(epsg=3857)
g2.plot(ax=ax[1], column="qf_no_residencial", cmap="magma_r", legend=True, edgecolor="white", linewidth=0.5,
        norm=plt.matplotlib.colors.LogNorm(), legend_kwds={"label": "Q_F no residencial (W/m²)", "shrink": 0.7})
ax[1].set_axis_off(); ax[1].set_title("Calor elèctrica alliberada pels edificis no residencials")
plt.tight_layout(); plt.savefig("qf_vs_chi_v12.png", dpi=200); plt.show()
```

## A5. Comparació v0.3 → v0.4: què explica la susceptibilitat?

**Què fa:** junta en una sola taula, per a cada CP, la susceptibilitat $\chi^0$ (estacions, com la v0.3), la illa de calor del satèl·lit, la renda (secció A3) i la calor no residencial (secció A4). Després respon tres preguntes:

1. **Canvia el mapa si corregim per amplificació?** $\chi^0_{\text{corregida}} = \chi^0/s_i$. Si l'ordre dels CP es manté (Spearman alt), la geografia de $\chi^0$ no és illa de calor. Com que la superfície exagera, és la correcció **màxima** possible.
2. **Què pesa més: la renda, la illa de calor o la calor dels no residencials?** Regressió amb les variables estandarditzades: cada coeficient diu quantes desviacions típiques canvia $\chi^0$ per cada desviació típica de la variable, i així es poden comparar. Mira el $t$: per sota de 2 en valor absolut, no és concloent.
3. **Hipòtesi multiplicativa:** $\chi^0 \sim$ renda $\times$ escalfament. Si $\chi^0$ correlaciona més amb el producte que amb la renda sola, la hipòtesi es reforça.

**Compte:** tenim ~40 CP, i les explicatives estan correlacionades entre elles (els barris densos són més calents i tenen més comerç). La taula de correlacions avisa si els coeficients es poden confondre. És una anàlisi exploratòria, no causal.

```python
# Taula conjunta per CP: susceptibilitat (v0.3, estacions), illa de calor (Landsat), renda i calor no residencial
T = mapa.set_index("cp")[["chi_pct", "chi_estatica_se", "signif"]].join(illa)
if "renda_cp" in globals():
    T = T.join(renda_cp.rename("renda"))
if "qf" in globals():
    T = T.join(np.log(qf["qf_no_residencial"]).rename("ln_qf"))
if "taula" in globals() and "barris_principals" in taula.columns:
    T = T.join(taula.set_index("cp")["barris_principals"].str.split(" \\(").str[0].rename("barri"))

# ---------- 1) Correcció per amplificació tèrmica ----------
# Si la superfície d'un CP s'escalfa s vegades més que la mitjana, una part de la seva χ⁰ és calor extra, no resposta.
# χ⁰_corregida = χ⁰ / s. Compte: la superfície oscil·la més que l'aire, és una correcció MÀXIMA.
T["chi_corregida"] = T["chi_pct"] / T["s_amplificacio"]
Ts = T[T["signif"]].dropna(subset=["chi_corregida"])
rho, _ = stats.spearmanr(Ts["chi_pct"], Ts["chi_corregida"])
print(f"Ordre dels CP abans i després de corregir per amplificació: Spearman ρ = {rho:.2f}")
print(f"Amplificació s: de {T['s_amplificacio'].min():.2f} a {T['s_amplificacio'].max():.2f} (1 = com la mitjana)")
rho_i, p_i = stats.spearmanr(Ts["dLST"], Ts["chi_pct"])
print(f"χ⁰ vs illa de calor ΔLST: ρ = {rho_i:.2f} (p = {p_i:.3f})")

# ---------- 2) Què explica χ⁰? Regressió amb variables estandarditzades ----------
# Coeficients en "desviacions típiques de χ⁰ per desviació típica de la variable": així es poden comparar
cols = [c for c in ["renda", "dLST", "ln_qf"] if c in Ts.columns and Ts[c].notna().sum() > 10]
D = Ts.dropna(subset=cols)
Z = (D[cols] - D[cols].mean()) / D[cols].std()
y = (D["chi_pct"] - D["chi_pct"].mean()) / D["chi_pct"].std()
X = np.column_stack([np.ones(len(D)), Z.values])
b, *_ = np.linalg.lstsq(X, y.values, rcond=None)
res = y.values - X @ b
se = np.sqrt(np.diag(res @ res / (len(D) - X.shape[1]) * np.linalg.inv(X.T @ X)))
print(f"\nRegressió estandarditzada χ⁰ ~ {' + '.join(cols)}   ({len(D)} CP, R² = {1 - res.var() / y.var():.2f})")
for nom, bi, si in zip(["constant"] + cols, b, se):
    print(f"  {nom:9s} {bi:+.2f} ± {si:.2f}   (t = {bi/si:+.1f})")
print("Correlacions entre explicatives (si són altes, els coeficients es confonen):")
print(D[cols].corr().round(2).to_string())

# ---------- 3) Hipòtesi multiplicativa: χ⁰ ∝ renda × escalfament ----------
if "renda" in cols:
    D = D.assign(producte=(D["renda"] / D["renda"].mean()) * D["s_amplificacio"])
    r1, _ = stats.spearmanr(D["renda"], D["chi_pct"]); r2, _ = stats.spearmanr(D["producte"], D["chi_pct"])
    print(f"\nSpearman χ⁰ vs renda: {r1:.2f} · χ⁰ vs renda × amplificació: {r2:.2f}")

# ---------- Figures ----------
fig, ax = plt.subplots(1, 2, figsize=(13, 5))
ax[0].scatter(Ts["chi_pct"], Ts["chi_corregida"], s=20); m = Ts[["chi_pct", "chi_corregida"]].max().max() * 1.05
ax[0].plot([0, m], [0, m], "k--", lw=0.8)
for cp_, r in Ts.iterrows():
    ax[0].annotate(cp_[-2:], (r["chi_pct"], r["chi_corregida"]), fontsize=7)
ax[0].set_xlabel("χ⁰ v0.3 (%/K)"); ax[0].set_ylabel("χ⁰ corregida per amplificació (%/K)")
ax[0].set_title(f"Correcció per illa de calor (ρ = {rho:.2f})")
ax[1].scatter(Ts["dLST"], Ts["chi_pct"], s=20)
for cp_, r in Ts.iterrows():
    ax[1].annotate(cp_[-2:], (r["dLST"], r["chi_pct"]), fontsize=7)
ax[1].set_xlabel("ΔLST: illa de calor de superfície (°C)"); ax[1].set_ylabel("χ⁰ (%/K)")
ax[1].set_title(f"Susceptibilitat vs illa de calor (ρ = {rho_i:.2f})")
plt.tight_layout(); plt.savefig("comparacio_v03_v12.png", dpi=200); plt.show()

T.round(3).to_csv("taula_conjunta_v12.csv")
print("\n[OK] Guardat: taula_conjunta_v12.csv")
print(T.sort_values("chi_pct", ascending=False).round(2).to_string())
```

## A6. Illa de calor dins del model: temperatura ERA5 corregida i llindar $T_c$ per CP

**Per què:** a la secció A5 comparem $\chi^0$ amb la illa de calor *després* d'ajustar. Aquí la posem *dins* de l'ajust, de dues maneres.

**18a. Temperatura ERA5 corregida per Landsat.** Per a cada CP construïm una temperatura pròpia a partir d'ERA5:
$$T_i(d) = \bar T + s_i\,\big(T_{ERA5}(d) - \bar T\big) + k\,\Delta LST_i$$
on $\bar T$ és la mitjana d'estiu d'ERA5, $s_i$ l'amplificació i $\Delta LST_i$ la illa de calor del CP (secció 10). $k$ diu quina fracció del contrast de superfície passa a l'aire: la superfície exagera, així que $k<1$ (no en tenim cap valor verificat; provem $k = 1$ i $k = 0{,}5$). Tornem a ajustar $\chi^0$ amb el $T_c$ de la ciutat i el comparem amb el $\chi^0$ d'ERA5.

**Què esperar:** sumar una constant ($k\,\Delta LST_i$) a la temperatura **no canvia la pendent**, només mou el llindar: $\max(T + \Delta - T_c, 0) = \max(T - (T_c - \Delta), 0)$. Per tant, $\chi^0$ només canviarà per l'amplificació $s_i$ i per uns quants dies que entren o surten de la zona calenta. Si l'ordre dels CP es manté (Spearman alt), es confirma que la illa de calor no és el que fa el mapa.

**18b. El test nou: llindar $T_c$ propi per a cada CP.** Si la illa de calor importa, un barri calent hauria de començar a refrigerar **abans**, mesurat amb el termòmetre comú (ERA5):
$$T_{c,i}^{ERA5} \approx T_{c,0} - k\,\Delta LST_i \quad\Rightarrow\quad \text{pendent negativa de } T_{c,i} \text{ contra } \Delta LST_i$$
Per a cada CP provem $T_c$ de 16 a 25 °C (cada 0,25 °C) i ens quedem amb el que fa mínim l'error. L'interval de confiança aproximat (95 %) són els $T_c$ amb error $\le$ mínim $\times\, e^{3{,}84/n}$ ($n$ = dies), la regla de la versemblança perfilada amb errors gaussians.

**Què mirar a la sortida:**
- La taula de $T_{c,i}$ amb el seu interval. Si molts CP tenen intervals de més de 2-3 °C, el llindar per CP és massa sorollós i el test no té potència.
- Els CP amb $T_c$ a la vora de la graella (16 o 25 °C) no són fiables: el mínim no s'ha trobat.
- **La pendent de $T_{c,i}$ contra $\Delta LST_i$**: negativa i amb $|t| > 2$ vol dir que la illa de calor mou el **llindar**, encara que no mogui la **susceptibilitat**. Seria un resultat nou.
- També $T_{c,i}$ contra la renda: un llindar més baix als barris rics seria coherent amb més aire condicionat.

**Compte:** els errors estàndard són optimistes (residus autocorrelacionats) i és una anàlisi exploratòria. No ho posis a la presentació fins que no ho hàgim mirat junts.


```python
# ============ A6. Illa de calor DINS del model ============
# Necessita: temp (ERA5, amb la columna 'onada'), diari (consum per CP), illa (secció 10),
#            res_era5 (secció 17), TC (secció 15) i, si hi és, renda_cp (secció A3).
from scipy import stats

estiu = temp["data"].dt.month.isin(MESOS_ESTIU) & temp["data"].dt.year.isin(ANYS)
T_MITJA = temp.loc[estiu, "tmean"].mean()      # mitjana d'estiu d'ERA5 (punt fix de l'amplificació)

def temp_corregida(cp, k=1.0):
    # Temperatura pròpia del CP: ERA5 escalada per l'amplificació s i desplaçada per k·ΔLST
    t = temp.copy()
    s, d = illa.loc[cp, "s_amplificacio"], illa.loc[cp, "dLST"]
    if not (np.isfinite(s) and np.isfinite(d)):
        return None
    t["tmean"] = T_MITJA + s * (t["tmean"] - T_MITJA) + k * d
    return t

def chi0_cp(serie_cp, Tc, tsrc=None):
    # Ajusta un CP i retorna (χ⁰, error, error quadràtic mitjà, nombre de dies)
    taula = prepara(serie_cp[["data", "kwh"]], Tc, tsrc)
    normals = taula[~taula["onada"]]
    if len(normals) < 100:
        return np.nan, np.nan, np.nan, 0
    beta, se, res = ajusta(normals, LAGS)
    chi = sum(beta[f"cdd{j}"] for j in range(LAGS + 1))
    chi_se = np.sqrt(sum(se[f"cdd{j}"] ** 2 for j in range(LAGS + 1)))
    return chi, chi_se, np.mean(res ** 2), len(normals)

# ---------- 18a. χ⁰ amb ERA5 corregida per Landsat ----------
files = []
for cp, serie_cp in diari.groupby("cp"):
    if cp not in illa.index:
        continue
    fila = {"cp": cp}
    for k in [1.0, 0.5]:
        tsrc = temp_corregida(cp, k)
        if tsrc is not None:
            fila[f"chi_k{k}"] = chi0_cp(serie_cp, TC, tsrc)[0]
    files.append(fila)
corr = pd.DataFrame(files).merge(res_era5[["cp", "chi_estatica"]], on="cp").dropna()

print(f"18a · χ⁰ amb ERA5 (Tc = {TC} °C) vs χ⁰ amb ERA5 corregida per Landsat ({len(corr)} CP)")
for k in [1.0, 0.5]:
    rho, _ = stats.spearmanr(corr["chi_estatica"], corr[f"chi_k{k}"])
    canvi = 100 * (corr[f"chi_k{k}"] - corr["chi_estatica"])
    print(f"  k = {k}: Spearman ρ = {rho:.2f} · canvi de χ⁰ de {canvi.min():+.1f} a {canvi.max():+.1f} %/K "
          f"(mediana {canvi.median():+.1f})")

# ---------- 18b. Llindar Tc propi per a cada CP (amb ERA5, sense corregir) ----------
TC_GRAELLA = np.arange(16.0, 25.01, 0.25)
files = []
for cp, serie_cp in diari.groupby("cp"):
    perfil = [chi0_cp(serie_cp, Tc) for Tc in TC_GRAELLA]
    mse = np.array([p[2] for p in perfil]); n = perfil[0][3]
    if n == 0 or np.all(np.isnan(mse)):
        continue
    i = np.nanargmin(mse)
    dins = TC_GRAELLA[mse <= mse[i] * np.exp(3.84 / n)]        # interval de confiança aprox. del 95 %
    files.append({"cp": cp, "Tc_cp": TC_GRAELLA[i], "Tc_baix": dins.min(), "Tc_alt": dins.max(),
                  "chi_Tc_propi": perfil[i][0], "vora": i in (0, len(TC_GRAELLA) - 1), "n_dies": n})
llindars = pd.DataFrame(files).set_index("cp").join(illa[["dLST", "s_amplificacio"]])
if "renda_cp" in globals():
    llindars = llindars.join(renda_cp.rename("renda"))
llindars["amplada_IC"] = llindars["Tc_alt"] - llindars["Tc_baix"]

print(f"\n18b · Tc per CP amb ERA5 (Tc de la ciutat = {TC} °C)")
print(f"  Mediana {llindars['Tc_cp'].median():.2f} °C · de {llindars['Tc_cp'].min():.2f} a {llindars['Tc_cp'].max():.2f} °C")
print(f"  Amplada mediana de l'interval del 95 %: {llindars['amplada_IC'].median():.2f} °C")
print(f"  CP amb el mínim a la vora de la graella (no fiables): {list(llindars.index[llindars['vora']])}")

# Test: es mou el llindar amb la illa de calor? (només CP fiables)
ok = llindars[~llindars["vora"]].dropna(subset=["dLST"])
reg = stats.linregress(ok["dLST"], ok["Tc_cp"])
rho, p = stats.spearmanr(ok["dLST"], ok["Tc_cp"])
print(f"\n  Tc_cp vs ΔLST ({len(ok)} CP): pendent {reg.slope:+.3f} ± {reg.stderr:.3f} °C/°C "
      f"(t = {reg.slope / reg.stderr:+.1f}) · Spearman ρ = {rho:.2f} (p = {p:.3f})")
print("  Pendent negativa i |t| > 2 → la illa de calor avança el llindar.")
if "renda" in ok.columns:
    okr = ok.dropna(subset=["renda"])
    rho_r, p_r = stats.spearmanr(okr["renda"], okr["Tc_cp"])
    print(f"  Tc_cp vs renda ({len(okr)} CP): Spearman ρ = {rho_r:.2f} (p = {p_r:.3f})")

# ---------- Figures ----------
fig, ax = plt.subplots(1, 2, figsize=(13, 5))
for k, c in [(1.0, "C0"), (0.5, "C1")]:
    ax[0].scatter(100 * corr["chi_estatica"], 100 * corr[f"chi_k{k}"], s=18, c=c, label=f"k = {k}")
m = 100 * corr[["chi_estatica", "chi_k1.0", "chi_k0.5"]].max().max() * 1.05
ax[0].plot([0, m], [0, m], "k--", lw=0.8); ax[0].legend()
ax[0].set_xlabel("χ⁰ amb ERA5 (%/K)"); ax[0].set_ylabel("χ⁰ amb ERA5 corregida per Landsat (%/K)")
ax[0].set_title("18a · Corregir la temperatura amb la illa de calor")

ax[1].errorbar(ok["dLST"], ok["Tc_cp"], yerr=[ok["Tc_cp"] - ok["Tc_baix"], ok["Tc_alt"] - ok["Tc_cp"]],
               fmt="o", ms=4, capsize=2, elinewidth=0.7, alpha=0.8)
x = np.linspace(ok["dLST"].min(), ok["dLST"].max(), 50)
ax[1].plot(x, reg.intercept + reg.slope * x, "r--", lw=1, label=f"pendent {reg.slope:+.2f} (t = {reg.slope / reg.stderr:+.1f})")
for cp_, r in ok.iterrows():
    ax[1].annotate(cp_[-2:], (r["dLST"], r["Tc_cp"]), fontsize=7, xytext=(3, 3), textcoords="offset points")
ax[1].set_xlabel("ΔLST: illa de calor de superfície (°C)"); ax[1].set_ylabel("Tc del CP amb ERA5 (°C)")
ax[1].set_title("18b · Es mou el llindar amb la illa de calor?"); ax[1].legend()
plt.tight_layout(); plt.savefig("illa_dins_model_v12.png", dpi=200); plt.show()

llindars.round(3).to_csv("llindars_per_cp_v12.csv")
print("\n[OK] Guardat: llindars_per_cp_v12.csv")
print(llindars.sort_values("Tc_cp")[["Tc_cp", "Tc_baix", "Tc_alt", "dLST", "chi_Tc_propi"]].round(2).to_string())
```
